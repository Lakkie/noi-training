# 1 Logical Structure 逻辑结构

## 1.1 Set Structure 集合结构‌

定义‌：数据元素之间除了“同属一个集合”的关联外，不存在其他任何特定逻辑关系，元素组织形式松散。

```python
# 1. 集合的创建

# 方法一：使用花括号 {}
# 注意：空的花括号 {} 创建的是字典，不是集合
fruits = {"apple", "banana", "cherry"}
print(fruits)  # 输出: {'apple', 'banana', 'cherry'} (顺序可能不同)

# 方法二：使用 set() 构造函数
# 适用于创建空集合或将其他可迭代对象转换为集合
empty_set = set()
print(empty_set)  # 输出: set()

# 从列表转换（自动去重）
numbers = set([1, 2, 2, 3, 4, 4, 5])
print(numbers)  # 输出: {1, 2, 3, 4, 5}

# 从字符串转换
chars = set("hello")
print(chars)  # 输出: {'h', 'e', 'l', 'o'}
```

```python
# 2. 基本操作：增删改查 集合不支持索引访问（因为无序），但支持高效的成员检测和元素增减。

s = {1, 2, 3}

# --- 添加元素 ---
s.add(4)          # 添加单个元素
s.update([5, 6])  # 添加多个元素（传入列表、元组或另一个集合）
print(s)          # 输出: {1, 2, 3, 4, 5, 6}

# --- 删除元素 ---
s.remove(2)       # 删除指定元素，若元素不存在会报错 (KeyError)
s.discard(10)     # 删除指定元素，若元素不存在不会报错（推荐用于不确定元素是否存在时）
popped = s.pop()  # 随机移除并返回一个元素（因为无序，无法指定移除哪个）
s.clear()         # 清空集合

# --- 成员检测 ---
# 性能远优于列表，时间复杂度为 O(1)
my_set = {1, 2, 3, 4, 5}
print(3 in my_set)    # 输出: True
print(10 not in my_set) # 输出: True
```

```python
# 3. 集合数学运算

A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

# 并集 (Union): 包含两个集合所有不重复元素
print(A | B)            # 运算符方式: {1, 2, 3, 4, 5, 6, 7, 8}
print(A.union(B))       # 方法方式

# 交集 (Intersection): 两个集合共有的元素
print(A & B)            # 运算符方式: {4, 5}
print(A.intersection(B)) # 方法方式

# 差集 (Difference): 在 A 中但不在 B 中的元素
print(A - B)            # 运算符方式: {1, 2, 3}
print(A.difference(B))   # 方法方式

# 对称差集 (Symmetric Difference): 只在 A 或只在 B 中的元素（排除共有部分）
print(A ^ B)            # 运算符方式: {1, 2, 3, 6, 7, 8}
print(A.symmetric_difference(B)) # 方法方式
```

```python
# 4. 集合推导式

# 生成 0-9 中偶数的平方集合
squares = {x**2 for x in range(10) if x % 2 == 0}
print(squares)  # 输出: {0, 4, 16, 36, 64}
```

## 1.2 Linear Structure 线性结构‌

定义‌：数据元素之间存在‌一对一‌的线性相邻关系，有且仅有一个开始结点和终端结点，所有结点最多仅有一个直接前驱和一个直接后继。

### 1.2.1 General Linear List 一般线性表‌

定义‌：一般线性表是具有相同数据类型的n（n≥0）个数据元素的有限序列，除首元素无直接前驱、尾元素无直接后继外，其余每个元素有且仅有一个直接前驱和一个直接后继，且支持在表中‌任意位置‌执行插入、删除、查找、遍历等操作，没有操作位置的特殊约束。

#### 1.2.1.1 Sequential List 顺序表

定义:顺序表是采用‌顺序存储结构‌实现的线性表，它将线性表中所有元素依次存放在一组地址连续的内存单元中，借助物理上的邻接关系直接体现数据元素之间的一对一逻辑相邻关系。

#### 1.2.1.2 Singly Linked List 单向链表

定义：单向链表是用‌任意地址的内存单元‌存储线性表元素的链式结构，每个结点由数据域和唯一的指针域组成，数据域存放元素值，指针域指向序列中的下一个结点，所有结点通过指针串联成一条单向的链式序列。

#### 1.2.1.3 Doubly Linked List 双向链表

定义：双向链表的每个结点都包含‌数据域、前驱指针域、后继指针域‌三个部分，数据域存放元素值，前驱指针指向该结点的前一个结点，后继指针指向该结点的后一个结点，所有结点通过双向指针串联成可双向遍历的链式序列。

#### 1.2.1.4 Circular Linked List 循环链表

定义：循环链表是将普通链表的首尾结点通过指针相连形成的环形链式结构，最常见的实现是单向循环链表，它把最后一个结点的后继指针指向链表的头结点，让整个链表形成一个首尾相接的闭合环形。

### 1.2.2 Restricted Linear List 受限线性表‌

‌定义：受限线性表是对基础线性表的操作位置做了明确约束的线性结构，它保留了线性表“元素为一对一的有限序列”的基础逻辑特征，但仅允许在‌指定的一个或两个端点‌完成元素的插入、删除操作，不支持在表中任意位置执行增删操作。

#### 1.2.2.1 Stack 栈

定义：栈是限定仅能在表的同一端进行插入和删除操作的受限线性表，允许操作的一端被称为‌栈顶‌，另一端被称为‌栈底‌，元素的访问遵循‌后进先出（LIFO, Last In First Out）‌的核心规则。

#### 1.2.2.2 Queue 队列

定义：队列是限定仅能在表的不同两端分别进行插入和删除操作的受限线性表，允许插入元素的一端被称为‌队尾‌，允许删除元素的一端被称为‌队头‌，元素的访问遵循‌先进先出（FIFO, First In First Out）‌的核心规则。

#### 1.2.2.3 Deque (Double-Ended Queue) 双端队列

定义：双端队列（Deque，全称为Double-Ended Queue）是允许在队列的‌两端‌都执行插入和删除操作的受限线性表，它没有普通队列“仅队尾插入、仅队头删除”的严格约束，两端都可作为队头或队尾完成元素的增删操作。

#### 1.2.2.4 String 串

定义：串在计算机数据结构中通常指字符串，是由零个或多个字符组成的有限序列，是一种特殊的线性表，它的数据元素被限定为字符集中的字符。

### 1.2.3 Extended Linear List 扩展线性表‌（推广线性表）

定义：扩展线性表是对基础线性表的元素组成规则进行扩展的线性结构，它保留了线性表的部分基础逻辑特征，同时突破了“元素必须为原子类型”的约束，允许元素为子表或多维结构，以此实现更复杂的数据关系表达。

#### 1.2.3.1 Array 数组

定义：数组是用一组‌连续的内存空间‌，存储一组‌相同数据类型元素‌的线性结构，每个元素都有唯一的整数索引（通常从0开始计数），可通过索引直接定位元素，实现O(1)时间复杂度的随机访问。

#### 1.2.3.2 Generalized List 广义表

定义：广义表是n（n≥0）个元素的有限序列，其中每个元素既可以是不可拆分的原子数据元素，也可以是另一个广义表（子表），它突破了基础线性表元素必须为原子类型的约束，是线性表的深度扩展形式。

## 1.3 Tree Structure 树结构‌

定义‌：数据元素之间存在‌一对多‌的层次关系，有且仅有一个根结点，其余结点有且仅有一个直接前驱，可拥有多个直接后继。

根节点(root): 树的最上层的节点，任何非空的树都有一个节点  
路径(path): 从起始节点到终止节点经历过的边  
父亲(parent)：除了根节点，每个节点的上一层边连接的节点就是它的父亲(节点)  
孩子(children): 每个节点由边指向的下一层节点  
兄弟(siblings): 同一个父亲并且处在同一层的节点  
子树(subtree): 每个节点包含它所有的后代组成的子树  
叶子节点(leaf node): 没有孩子的节点成为叶子节点  

```python
# -*- coding: utf-8 -*-

class BinTreeNode(object):
    """
    二叉树节点类
    每个节点包含数据域(data)和两个指针域(left, right)
    """
    def __init__(self, data, left=None, right=None):
        self.data = data      # 节点存储的数据
        self.left = left      # 指向左子节点的指针
        self.right = right    # 指向右子节点的指针


class BinTree(object):
    """
    二叉树类
    支持从列表构建、前序遍历和镜像反转操作
    """
    def __init__(self, root=None):
        self.root = root      # 树的根节点

    @classmethod
    def build_from(cls, node_list):
        """
        根据节点列表构建二叉树
        
        :param node_list: 字典列表，每个字典包含 'data', 'left', 'right', 'is_root'
                          'left' 和 'right' 的值对应其他节点的 'data' 值，若无则为 None
        :return: 构建好的 BinTree 实例
        """
        node_dict = {}
        
        # 第一遍遍历：创建所有节点对象，并存入字典以便通过 data 快速查找
        for node_data in node_list:
            data = node_data['data']
            # 此时左右子节点暂为 None，后续第二步再连接
            node_dict[data] = BinTreeNode(data)
        
        root = None
        # 第二遍遍历：建立节点之间的父子关系（连接左右指针）
        for node_data in node_list:
            data = node_data['data']
            node = node_dict[data]
            
            # 识别根节点
            if node_data['is_root']:
                root = node
            
            # 获取左子节点对象：如果 node_data['left'] 为 None，get 返回 None；否则从字典取对象
            node.left = node_dict.get(node_data['left'])
            # 获取右子节点对象
            node.right = node_dict.get(node_data['right'])
            
        return cls(root)

    def preorder_trav(self, subtree):
        """
        前序遍历 (Pre-order Traversal)
        顺序：根节点 -> 左子树 -> 右子树
        
        :param subtree: 当前子树的根节点
        """
        if subtree is not None:
            # 1. 访问根节点
            print(subtree.data)
            # 2. 递归遍历左子树
            self.preorder_trav(subtree.left)
            # 3. 递归遍历右子树
            self.preorder_trav(subtree.right)

    def reverse(self, subtree):
        """
        二叉树镜像反转 (Mirror/Reverse)
        核心逻辑：递归交换每个节点的左右子指针
        
        :param subtree: 当前需要反转的子树根节点
        """
        if subtree is not None:
            # 1. 交换当前节点的左右子树
            # Python 特有的元组解包赋值，无需临时变量
            subtree.left, subtree.right = subtree.right, subtree.left
            
            # 2. 递归处理新的左子树（原右子树）
            self.reverse(subtree.left)
            # 3. 递归处理新的右子树（原左子树）
            self.reverse(subtree.right)


# 定义二叉树的结构数据
# 注意：'left' 和 'right' 的值必须与某个节点的 'data' 一致，或者为 None
node_list = [
    {'data': 'A', 'left': 'B', 'right': 'C', 'is_root': True},
    {'data': 'B', 'left': 'D', 'right': 'E', 'is_root': False},
    {'data': 'D', 'left': None, 'right': None, 'is_root': False},
    {'data': 'E', 'left': 'H', 'right': None, 'is_root': False},
    {'data': 'H', 'left': None, 'right': None, 'is_root': False},
    {'data': 'C', 'left': 'F', 'right': 'G', 'is_root': False},
    {'data': 'F', 'left': None, 'right': None, 'is_root': False},
    {'data': 'G', 'left': 'I', 'right': 'J', 'is_root': False},
    {'data': 'I', 'left': None, 'right': None, 'is_root': False},
    {'data': 'J', 'left': None, 'right': None, 'is_root': False},
]

# 1. 构建二叉树
btree = BinTree.build_from(node_list)

# 2. 打印原始树的前序遍历结果
print('原始树前序遍历:')
btree.preorder_trav(btree.root)

# 3. 执行镜像反转
# 反转后，所有节点的左右子树互换
btree.reverse(btree.root)

print('====我是华丽丽滴分割线=====')

# 4. 打印反转后树的前序遍历结果
print('反转后树前序遍历:')
btree.preorder_trav(btree.root)
```

## 1.4 Graph Structure 图结构‌

定义‌：数据元素之间存在‌多对多‌的任意关联关系，任意两个结点之间都可能存在连接路径。

### 广度优先搜索 BFS

```python
# -*- coding: utf-8 -*-

# 导入双端队列，用于高效实现队列的头部弹出和尾部追加操作
from collections import deque


# 定义图的邻接表
# 键(Key)代表节点，值(Value)是该节点直接相连的邻居节点列表
GRAPH = {
    'A': ['B', 'F'],
    'B': ['C', 'I', 'G'],
    'C': ['B', 'I', 'D'],
    'D': ['C', 'I', 'G', 'H', 'E'],
    'E': ['D', 'H', 'F'],
    'F': ['A', 'G', 'E'],
    'G': ['B', 'F', 'H', 'D'],
    'H': ['G', 'D', 'E'],
    'I': ['B', 'C', 'D'],
}


class Queue(object):
    """
    基于 collections.deque 封装的简单队列类
    遵循先进先出 (FIFO) 原则
    """
    def __init__(self):
        # 初始化内部使用的双端队列
        self._deque = deque()

    def push(self, value):
        """
        入队操作：将元素添加到队列尾部
        :param value: 要加入队列的元素
        """
        return self._deque.append(value)

    def pop(self):
        """
        出队操作：从队列头部移除并返回元素
        :return: 队列头部的元素
        """
        return self._deque.popleft()

    def __len__(self):
        """
        返回队列当前长度，支持 len(queue) 调用
        """
        return len(self._deque)


def bfs(graph, start):
    """
    广度优先搜索 (Breadth-First Search) 生成器函数
    
    :param graph: 图的邻接表字典
    :param start: 起始搜索节点
    :yield: 按 BFS 顺序遍历到的节点
    """
    # 初始化搜索队列，并将起始节点入队
    search_queue = Queue()
    search_queue.push(start)
    
    # 使用集合记录已访问过的节点，防止重复访问导致死循环
    searched = set()
    
    # 当队列不为空时，持续进行遍历
    while search_queue:
        # 从队列头部取出一个节点作为当前处理节点
        cur_node = search_queue.pop()
        
        # 如果当前节点尚未被访问过
        if cur_node not in searched:
            yield cur_node               # 1. 产出当前节点供外部迭代使用
            searched.add(cur_node)       # 2. 立即标记为已访问，避免后续重复处理
            
            # 遍历当前节点的所有邻居节点
            for node in graph[cur_node]:
                # 将邻居节点加入队列尾部，等待后续处理
                # 注意：这里不立即标记 visited，而是等到 pop 出来时再判断，
                # 这样允许同一节点多次入队，但只会被处理一次
                search_queue.push(node)  # 3. 将邻居加入队列

# --- 主程序执行部分 ---

print('bfs:')

# 调用 bfs 生成器，并将其转换为列表以获取完整的遍历顺序
# list() 会不断调用 next() 直到生成器耗尽，从而触发 bfs 中的 while 循环
result = list(bfs(GRAPH, 'A'))

# 打印最终的 BFS 遍历结果
print(result)
```

### 深度优先搜索 DFS

```python
# -*- coding: utf-8 -*-

# 定义图的邻接表
GRAPH = {
    'A': ['B', 'F'],
    'B': ['C', 'I', 'G'],
    'C': ['B', 'I', 'D'],
    'D': ['C', 'I', 'G', 'H', 'E'],
    'E': ['D', 'H', 'F'],
    'F': ['A', 'G', 'E'],
    'G': ['B', 'F', 'H', 'D'],
    'H': ['G', 'D', 'E'],
    'I': ['B', 'C', 'D'],
}

# 全局集合，用于存储已访问过的节点
# 使用全局变量是为了在递归调用之间共享状态，避免重复访问
DFS_SEARCHED = set()


def dfs(graph, start):
    """
    深度优先搜索 (Depth-First Search) 递归实现
    
    :param graph: 图的邻接表字典
    :param start: 当前正在访问的起始节点
    """
    # 1. 检查当前节点是否已被访问
    if start not in DFS_SEARCHED:
        # 2. 如果未访问，则处理当前节点（这里表现为打印）
        print(start, end=' ')  # 使用 end=' ' 使输出在同一行，更清晰
        
        # 3. 将当前节点标记为“已访问”，防止后续递归再次进入形成死循环
        DFS_SEARCHED.add(start)
        
        # 4. 递归遍历当前节点的所有邻居
        for node in graph[start]:
            # 只有当邻居节点未被访问时，才深入递归
            # 这体现了 DFS "一条路走到黑" 的特性：优先深入第一个邻居的子树
            if node not in DFS_SEARCHED:
                dfs(graph, node)


# --- 主程序执行部分 ---
print('dfs:')
# 重置全局集合，确保每次运行都是干净的状态（如果是多次调用需要此步）
DFS_SEARCHED.clear() 
dfs(GRAPH, 'A')
print()  # 换行
```

# 2 Storage Structure / Physical Structure 存储结构 / 物理结构

## 2.1 Sequential Storage Structure 顺序存储结构‌

定义‌：将逻辑上相邻的元素，存放在物理地址也连续的存储单元中，元素间的逻辑关系通过存储单元的邻接关系来体现，典型实现为数组。

核心特点‌：存储密度大、支持随机存取，但插入和删除元素时需要移动大量元素，操作效率较低。

## 2.2 Linked Storage Structure 链式存储结构‌

‌定义‌：不要求数据元素在物理地址上连续，每个元素结点包含数据域和指针域，通过指针记录下一个结点的地址，以此串联起所有元素，典型实现为链表。

‌核心特点‌：插入和删除操作仅需修改指针，无需移动元素，但不支持随机存取，查找元素需要沿指针依次遍历。

## 2.3 Indexed Storage Structure 索引存储结构‌

定义‌：在存储数据元素的同时，额外建立一张索引表，索引表的每一项都包含关键字和对应元素的存储地址，通过索引可以快速定位目标元素。

‌核心特点‌：元素查找效率高，但需要额外的空间存储索引表，会增加一定的空间开销。

## 2.4 Hash Storage Structure 哈希存储结构（散列存储结构）

‌定义‌：不依赖元素的物理位置或指针关联，直接根据元素的关键字，通过预先设定的散列函数计算出该元素的存储地址，将元素存入对应位置。

‌核心特点‌：元素的查找、插入和删除操作速度极快，但如果散列函数设计不合理，容易出现地址冲突，需要额外机制处理冲突。
