# Divide and Conquer 分治

是将待求解的原问题划分成k个较小规模的子问题，对这k个子问题分别求解。如果子问题的规模仍然不够小，则再将每个子问题划分为k个规模更小的子问题，如此分解下去，直到问题规模足够小，很容易求出其解为止（子问题求解思路一致），再将子问题的解合并为一个更大规模的问题的解，自底向上逐步求出原问题的解。

# Dynamic Programing 动态规划

是将待求解问题分解成若干个相互重叠的子问题，每个子问题对应决策过程的一个阶段，一般来说，子问题的重叠关系表现在对给定问题求解的递推关系（也就是动态规划函数）中，将子问题的解求解一次并填入表中，当需要再次求解此子问题时，可以通过查表获得该子问题的解而不用再次求解，从而避免了大量重复计算。

# Greedy 贪心

贪心法在解决问题的策略上目光短浅，只根据当前已有的信息就做出选择，而且一旦做出了选择，不管将来有什么结果，这个选择都不会改变。换言之，贪心法并不是从整体最优考虑，它所做出的选择只是在某种意义上的局部最优。这种局部最优选择并不总能获得整体最优解（Optimal Solution），但通常能获得近似最优解（Near-Optimal Solution）。

贪心算法（Greedy Algorithm）是一种在每一步选择中都采取在当前状态下最优的选择，希望通过每一步的局部最优，最终得到全局最优解的方法。通俗地讲，贪心算法总是只看眼前利益，每一步都“贪”当前最好的，不回溯、不修改。

贪婪算法是一种通过每一步选择当前状态下的最优局部决策，逐步构造可行解的算法策略。它采用自顶向下、迭代无回溯的方式——一旦做出选择，就绝不回头。用一句话概括：走一步看一步，选了就不后悔。

使用贪心策略解决问题，通常需要满足以下两个条件：  
1. 贪心选择性质（Greedy Choice Property）当前最优选择可以导出问题的整体最优解。这意味着不需要考虑子问题的解，当前选择不受后续决策影响。  
2. 最优子结构（Optimal Substructure）问题的最优解包含其子问题的最优解。该性质与动态规划算法共享，但贪心算法通过单一选择路径达成，而动态规划需要比较多个子问题解。只有在这两个条件都满足时，贪心算法才有可能得到正确答案。  

贪心算法没有固定的模板，但通常有以下几个步骤：  
模型分解：将问题转换为一系列决策，多个相互关联的阶段  
选择策略：确定每个阶段的局部最优选择标准  
决策实施：做出当前最优选择且不可回退  
迭代求解：将剩余问题转化为新的子问题重复上述过程  

贪婪算法的求解一般遵循以下步骤：  
步骤1：建立数学模型——明确目标函数和约束条件。  
步骤2：划分子问题——把求解的问题分成若干个子问题。  
步骤3：确定贪婪策略——定义每一步的局部最优选择规则。  
步骤4：迭代求解——对每个子问题求解，得到局部最优解。  
步骤5：合并解集——将局部最优解合成为原问题的解。  
这个流程可以简化为一个迭代模型：初始化解集 → 每一步做出当前最优选择 → 更新问题状态 → 直到满足终止条件 → 输出结果。

正确性证明  
不是所有问题都适合用贪心解法，只有满足特定条件的问题，贪心算法才可能得到最优解。  
因此，在实际应用中，我们必须验证：  
贪心策略是否能产生最优解。  
是否满足贪心选择性质和最优子结构这两个核心理论基础。  

数学归纳法：  
证明每个选择步骤保持最优解结构。适合一些具有结构性或递推性质的问题，比如区间覆盖、最小生成树。
步骤：  
证明在初始状态下贪心是最优的；  
假设前 𝑘 步贪心策略都是对的；  
推导第 𝑘+1 步仍然正确。  

反证法：  
这是最常见的贪心算法证明方式。  
核心思想：  
假设存在一个比贪心更优的解，试图将这个更优解一步步“交换”为贪心解，且不会变差，最终证明贪心就是最优的。  
经典例子：区间调度问题（按结束时间排序）  
a. 假设贪心选择是 𝐺，另一个更优的解是 𝑂；  
b. 将 𝑂 中与 𝐺 冲突的部分替换为 𝐺 的选择；  
c. 替换后仍是合法解且不会更差；  
d. 最终能得到与 𝐺 相同甚至更好的结果 ⇒ 说明 𝐺 是最优的。  

构造法 / 局部最优可扩展性  
也可以通过构造过程分析，说明每一步的贪心决策可以扩展到最优解，常用于像哈夫曼编码这样的结构性问题。 

证明一个贪婪算法是否正确，通常有两种方法：  

方法一：交换论证（Exchange Argument） —— 考察一个全局最优解，证明可以将其逐步“交换”成贪婪算法选择的解，且每一步交换都不会让解变差。最终，贪婪解就是一个全局最优解。

方法二：贪婪领先（Greedy Stays Ahead） —— 证明贪婪算法在每一步的某个度量指标上都不落后于任何其他算法。如果贪婪算法始终“领先”或“持平”，最终它必然达到最优。

# Backtracking 回溯

回溯法就是一种有组织的系统化搜索技术，可以看作是蛮力法穷举搜索的改进。回溯法每次只构建可能解的一部分，然后评估这个部分解，如果这个部分有可能导致一个完全解，对其进一步搜索，否则，就不必继续构造这部分的解了，回溯法常常可以避免搜索所有可能的解，所以，它往往比满立法效率更高，适用于求解组合数组较大的问题。

# Branch and Bound 分支限界

分支限界法按广度优先策略遍历问题的解空间，在遍历过程种，对已经处理的每一个结点根据限界函数估算目标函数的可能值，从中选取使目标函数取得极值（极大或极小）的结点优先进行广度优先搜索，从而不断调整搜索方向，尽快找到问题的解。因为界限函数常常是基于问题的目标函数而确定的，所以，分支限界法适用于求解最优化问题。
