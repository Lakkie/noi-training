`````python
# 打印空心三角形 带背景

side = 11
center = side // 2 + 1  # 三角形的行数（高度）

symbol_background = "." 
symbol_graph = "+" 

for i in range(1, center + 1):
    leading = symbol_background * (center - i)
    if i == 1:                                                     # 顶点：只有一个图形符
        line = leading + symbol_graph + leading
    elif i == center:                                              # 底边：全部是图形符，宽度为 2*center - 1
        line = leading + symbol_graph * (2 * i - 1)
    else:
        mid_bg = symbol_background * (2 * i - 3)
        line = leading + symbol_graph + mid_bg + symbol_graph + leading
        
    print(line)
`````

`````python
# 打印实心矩形

row = 10                                       # row 指横着的行
column = 10                                    # ‌column 指竖着的列
symbol_background = "*"                        # 背景地图的符号

for i in range(row):                           # 使用外层循环来处理行
    row_list = []                              # 用于存放每行需要打印的符号

    for j in range(column):                    # 使用内层循环来处理列
        row_list.append(symbol_background)     # 将符号存入列表

    row_str = "".join(row_list)                # 将列表转换为字符串
    print(row_str)                             # 打印当前行
`````

`````python
# 打印矩形（长方形 正方形）

row = 10 
column = 10 
symbol_background = "*" 
symbol_graph = "x"                                      # 图案的符号

for i in range(row):            
    row_list = [] 

    for j in range(column): 
        if i==0 or i==row-1 or j==0 or j==column-1:     # 地图最外围的位置 (判断需要打印图案的位置)
            row_list.append(symbol_graph)               # 如果符合条件，往列表中存入图案的符号
        else: 
            row_list.append(symbol_background)          # 如果不符合条件，则往列表中存入背景地图的符号
    
    row_str = "".join(row_list) 
    print(row_str)
`````

`````python
# 打印矩形（长方形 正方形）

row = 10 
column = 10 
symbol_background = "*" 
symbol_graph1 = "+"                                     # 图案的符号
symbol_graph2 = "-"                                     # 图案的符号
symbol_graph3 = "|"                                     # 图案的符号

for i in range(row):            
    row_list = [] 

    for j in range(column): 
        if (i==0 and j==0) or (i==0 and j==column-1) or (i==column-1 and j==0) or (i==column-1 and j==column-1):     # 地图的四个角
            row_list.append(symbol_graph1) 
        elif (i==0 and j!=0 and j!=column-1) or (i==column-1 and j!=0 and j!=column-1): 
        # elif i==0 or i==column-1: 
            row_list.append(symbol_graph2) 
        elif (j==0 and i!=0 and i!=column-1) or (j==column-1 and i!=0 and j!=column-1): 
        # elif j==0 or j==column-1: 
            row_list.append(symbol_graph3) 
        else: 
            row_list.append(symbol_background) 
    
    row_str = "".join(row_list) 
    print(row_str)
`````

`````python
# 打印实心菱形

side = 11
center = side//2 + 1

symbol_background = "." 
symbol_graph = "+" 

# 上半部分（包括中间行）
for i in range(1, center + 1):
    spaces = symbol_background * (center - i)
    stars = symbol_graph * (2 * i - 1)
    print(spaces + stars + spaces)
    
# 下半部分
for i in range(center - 1, 0, -1):
    spaces = symbol_background * (center - i)
    stars = symbol_graph * (2 * i - 1)
    print(spaces + stars + spaces)
`````

`````python
# 打印空心菱形
side = 11
center = side // 2 + 1

symbol_background = "." 
symbol_graph = "+" 

# 上半部分（包括中间行）
for i in range(1, center + 1):
    leading_spaces = symbol_background * (center - i)
    if i == 1:                                                    # 顶点：只有一个图形符号
        line = leading_spaces + symbol_graph + leading_spaces
    else:
        mid_background = symbol_background * (2 * i - 3)
        line = leading_spaces + symbol_graph + mid_background + symbol_graph + leading_spaces
        
    print(line)
    
# 下半部分
for i in range(center - 1, 0, -1):
    leading_spaces = symbol_background * (center - i)
    if i == 1:                                                    # 底点：只有一个图形符号
        line = leading_spaces + symbol_graph + leading_spaces
    else:
        mid_background = symbol_background * (2 * i - 3)
        line = leading_spaces + symbol_graph + mid_background + symbol_graph + leading_spaces
        
    print(line)
`````

`````python
# 打印空心菱形 高阶公式

side = 11
center = side//2 + 1

symbol_background = "+" 
symbol_graph = "." 

for i in range(side):
    row_list = []

    for j in range(side):                        
        if abs(i + 1 - center) + abs(j + 1 - center) == center - 1: # L1距离公式（高中数学知识）
            row_list.append(symbol_background)
        else:
            row_list.append(symbol_graph)
    
    row_str = "".join(row_list) 
    print(row_str)
`````
