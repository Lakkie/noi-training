```python
h1=int(input())
h2=int(input())
h3=int(input())
h4=int(input())
a=h1
f=[h2,h3,h4]
m=10000
hh=0
for h in f:
    d=abs(h-a)
    if(d<m):
        m=d
        hh=h
    elif(d==m):
        if(h<hh):
            hh=h
print(hh)
```
