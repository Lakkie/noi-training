```python
# P11966 [GESP202503 八级] 上学

import heapq
import sys

sys.setrecursionlimit(200005)

def solve():
    input = sys.stdin.readline
    try:
        line1 = input().split()
        if not line1:
            return
        n, m, s, q = map(int, line1)
    except ValueError:
        return

    # 构建邻接表
    # graph[u] 存储列表 [(v, weight), ...]
    graph = [[] for _ in range(n + 1)]
    for _ in range(m):
        u, v, w = map(int, input().split())
        graph[u].append((v, w))
        graph[v].append((u, w))  # 无向图，双向添加

    # Dijkstra 算法初始化
    INF = float('inf')
    dist = [INF] * (n + 1)
    dist[s] = 0
    # 优先队列：存储 (距离，节点编号)
    # Python 的 heapq 是小顶堆，默认按元组第一个元素排序
    pq = [(0, s)]
    # 标记数组可选，但在 Python 中通常通过判断 dist 是否更小来避免重复处理
    # 这里采用 lazy deletion 策略：如果取出的距离大于当前记录的最短距离，则跳过
    while pq:
        d, u = heapq.heappop(pq)
        # 如果当前取出的距离比已知的最短距离大，说明是旧数据，跳过
        if d > dist[u]:
            continue
        # 遍历邻接点
        for v, w in graph[u]:
            new_dist = dist[u] + w
            if new_dist < dist[v]:
                dist[v] = new_dist
                heapq.heappush(pq, (new_dist, v))

    # 处理查询并输出
    results = []
    for _ in range(q):
        h = int(input())
        # 如果不可达，题目通常保证连通或输出特定值，此处按题意输出计算结果
        # 若 dist[h] 仍为 INF，说明不可达，但根据题意通常有解
        results.append(str(dist[h] if dist[h] != INF else -1))
    
    print('\n'.join(results))

if __name__ == "__main__":
    solve()
```
