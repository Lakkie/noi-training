```python
# B4497 [GESP202603 二级] 数数

# 输入
L = int(input())
R = int(input())

# 计算
s = 0                         # 初始化计数器
for num in range(L, R + 1):       # 遍历区间 [L, R] 中的每一个整数
    if str(num).count('2') == 3:  # 将整数转换为字符串，统计字符 '2' 出现的次数
        s += 1

# 输出
print(s)
```

```python
# B4553 [GESP202606 二级] 完全平方数计数

import math

# 输入
l = int(input())
r = int(input())

# 计算
# 原理：
# sqrt(r) 向下取整得到的是 1 到 r 之间完全平方数的个数
# sqrt(l-1) 向下取整得到的是 1 到 l-1 之间完全平方数的个数
# 两者相减即为 l 到 r 之间的完全平方数个数
count = int(math.isqrt(r)) - int(math.isqrt(l - 1))

# 输出
print(count)
```

```python
# 对于乘方运算规则 a ^ b = N，其中a为底数base，b为指数exponent，N为真数radicand。
import math

# 已知a和b，求N
def compute_n(a, b):
    # n = a ** b
    n = math.pow(a, b)
    return int(n)

# 已知b和N，求a
def compute_a(b, n):
    # a = n ** (1/b)
    a = math.pow(n, 1/b)
    return int(a)

# 已知a和N，求b
def compute_b(a, n):
    b = math.log(n, a)
    return int(b)

# 示例 3^2=9，其中a=3，b=2，N=9
print(compute_n(3,2))
print(compute_a(2,9))
print(compute_b(3,9))
```

```python
# B4554 [GESP202606 二级] 菱形

# 输入
n = int(input())

# 计算
size = 2 * n - 1                                        # 网格大小
center = n                                              # 中心行/列索引（基于1的坐标系）

for i in range(1, size + 1):                            # 遍历每一行
    row_chars = []
    for j in range(1, size + 1):                        # 遍历每一列
        if abs(i - center) + abs(j - center) == n - 1:  # 判断当前点 (i, j) 是否在菱形边框上
            row_chars.append('+')
        else:
            row_chars.append('.')
    
    # 输出
    print("".join(row_chars))
```

```python
# B4259 [GESP202503 二级] 等差矩阵

# 输入
n = int(input())
m = int(input())

# 计算&输出
for i in range(1, n + 1):        # 外层循环遍历每一行，i 从 1 到 n
    for j in range(1, m + 1):    # 内层循环遍历每一列，j 从 1 到 m
        print(i * j, end=" ")    # 计算当前元素的值 i * j end=" " 表示打印后不换行，而是加一个空格
    print()                      # 每一行打印结束后，需要换行
```

```python
# B4260 [GESP202503 二级] 时间跨越
from datetime import datetime, timedelta

# 输入
y = int(input())
m = int(input())
d = int(input())
h = int(input())
k = int(input())

# 计算
current_time = datetime(y, m, d, h)                # 创建当前时间的 datetime 对象
future_time = current_time + timedelta(hours=k)    # 增加 k 小时

# 输出
print(future_time.year, future_time.month, future_time.day, future_time.hour)
```

```python
# B4356 [GESP202506 二级] 数三角形

# 输入
n = int(input())

# 计算
count = 0                        # 初始化计数器
for a in range(1, n + 1):        # 外层循环枚举第一条直角边 a，范围从 1 到 n
    for b in range(a, n + 1):    # 内层循环枚举第二条直角边 b。为了保证不重复（a <= b），b 从 a 开始枚举到 n。
        if (a * b) % 2 == 0:     # 面积 = (a * b) / 2，若 a * b 为偶数，则面积为整数
            count += 1

# 输出
print(count)
```

```python
# B4357 [GESP202506 二级] 幂和数

# 输入
l = int(input())
r = int(input())

# 计算
count = 0
for i in range(l, r + 1):
    if bin(i).count('1') <= 2 and i >= 2:  # bin(i).count('1') 计算二进制中1的个数
        count += 1

# 输出
print(count)
```

```python
# B4411 [GESP202509 二级] 优美的数字

# 输入
n = int(input())

# 计算
count = 0

for d in range(1, 10):        # 枚举优美数的首位数字 d (1-9)
    val = 0
    
    for _ in range(4):        # 枚举位数，最多4位（因为 n <= 2025）
        val = val * 10 + d
        if val <= n:
            count += 1
        else:
            break            # 如果当前位数生成的数已经超过n，更长的位数肯定也超过，跳出内层循环

# 输出
print(count)
```

```python
# B4412 [GESP202509 二级] 菱形

# 输入
n = int(input())

# 计算
k = n // 2                                  # n 是奇数，k = n // 2 即为从中心到顶点的曼哈顿距离
for i in range(n):                          # 遍历每一行
    row_chars = []
    for j in range(n):                      # 遍历每一列
        if abs(i - k) + abs(j - k) == k:    # 判断当前点 (i, j) 是否在菱形边框上
            row_chars.append('#')
        else:
            row_chars.append('.')
    
    # 输出
    print("".join(row_chars))
```

```python
# B4448 [GESP202512 二级] 黄金格

# 输入
H = int(input())
W = int(input())
x = int(input())

# 计算
count = 0 

for r in range(1, H + 1):      # 遍历所有格子 r 代表行号，从 1 到 H
    for c in range(1, W + 1):  # c 代表列号，从 1 到 W
        right_val = x + r - c  # 计算不等式右边的值: right = x + r - c
        # 优化：如果右边小于0，因为左边 sqrt(r^2+c^2) >= 0，所以不等式一定不成立
        # 这一步可以省略，因为下面的平方判断也会处理，但加上可以提高效率
        if right_val < 0:
            continue

        # 使用平方后的不等式进行判断，避免浮点数精度问题
        # 原不等式: sqrt(r^2 + c^2) <= right_val
        # 平方后:   r^2 + c^2 <= right_val^2
        left_sq = r * r + c * c
        right_sq = right_val * right_val
        
        if left_sq <= right_sq:
            count += 1

# 输出
print(count)
```
