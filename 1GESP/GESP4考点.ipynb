4.1 函数的定义与调用  

```python
def 函数名(参数列表):
    """文档字符串（可选，用于说明函数功能）"""
    # 函数体：执行的具体代码
    return 返回值  # 可选，若无 return 则默认返回 None

# 基本调用
result = 函数名(参数值)
```

```python
# 1. 定义函数
def calculate_area(width, height, unit="m"):
    """计算矩形面积"""
    area = width * height
    return f"面积为: {area} 平方{unit}"

# 2. 调用函数
# 使用位置参数
print(calculate_area(5, 10)) 

# 使用关键字参数
print(calculate_area(height=10, width=5)) 

# 使用默认参数
print(calculate_area(5, 10, "cm")) 
```

4.2 形参与实参、作用域  

形参（Formal Parameter）‌

定义‌：在‌函数定义‌时括号内声明的变量名。
特点‌：它只是一个占位符，没有实际值，仅在函数内部有效。
示例‌：def add(a, b): 中的 a 和 b 就是形参。

实参（Actual Argument）‌

定义‌：在‌函数调用‌时传递给函数的具体数据或变量。
特点‌：具有实际的值，用于初始化对应的形参。
示例‌：add(3, 5) 中的 3 和 5 就是实参。

作用域（Scope）
作用域决定了变量在代码中的‌可见范围‌和‌生命周期‌。Python 遵循 ‌LEGB‌ 规则查找变量：

局部作用域（Local）‌
范围‌：函数内部定义的变量。
特点‌：只能在当前函数内访问，函数执行结束后变量销毁。
示例‌：add 函数内的 result = a + b，result 是局部变量。

全局作用域（Global）‌
范围‌：模块级别（文件顶层）定义的变量。
特点‌：在整个文件中可见。若要在函数内‌修改‌全局变量，需使用 global 关键字；若仅读取，可直接访问。

嵌套/闭包作用域（Enclosing）‌
范围‌：外层函数的局部作用域。
特点‌：内部函数可以访问外部函数的变量。若需修改，需使用 nonlocal 关键字。

内置作用域（Built-in）‌
范围‌：Python 预定义的标识符（如 print, len, int）。

```python
x = 10  # x是全局变量

def test_scope(y):           # y是形参
    print(f"参数 y: {y}")
    z = 5                    # z是局部变量
    print(f"局部变量 z: {z}")
    # global x; x = 20       # 修改全局变量 x
    print(f"全局变量 x: {x}")    

test_scope(3)                # 3是实参
# print(z)                   # 报错：NameError，z 在局部作用域外不可见
```

4.3 C++指针类型的概念及基本应用  

指针(Pointer)是 C++ 中的一种变量类型，它的作用是存储另一个变量的内存地址。简单来说，指针可以“指向”其他变量的内存位置。每个变量都有一个固定的内存地址，指针的值就是该内存地址。  

取地址符 &：用于获取变量的内存地址。例如，&a 表示获取变量 a 的内存地址。  
解引用符 *：用于访问指针所指向的内存位置中的值。例如，*p 表示访问指针 p 所指向的变量的值。

指针的声明与普通变量相似，但需要在变量类型前加上 * 符号（解引用符）。比如，int* p 声明了一个指向 int 类型的指针变量 p。  

```python
#include <iostream>
using namespace std;

int main()
{
    int a = 10;               // 声明一个整数变量a并赋值为10
    int* p = &a;              // 声明一个指针p，存储a的地址
    
    cout << *p << endl;       // 通过指针p解引用，输出a的值（即10）
    cout << a << endl;        // 输出变量a的值（即10）
    cout << "---------" << endl;
    cout << &p << endl;       // 输出指针p存储的地址，即指针p本身的地址
    cout << "---------" << endl;
    cout << &a << endl;       // 输出变量a的地址
    cout << *&p << endl;      // 解引用指针p的地址，输出指针p存储的值，即a的地址
    cout << p << endl;        // 输出指针p存储的值，即a的地址
    
    return 0;	
}
```

```python
# 控制台输出
10                            // *p：通过解引用指针 p，访问指针所指向地址的值，即 a 的值，输出为 10。
10                            // a：直接输出变量 a 的值，结果为 10。
---------
0x70fe10                      // &p：输出指针 p 存储的地址，即指针变量 p 在内存中的位置。
---------
0x70felc                      // &a：输出变量 a 的地址，表示 a 存储在内存中的位置。
0x70felc                      // *&p：这实际上是先取 p 的地址，再解引用其值。结果是输出 p 所指向的地址，即 a 的地址。
0x70felc                      // p：直接输出指针 p 存储的值，表示 a 的内存地址。
```

空指针‌：一个指针初始化为nullptr（C++11及以后版本）或NULL（在C++中通常等同于0或nullptr），表示它不指向任何东西。这是一种好的编程习惯，可以避免野指针的问题。

‌野指针‌：未初始化或已经释放的指针称为野指针。使用野指针访问内存可能会导致未定义行为。始终确保在使用指针之前初始化它们。
