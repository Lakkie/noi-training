```python
# B4495 [GESP202603 一级] 交朋友 （无数组，仅使用顺序结构）

# 输入
h1 = int(input())
h2 = int(input())
h3 = int(input())
h4 = int(input())

# 计算
alice = h1
min_diff = 10000           # 最小身高差，初始化一个足够大的差值
best_h = 0                 # 身高最接近Alice的朋友身高

diff = abs(h2 - alice)     # 朋友和Alice的身高差取绝对值 15
if diff < min_diff:        # 如果找到更小的差值
    min_diff = diff        # 15
    best_h = h2            # 165
elif diff == min_diff:     # 如果差值一样
    if h2 < best_h:        # 选更矮的
        best_h = h2

diff = abs(h3 - alice)     # 朋友和Alice的身高差取绝对值 135 - 150 = -15
if diff < min_diff:        # 如果找到更小的差值
    min_diff = diff
    best_h = h3
elif diff == min_diff:     # 如果差值一样 15
    if h3 < best_h:        # 选更矮的 135 < 165
        best_h = h3        # 135

diff = abs(h4 - alice)     # 朋友和Alice的身高差取绝对值 133 - 150 = -17
if diff < min_diff:        # 如果找到更小的差值 17 < 15
    min_diff = diff
    best_h = h4
elif diff == min_diff:     # 如果差值一样 17 == 15
    if h4 < best_h:        # 选更矮的
        best_h = h4

# 输出
print(best_h)
```

```python
# B4495 [GESP202603 一级] 交朋友 （数组+循环结构）

# 输入
h1 = int(input())
h2 = int(input())
h3 = int(input())
h4 = int(input())

# 计算
alice = h1
friends = [h2, h3, h4]

min_diff = 10000           # 最小身高差，初始化一个足够大的差值
best_h = 0                 # 身高最接近Alice的朋友身高

for h in friends:
    diff = abs(h - alice)  # 身高差取绝对值
    if diff < min_diff:    # 如果找到更小的差值
        min_diff = diff
        best_h = h
    elif diff == min_diff: # 如果差值一样
        if h < best_h:     # 选更矮的
            best_h = h

# 输出
print(best_h)
```

```python
# B4551 [GESP202606 一级] 去旅行

# 输入
a = int(input())
b = int(input())
c = int(input())
d = int(input())

# 计算
price1 = a          # 方案1: 直飞
price2 = b + c      # 方案2: 高铁(A->C) + 飞机(C->B)
price3 = b + d      # 方案3: 高铁(A->C) + 高铁(C->B)
min_price = min(price1, price2, price3) # 找出最小价格

# 输出
print(min_price)
```

```python
# B4552 [GESP202606 一级] 交税

# 输入和计算
total_tax = 0.0                        # 初始化总税额

for _ in range(12):                    # 循环处理12个月的数据
    income = float(input())            # 输入当月劳务报酬
    if income > 800:                   # 判断是否超过800元
        tax = (income - 800) * 0.2     # 计算超过部分的20作为当月税额
        total_tax += tax               # 累加到总税额
                                       # 如果不超过800，税额为0，无需操作

# 输出
print(f"{total_tax:.2f}")
```

```python
# B4257 [GESP202503 一级] 图书馆里的老鼠

# 输入
n = int(input())                           # 书的数量 n
x = int(input())                           # 第二行：老鼠啃光一本书需要的时间 x
y = int(input())                           # 第三行：经过的总时间 y

# 计算
eaten_full_books = y // x                  # 老鼠啃掉的完整书本数
if y % x == 0:                             # 判断是否有正在啃但未完成的书
    damaged_books = eaten_full_books       # 刚好啃完整数本，没有残缺的书
else:
    damaged_books = eaten_full_books + 1   # 有残缺的书，这本也不算完整，所以损坏的书本数要加1
remaining_books = n - damaged_books        # 剩余完整的书

# 输出
print(remaining_books)
```

```python
# B4258 [GESP202503 一级] 四舍五入

# 输入
n = int(input())
for _ in range(n):              # 循环 n 次，处理每一个整数
    a = int(input())            # 读取当前整数

    # 计算
    ge = a % 10                 # 获取个位数字
    shi = a // 10               # 获取十位及以上的数字
    if ge < 5:                  # 判断个位是否小于 5
        s = shi * 10            # 舍去个位
    else:
        s = (shi + 1) * 10      # 进位

    # 输出
    print(s)
```

```python
# B4354 [GESP202506 一级] 假期阅读

# 输入
n = int(input())     # 书的总页数 n
k = int(input())     # 每天最多阅读的页数 k
t = int(input())     # 假期天数 t

# 计算
max_read = k * t     # 假期总共能阅读的页数
if max_read > n:     # 如果能读的页数超过书的页数，则最多只能读 n 页
    s = n
else:
    s = max_read     # 否则，最多能读 max_read 页

# 输出
print(s)
```

```python
# B4355 [GESP202506 一级] 值日

# 输入
m = int(input())
n = int(input())

# 计算
i = 1                               # 从1开始枚举，寻找第一个同时能被m和n整除的数
while True:
    if i % m == 0 and i % n == 0:
        # 输出
        print(i)
        break
    i += 1
```

```python
# B4409 [GESP202509 一级] 商店折扣

# 输入
x = int(input())
y = int(input())
n = int(input())
p = int(input())

# 计算方案一：满减优惠
if p >= x:
    price1 = p - y
else:
    price1 = p

# 计算方案二：打折优惠
price2 = p * n / 10.0 # 注意使用 10.0 确保进行浮点数除法
min_price = min(price1, price2) # 取两种方案中的最小值

# 输出
print(f"{min_price:.2f}")
```

```python
# B4410 [GESP202509 一级] 金字塔

# 输入
n = int(input())

# 计算
total_stones = 0             # 总石块数
for i in range(1, n + 1):    # 循环从 1 到 n，累加每一层的石块数 (i * i)
    total_stones += i * i

# 输出
print(total_stones)
```

```python
# B4445 [GESP202512 一级] 小杨的爱心快递

# 输入
v = float(input())
g = float(input())
m = float(input())
n = float(input())

# 计算
cost_v = 0.5 * v # 体积运费
if g < 300: # 重量运费
    cost_g = m
else:
    cost_g = n
s = min(cost_v, cost_g) # 取最小值

# 输出
print("{0:.1f}".format(s)) 
```
