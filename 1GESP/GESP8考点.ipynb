```python
import sys
from collections import deque

sys.setrecursionlimit(2 * 10**6)

class TreeLCA:
    def __init__(self, n, edges, root=1):
        self.n = n
        self.LOG = n.bit_length()
        self.graph = [[] for _ in range(n + 1)]
        for u, v in edges:
            self.graph[u].append(v)
            self.graph[v].append(u)
        
        self.depth =  * (n + 1)
        # up[i][j]: 节点 i 的第 2^j 级祖先
        self.up = [ * self.LOG for _ in range(n + 1)]
        self._bfs(root)

    def _bfs(self, root):
        q = deque([root])
        self.depth[root] = 1
        self.up[root] = 0  # 根节点父节点设为0
        while q:
            u = q.popleft()
            for v in self.graph[u]:
                if v != self.up[u]:
                    self.depth[v] = self.depth[u] + 1
                    self.up[v] = u
                    q.append(v)
        
        # 预处理倍增表
        for j in range(1, self.LOG):
            for i in range(1, self.n + 1):
                mid = self.up[i][j-1]
                self.up[i][j] = self.up[mid][j-1] if mid else 0

    def lca(self, u, v):
        if self.depth[u] < self.depth[v]:
            u, v = v, u
        # 1. 提至同深
        diff = self.depth[u] - self.depth[v]
        for j in range(self.LOG - 1, -1, -1):
            if diff & (1 << j):
                u = self.up[u][j]
        if u == v: return u
        # 2. 同时上跳
        for j in range(self.LOG - 1, -1, -1):
            if self.up[u][j] != self.up[v][j]:
                u = self.up[u][j]
                v = self.up[v][j]
        return self.up[u]

    def kth_ancestor(self, node, k):
        """求节点 node 的第 k 级祖先"""
        for j in range(self.LOG - 1, -1, -1):
            if k & (1 << j):
                node = self.up[node][j]
                if not node: break
        return node
```
