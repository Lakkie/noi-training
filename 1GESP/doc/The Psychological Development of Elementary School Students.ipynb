# 小学儿童的学习发展

小学儿童的学习是为一生的学习活动奠定基础知识和学习能力的时期，是心理发展的一个重要阶段。老师为主导，要学会学习，学习规划、能力和技巧的提升。  
小学儿童的学习特点  
**学习动机**  
外部动机为主，内部动机逐步发展。  
逐渐觉得知识和学科很有趣，内驱力发展。  
**学习兴趣**  
最初对学习过程、外部活动有兴趣，后逐渐对学习内容、对需要独立思考的学习作业更有兴趣。  
对不同学科的学习兴趣从不分化到初步分化，但很不稳定。不要给孩子贴标签，比如说你喜欢英语、不喜欢数学等等。对学科兴趣的分化时间一般从三年级开始。  
对具体事实和经验的知识有兴趣，对抽象因果知识的兴趣在逐步发展。  
**学习态度**  
对教师的态度。低年级儿童（一、二年级）对教师怀有特殊的尊敬和依恋之情。高年级儿童逐渐对教师产生选择性的、怀疑的态度。例如高年级孩子会表达出某某老师讲课特别好，某某老师对学生特别好。如果孩子对老师有怀疑或者不满情绪时，家长要对孩子正确的引导。  
对集体的态度。低年级儿童对集体尚未形成归属感。高年级儿童在老师的带领下，逐渐有了班集体的概念，有了归属感，责任感，荣誉感。会以自己属于班级的一员而自豪，会说这是我们班的，我们班怎么怎么样。  

# 小学儿童的认知发展

**感觉的发展**  
视觉的感受性提高，7岁时增长最快。视觉的感受性指的是从明亮的环境进入黑暗环境之后，我们能看清楚的反应。

**知觉的发展**  
空间知觉：对上下、前后可以很好的辨别；左右还不完善。9-11岁，可灵活掌握左右概念。
时间知觉：时间观念较快发展，对时距估计的正确性也在不断提高。  

**注意力的发展**  
无意注意占主导，小学五年级左右， 有意注意才占主导。越小的孩子注意力越容易受到外界的干扰，越容易分散。
具体生动、直观形象的事物易引起注意。注意带有情绪色彩（情绪性非常重要，快乐教育的必要性）。 教具需要灵活多变，生动形象。
注意时间在逐渐延长。
一般情况：
4-7岁，10分钟
7-10岁，20分钟
10-12岁，25分钟
12岁以上，30分钟  

**记忆力的发展**  
记忆容量增加。小学儿童对数字的短时记忆容量最大，字母次之，部首的记忆容量最小。
记忆策略的发展。小学中高年级（10 岁左右）儿童已经较好掌握复述、组织的记忆策略。
对低年级的儿童进行提示和训练能有效提高记忆量。
记忆策略有：复诵、组织、系统化、巧妙加工。
对 6-8 岁的小学低年级和中年级儿童进行记忆策略指导和提示效果最好。
形象记忆和抽象记忆的发展。小学低年级儿童以形象记忆为主；高年级抽象记忆逐步发展。
对小学低年级的孩子进行辅导时建议具体形象多样化。  

**思维的发展**  
基本特征：具体形象思维向抽象逻辑思维过渡
转折年龄：9-10 岁之间。（我国研究发现一般在四年级左右）
在低年级可以进行围棋等逻辑思维的训练，但切记不可过度。
（1）概括能力的发展：三阶段
直观形象：7-8岁
形象抽象：8-10岁
初步本质抽象：10-12岁 三人成虎
（2）比较能力的发展
对不同点的比较发展较早，对相同点的比较发展较晚。
（3）自我中心表现和脱自我中心化
4-7 岁儿童具有自我中心现象。
9 岁以后儿童基本上摆脱了自我中心的影响。 能够从他人的角度去考虑问题。
8 岁组儿童处于脱自我中心化的转折时段。

# 小学儿童的社会性发展

**自我意识的发展**  
趋势：1-3年级上升期，3-5年级平稳期， 5-6年级上升期
自我意识的发展在家庭教育中经常体现为反抗。

自我评价能力  
三岁都出现了，自卖自夸。
多方面的评价，如学业成绩、运动能力等。当孩子有这种方式评价的时候，我们不能仅仅说你很棒啊，这个太笼统了。就像夸小孩子一样要具体。  
社会支持因素起重要作用，尤其父母和同学。中高年级时对自己的评价表现出独立性。自我评价能力和自我意识的发展紧密相关。  
自我评价与情感密切联系。喜欢自己的儿童，情绪最快乐。情感的作用真的太重要了，记忆力，注意力和自我评价都密切相关。小学儿童自我评价与学业经验、同伴交往、自信心等都有密切关系。当孩子有比较负面的自我评价时，这时就需要我们去关注下背后的原因了。

自我控制能力  
罗腾博格-延迟满足研究  
自我控制的能力逐渐增加。
6-8 岁 有 1/3 选择等待
9-11 岁有 1/2 选择等待
12-15 岁的儿童几乎都愿意等待
在6-11岁之间我们可以有意识的进行延迟满足的训练，从而训练孩子自我控制的能力。  
家庭教育  
民主型---容易抗拒诱惑的自我控制能力。 在这种环境中的孩子会经常思考我怎么做会比较好，有自主的选择权。  
独裁型、惩罚型或溺爱型---剥夺儿童练习自我控制的机会和动力，使其缺乏自我控制能力。

**道德的发展**  
道德情感的发展：日益丰富。如：羞愧、内疚、情感共鸣等  
道德认知的发展：三阶段（皮亚杰）  
前道德 4-5岁，不能判断，只能直接接受行为结果。  
他律道德 5-8、9岁，严格遵守规则，不能变。根据后果评定行为的好坏。  
自律道德 8、9岁后，规则可变。依据动机判断行为的好坏。  
道德行为的发展：  
亲社会行为  
指对他人有益、对社会有利的积极行为及趋向，表现为分享、 合作、帮助等。
亲社会行为的获得应具备如下条件 ：
道德动机的发展。有一种需要想要去和他人合作。
替他人着想的能力。
掌握有效的帮助人的知识和技能。  
攻击行为  
是指针对他人的具有敌视性、伤害性或破坏性的行为，也称侵犯行为。有意的孤立也是一种攻击行为。
攻击行为的基本要素是伤害意图。
对攻击行为的控制：
改善儿童所处的环境条件。 儿童的玩伴，家庭和学校环境。
教给儿童减少冲突的有效策略。
增加对攻击行为有害后果的了解。对孩子进行同理心教育。
发挥榜样的作用等。给与孩子积极正向的引导。  

**小学儿童的人际关系** 
总体特征：  
与父母的交往随年龄增长而下降；
与同龄伙伴的交往随年龄增长而上升；
与教师的交往在小学中年级前随年龄的增长而上升，之后则一直维持在交往比率的20%左右。  

亲子关系  
直接交往时间明显减少
父母教养关注重点的转移为学习、交友问题等
父母对儿童控制和儿童自主管理的消长变化  

同伴关系  
总体特征  
出现了真正的同伴群体--有规则、互动、有归属感、形成自己规范的群体。有了明显的群体认同感

同伴交往中儿童的人气特点  
受欢迎的儿童
不受欢迎的儿童（易产生适应困难） ：会被群体排斥，产生一种交往的恐惧感。可能缺乏交友的技能或者表达不够准确等等。
受忽略的儿童  

友谊发展  
在小学阶段时是逐步的在学习交友的技巧。他在寻找哪一类人和他臭味相投。
6-9 岁：单向帮助关系。要求朋友能服从自己的愿望和要求，否则就不是朋友。
9-12 岁：双向帮助关系。具有相互性，有功利性特点，被称为“顺利时的合作”，不能“共患难” 。
12 岁以后：亲密而又相对持久的共享关系。

师生关系  
对教师的态度  
低年级儿童充满崇拜和敬畏。
三年级开始，不再无条件服从和信任教师。对老师产生独立的评判和怀疑的态度

教师的期望  
罗森塔尔效应：指人们对他人的期望会影响其行为和表现，从而使期望成为自我实现的预言。
提供不同的学习机会
