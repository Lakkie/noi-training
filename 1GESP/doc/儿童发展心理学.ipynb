一、0-3岁 安全感建立关键期
核心心理特点‌：80%的大脑发育在此阶段完成，是神经突触爆发期，逐步建立自我意识，情绪调节能力弱，依赖照料者获得安全感。
实操育儿要点‌：及时回应孩子的需求，通过拥抱、眼神交流建立安全依恋；提供不同材质的感官玩具，多和孩子对话互动，引导孩子用语言表达情绪而非哭闹。

二、4-6岁 学前能力启蒙期
核心心理特点‌：处于皮亚杰前运算阶段，思维具象化，想象力达到顶峰，有意注意力时长仅15-20分钟，社交模仿行为增多。
实操育儿要点‌：不把孩子的幻想行为当成“撒谎”，保护其创造力；用绘本和情景游戏教孩子识别情绪，通过简单规则培养孩子的秩序感。

三、7-12岁 小学习惯养成期
核心心理特点‌：自我意识快速发展，开始在意同伴评价，规则意识逐步建立，学习兴趣成为核心成长动力。
实操育儿要点‌：尊重孩子的独立想法，不随意否定；帮孩子规划学习节奏，引导孩子客观看待他人评价，培养良好的学习习惯。

四、13-18岁 青春期人格定型期
核心心理特点‌：叛逆心理凸显，情绪波动大，极度渴望隐私与独立，自我认知逐步成熟，开始思考未来发展方向。
实操育儿要点‌：放下家长姿态平等沟通，不强行窥探孩子隐私；给予孩子充分信任，帮孩子疏导学业压力，和孩子共同探讨未来规划。
