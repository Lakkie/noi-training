```python
n=int(input())
print("."*(n//2)+"#"+"."*(n//2))
for a in range(1,n//2+1):
    print("."*(n//2-a)+"#"+"."*(a*2-1)+"#"+"."*(n//2-a))
i=n//2-1
while(i>0):
    print("."*(n//2-i)+"#"+"."*(i*2-1)+"#"+"."*(n//2-i))
    i-=1
print("."*(n//2)+"#"+"."*(n//2))
```

```python
n=int(input())
print("."*(n-1)+"+"+"."*(n-1))
for a in range(1,n-1):
    print("."*(n-a-1)+"+"+"."*(a*2-1)+"+"+"."*(n-a-1))
i=n-1
while(i>0):
    print("."*(n-i-1)+"+"+"."*(i*2-1)+"+"+"."*(n-i-1))
    i-=1
print("."*(n-1)+"+"+"."*(n-1))
```

```python
import math
h=int(input())
w=int(input())
x=int(input())
cc=0
for r in range(1,h+1):
    for c in range(1,w+1):
        r2=x+r-c
        if(r2<0):
            continue
        l=math.sqrt(r*r+c*c)
        if(l<=r2):
            cc+=1
print(cc)
```
