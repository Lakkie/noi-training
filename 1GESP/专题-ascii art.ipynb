`````python
# 图案组合
pattern1 = """
 ┏┓   ┏┓  
┏┛┻━━━┛┻┓  
┃       ┃  
┃   ━   ┃  
┃ ┳┛ ┗┳ ┃  
┃       ┃  
┃   ┻   ┃  
┃       ┃  
┗━┓   ┏━┛  
  ┃   ┃   
  ┃   ┃   
  ┃   ┗━━━┓
  ┃       ┣┓
  ┃       ┏┛
  ┗┓┓┏━┳┓┏┛
   ┃┫┫ ┃┫┫  
   ┗┻┛ ┗┻┛  
"""

pattern2 = """
      ==
      ][
      ][
      ][
  ((==<>==))
     /  \\
     \\  /
     |  |
     |  |
     |  |
     |  |
     |  |
   __]__[__
  / ~~~~~~ \\
 / ~~~~~~~~ \\
______________
"""

p1 = pattern1.splitlines()
p2 = pattern2.splitlines()
pmax = max(len(p1),len(p2))
pmin = min(len(p1),len(p2))

for i in range(pmin):
    print(p1[i] + "\t" + p2[i])
`````
