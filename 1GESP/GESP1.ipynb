1.4 基本数据类型（整型、浮点型、字符型、布尔型）

```python
# Numbers 整数类型
age = 25
print("Age:", age)
```

```python
# Numbers 浮点类型
height = 5.9
print("Height:", height)
```

```python
# String 字符串类型
name = "Alice"
print("Name:", name)
```

```python
# List 列表类型
fruits = ["apple", "banana", "cherry"]
print("Fruits:", fruits)
```

```python
# Tuple 元组类型
fruits = ("apple", "banana", "cherry")
print("Fruits:", fruits)
```

```python
tuple = ( 'runoob', 786 , 2.23, 'john', 70.2 )
tinytuple = (123, 'john')
 
print(tuple)               # 输出完整元组
print(tuple[0])            # 输出元组的第一个元素
print(tuple[1:3])          # 输出第二个至第四个（不包含）的元素 
print(tuple[2:])           # 输出从第三个开始至列表末尾的所有元素
print(tinytuple * 2)       # 输出元组两次
print(tuple + tinytuple)   # 打印组合的元组
```

```python
# Dictionary 字典类型
person = {"name": "Bob", "age": 30}
print("Person:", person)
```
