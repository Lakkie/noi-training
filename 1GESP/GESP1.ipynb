1.5 控制语句结构（顺序、循环、选择）

```python
# 顺序结构 Sequential Structure

a = int(input())
b = int(input())
c = int(input())
d = int(input())

print(a)
print(b)
print(c)
print(d)
```

```python
# 循环结构 Loop Structure

# for循环 适用于已知循环次数或遍历序列（如列表、字符串、范围等）。

for _ in range(12): 
    print(_)
```

```python
# while循环 适用于未知循环次数，只要条件为真 (True) 就持续执行。‌务必在循环体内修改条件变量‌，否则可能导致死循环。

# 示例：计算 1 到 50 的累加和
i = 1
total_sum = 0

while i <= 50:
    total_sum += i  # 累加
    i += 1          # 更新计数器，避免死循环

print(f"1到50的和为: {total_sum}")
```

```python
# 循环控制

# 打印 1 到 10 之间的奇数
for num in range(1, 11):
    if num % 2 == 0:
        continue  # 如果是偶数，跳过本次循环
    print(num)
    
    if num == 7:
        break     # 如果数字是7，提前结束循环
```

```python
# 选择结构 Selection Structure
# 单分支 (if)‌：满足条件则执行，否则跳过。

guess = int(input("请猜一个数字: "))

if guess == 6:
    print("你猜中了！")
```

```python
# 双分支 (if-else)‌：满足条件执行 A，不满足执行 B。

age = int(input("请输入年龄: "))

if age >= 18:
    print("你已经成年了！")
else:
    print("你还未成年！")
```

```python
# 多分支 (if-elif-else)‌：多个条件依次判断，执行第一个满足条件的分支。

score = float(input("请输入百分制成绩: "))

if score > 100 or score < 0:
    print("输入数据无效")
elif score >= 90:
    print("等级: A")
elif score >= 80:
    print("等级: B")
elif score >= 70:
    print("等级: C")
elif score >= 60:
    print("等级: D")
else:
    print("等级: E (不及格)")
```

1.6 基本运算（算术运算、关系运算、逻辑运算）

```python
# 算术运算 Arithmetic operations

a = 10
b = 3

# 基本加减乘
print(f"加法: {a} + {b} = {a + b}") 
print(f"减法: {a} - {b} = {a - b}") 
print(f"乘法: {a} * {b} = {a * b}") 

# 除法相关
print(f"除法: {a} / {b} = {a / b}")                     # 3.3333333333333335 (浮点数)
print(f"整除: {a} // {b} = {a // b}")                   # 3 (只保留整数部分)
print(f"取模: {a} % {b} = {a % b}")                     # 1 (余数)

# 幂运算
print(f"幂运算: {a} ** {b} = {a ** b}")                 # 1000 (10的3次方)
```

```python
# 关系运算 Relational operations

x = 10
y = 20

# 等于
print(f"x == y: {x == y}")  
# 不等于
print(f"x != y: {x != y}")  

# 大小比较
print(f"x > y: {x > y}")    
print(f"x < y: {x < y}")    
print(f"x >= 10: {x >= 10}")
print(f"y <= 15: {y <= 15}")

# 字符串比较 (按字典序/ASCII码比较)
print(f"'apple' < 'banana': {'apple' < 'banana'}")
```

```python
# 逻辑运算 Logical operations

age = 25
is_student = True
has_ticket = False

# and: 两者都满足
print(f"age > 18 and is_student: {age > 18 and is_student}")       # 年龄大于18 且 是学生

# or: 满足其一即可
print(f"has_ticket or is_student: {has_ticket or is_student}")     # 有票 或者 是学生

# not: 取反
print(f"not is_student: {not is_student}")                         # 不是学生

# 复杂组合
is_adult_working_age = (age >= 18) and (age <= 60)                 # 年龄在18到60之间 (包含18和60)
print(f"是否是适龄劳动力: {is_adult_working_age}")
```

1.8 Turtle绘图

```python
import turtle            #导入turtle模块

# 强制重置所有海龟和屏幕状态
try:
    turtle.reset()
except turtle.Terminator:
    pass # 如果之前已经终止，忽略错误
    
screen = turtle.Screen() # 创建画布（Canvas）
t = turtle.Turtle()      # 创建海龟（Turtle）对象
t.forward(100)           # 海龟向前移动100单位
t.right(90)              # 海龟向右转90度
t.forward(100)           # 再次向前移动100单位，形成一个直角

turtle.done()            # 保持窗口打开直到用户关闭它。
```

```python
import turtle

# 强制重置所有海龟和屏幕状态
try:
    turtle.reset()
except turtle.Terminator:
    pass # 如果之前已经终止，忽略错误

screen = turtle.Screen()
t = turtle.Turtle()
t.speed(2)               # 设置绘图速度为2
t.color("blue")          # 设置画笔颜色为蓝色

# 绘制正方形的四条边
for _ in range(4):
    t.forward(100)       # 向前移动100单位
    t.right(90)          # 向右转90度
    
turtle.done()            # 保持窗口打开直到用户关闭它。
```

```python
import turtle

# 强制重置所有海龟和屏幕状态
try:
    turtle.reset()
except turtle.Terminator:
    pass # 如果之前已经终止，忽略错误
    
screen = turtle.Screen()
t = turtle.Turtle()
t.speed(2)               # 设置绘图速度为2
t.color("blue")          # 设置画笔颜色为蓝色
    
# 绘制圆形（360度）
t.penup()                # 抬起画笔，移动时不绘制线条
t.goto(50, 0)            # 移动到新位置，不绘制线条到该位置（x=50, y=0）
t.pendown()              # 放下画笔，开始绘制线条到该位置以后的位置移动都会绘制线条了。
t.circle(50)             # 以当前位置为中心，半径为50单位画圆（360度）
    
turtle.done()            # 保持窗口打开直到用户关闭它。
```

```python
# B4495 [GESP202603 一级] 交朋友 （数组+循环结构）

# 输入
h1 = int(input())
h2 = int(input())
h3 = int(input())
h4 = int(input())

# 计算
alice = h1
friends = [h2, h3, h4]

min_diff = 10000           # 最小身高差，初始化一个足够大的差值
best_h = 0                 # 身高最接近Alice的朋友身高

for h in friends:
    diff = abs(h - alice)  # 身高差取绝对值
    if diff < min_diff:    # 如果找到更小的差值
        min_diff = diff
        best_h = h
    elif diff == min_diff: # 如果差值一样
        if h < best_h:     # 选更矮的
            best_h = h

# 输出
print(best_h)
```

```python
# B4552 [GESP202606 一级] 交税

# 输入和计算
total_tax = 0.0                        # 初始化总税额

for _ in range(12):                    # 循环处理12个月的数据
    income = float(input())            # 输入当月劳务报酬
    if income > 800:                   # 判断是否超过800元
        tax = (income - 800) * 0.2     # 计算超过部分的20作为当月税额
        total_tax += tax               # 累加到总税额
                                       # 如果不超过800，税额为0，无需操作

# 输出
print(f"{total_tax:.2f}")
```
