# GESP编程能力等级认证 一级

## 课程基本信息

课程名称：GESP一级  
授课对象：小学生  
前置基础：掌握电脑基本操作  
教学目标概述：覆盖GESP一级全部考纲知识点，帮助学生建立基础编程思维，达到官方认证考试合格水平。  

### 知识目标
1. 理解编程核心基础概念。  
2. 熟悉计算机基础知识与计算机历史。  
3. 了解C++标准的版本。
4. 熟练掌握对应语言（C++或Python）的程序基本结构。  

### 能力目标
1. 能独立完成编程环境搭建。（下载和安装开发工具）  
2. 能独立完成代码编写、调试运行的全流程操作。  
3. 能独立解决GESP1级考试中80%以上的编程题，自主排查缩进、符号、语法类常见错误。  
4. 能独立完成“分析需求-拆解步骤-编写代码-调试优化”的标准化解题步骤。   

### 素养目标
1. 建立计算思维，学会拆解复杂问题，用顺序、分支、循环逻辑梳理任务，养成严谨的逻辑思考习惯。   
2. 通过把创意转化为可运行的作品，在试错调试中提升动手能力与抗挫力。  
3. 将编程和数学、科学等课内知识结合，用代码模拟学科场景，用编程工具辅助理解课内知识点。  
4. 在组队完成项目的过程中学会分工配合，清晰讲解自己的作品思路，提升团队协作与公开表达能力。
5. 建立对AI、数字工具的正确认知，学会合理利用技术解决生活问题，养成规范的数字行为习惯。  
6. 通过趣味项目获得成就感，建立对信息学编程的长期兴趣，为后续更高等级学习打下基础。
