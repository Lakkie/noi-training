# GESP编程能力等级认证 一级

## 课程基本信息

课程名称：GESP一级  
授课对象：小学生  
前置基础：掌握电脑基本操作  
教学目标概述：覆盖GESP一级全部考纲知识点，帮助学生建立基础编程思维，达到官方认证考试合格水平。  

### 知识目标
1. 理解编程核心基础概念，熟练掌握对应语言的程序基本结构。  
2. 。  

### 能力目标
能独立完成编程环境搭建、代码编写、调试运行的全流程操作。
能独立解决GESP1级考试中80%以上的编程题，自主排查缩进、符号、语法类常见错误。
养成“分析需求-拆解步骤-编写代码-调试优化”的标准化解题习惯。 

### 素养目标
建立严谨的逻辑思维，养成规范书写代码的良好习惯。
通过趣味项目获得成就感，建立对信息学编程的长期兴趣，为后续更高等级学习打下基础。
