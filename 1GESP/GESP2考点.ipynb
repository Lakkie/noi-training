2.1 计算机的存储与网络

计算机存储‌：是计算机系统中用于保存和读取数据的核心部件，分为内存、外存等不同层级，实现数据的持久化和高速读写。

计算机网络‌：将多台独立计算机通过通信设备连接，实现数据传输、资源共享的系统，核心是完成设备间的信息交互。

2.3 流程图的概念与描述

流程图：是用特定符号直观表示算法和流程的图示工具。

```python
基本符号：  
1. 起止符：椭圆形，表示开始或者结束。
2. 处理符：矩形，表示操作步骤。
3. 判断符：菱形，表示决策判断，有多个出口。
4. 输入输出符：平行四边形。
5. 流程线：带箭头的直线，表示执行方向。
```
