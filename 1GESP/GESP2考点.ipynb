# 2.4 ASCII 编码

American Standard Code for Information Interchange 美国信息交换标准代码，它是一套基于拉丁字母的通用电脑编码系统，共定义了128个字符。

48-57 0-9  
65-90 A-Z  
97-122 a-z  

```python
print(f"{'Dec':<5} {'Hex':<5} {'Char':<10} {'Description'}")
print("-" * 40)

for i in range(128):
    char = chr(i)
    hex_val = hex(i)
    
    # 判断是否为可打印字符 (32-126)
    if 32 <= i <= 126:
        description = "Printable"
    elif i == 127:
        description = "DEL (Delete)"
    else:
        description = "Control Character"
        
    print(f"{i:<5} {hex_val:<5} {repr(char):<10} {description}")
```

# 2.5 数据类型的转换

数字类型 (Numeric Types)  
用于存储数值数据，支持算术运算。  

整型 (int)‌：表示整数，具有无限精度（即没有最大值限制，取决于内存大小）。  
浮点型 (float)‌：表示带有小数点的实数，通常对应 C 语言中的 double类型。  
复数型 (complex)‌：由实部和虚部组成，例如 3+4j。  
布尔型 (bool)‌：True 和 False。‌注意‌：在 Python中，布尔型是整型的子类，True 等价于 1，False 等价于 0。  

```python
序列类型 (Sequence Types)  
用于存储有序的元素集合，可以通过索引访问。  

字符串 (str)‌：不可变的字符序列，用于存储文本。  
列表 (list)‌：‌可变‌的有序序列，可以包含不同类型的元素，支持动态添加、删除和修改。  
元组 (tuple)‌：‌不可变‌的有序序列，一旦创建不能修改，常用于保护数据不被更改或作为字典的键。  
范围 (range)‌：表示一个不可变的数字序列，常用于 for 循环中生成整数序列。  
```

映射类型 (Mapping Type)  

字典 (dict)‌：以‌键值对‌（Key-Value）形式存储数据。  
键（Key）必须是不可变类型（如字符串、数字、元组）。  
值（Value）可以是任意类型。  
在 Python 3.7+ 版本中，字典保持插入顺序。  

集合类型 (Set Types)  
用于存储无序且不重复的元素集合，支持数学上的集合运算（如交集、并集）。  

集合 (set)‌：‌可变‌的无序不重复元素集。  
冻结集合 (frozenset)‌：‌不可变‌的无序不重复元素集，可作为字典的键或其他集合的元素。  

二进制类型 (Binary Types)  
用于处理原始字节数据，常用于文件操作和网络通信。  

字节 (bytes)‌：‌不可变‌的字节序列。  
字节数组 (bytearray)‌：‌可变‌的字节序列。  
内存视图 (memoryview)‌：允许在不复制数据的情况下访问对象的内部数据。  

特殊类型 (Special Types)  

空值 (NoneType)‌：只有一个实例 None，表示“无”或“空”，常作为函数的默认返回值。  
其他单例对象‌：如 NotImplemented（用于未实现的运算符方法）和 Ellipsis（即 ...，用于切片或占位）。

```python
# 字符串转数字
num_int = int("123")      # 结果: 123
num_float = float("3.14") # 结果: 3.14
# int("12abc") 会报错 ValueError
print(num_int)
print(num_float)
```

```python
# 数字转字符串
str_int = str(123)        # 结果: "123"
str_float = str(3.14)     # 结果: "3.14"
print(str_int)
print(str_float)
```

```python
# 浮点转整数‌：‌直接截断小数部分‌
num_int1 = int(3.99)      # 结果: 3
num_int2 = int(-3.99)     # 结果: -3 (向零取整)
print(num_int1)
print(num_int2)
```

```python
# 整数转浮点
num_float = float(10)     # 结果: 10.0
print(num_float)
```

```python
# 容器类型互转
lst = [1, 2, 3]           # 列表
tup = tuple(lst)          # 列表转元组，结果: (1, 2, 3)
st = set(lst)             # 列表转集合，结果: {1, 2, 3}
lst_back = list(st)       # 集合转列表，结果: [1, 2, 3] (顺序可能因集合无序性而改变)
print(tup)
print(st)
print(lst_back)
```

# 2.6 多层分支/循环结构

多层分支结构 Multi-level Branching Structure  

最佳实践建议  
条件互斥与顺序‌：确保条件之间逻辑清晰，范围判断时要注意顺序（如先判断 >=90 再判断 >=80），防止高分被低分条件提前捕获。  
保留 Else 兜底‌：始终建议保留 else 分支，用于处理意外输入或异常情况，提高程序的健壮性。  
避免过度嵌套‌：如果嵌套层数超过 3 层，代码可读性会大幅下降。此时应考虑提取函数或使用字典映射、策略模式等重构手段。  
利用短路特性‌：在复杂条件中，将计算成本低或更容易为假的条件放在前面，可以提升执行效率。  

```python
# 基础多条件判断
score = float(input("请输入你的成绩: "))

if score >= 90:
    print("优秀")
elif score >= 80:
    print("良好")
elif score >= 60:
    print("及格")
else:
    print("不及格")
```

```python
# 复杂逻辑组合
username = input("请输入用户名: ")
password = input("请输入密码: ")

if username == 'admin' and password == '123456':
    print("管理员登录成功！")
elif username == 'user' and password == 'password':
    print("普通用户登录成功！")
elif username == 'guest':
    print("访客模式，仅可浏览内容。")
else:
    print("用户名或密码错误，请重新输入！")
```

```python
# 嵌套分支结构
year = int(input("请输入年份: "))
month = int(input("请输入月份(1-12): "))

if month in [1, 3, 5, 7, 8, 10, 12]:
    days = 31
elif month in [4, 6, 9, 11]:
    days = 30
elif month == 2:
    # 嵌套分支：判断闰年
    if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
        days = 29
        print(f"{year}年是闰年")
    else:
        days = 28
        print(f"{year}年是平年")
else:
    print("输入的月份无效")
    days = 0

if days > 0:
    print(f"{year}年{month}月共有 {days} 天")
```

```python
# 使用字典映射简化离散值分支
def get_weekday_name(day_num):
    # 定义映射关系
    weekdays = {
        1: "星期一",
        2: "星期二",
        3: "星期三",
        4: "星期四",
        5: "星期五",
        6: "星期六",
        7: "星期日"
    }
    # get方法提供默认值，相当于 else 分支
    return weekdays.get(day_num, "无效的日期编号")

day = int(input("请输入数字(1-7): "))
print(get_weekday_name(day))
```

```python
多层循环结构 Multi-level Loop Structure‌

最佳实践建议
控制嵌套深度‌：建议嵌套层数不超过 3 层。过深的嵌套会导致代码难以阅读和维护，应考虑提取函数或使用递归。  
优化内层逻辑‌：内层循环执行次数最多，应将复杂的计算或判断尽量移至外层，或在内层中使用 break 提前终止无效搜索。  
变量命名清晰‌：多层循环中常用 i, j, k 作为索引，但在处理具体业务数据时，建议使用语义化名称（如 row, col 或 user, order）以提升可读性。  
```

```python
# 基础双层循环 打印 9x9 乘法表
for i in range(1, 10):                       # 外层循环：控制行 (1-9)
    for j in range(1, i + 1):                # 内层循环：控制列 (1-i)
        print(f"{j}x{i}={i*j}", end="\t")
    print()                                  # 每行结束后换行
```

```python
# 基础三层循环 生成 x, y, z 均在 0-1 范围内的所有整数坐标点
for x in range(2):
    for y in range(2):
        for z in range(2):
            print(f"({x}, {y}, {z})")
```

```python
# break 和 continue 默认只影响‌当前所在的最内层循环‌。若需跳出外层循环，通常需借助标志位或异常处理。

target = 5
found = False

for i in range(3):
    for j in range(3):
        if i * 3 + j == target:
            print(f"找到目标: i={i}, j={j}")
            found = True
            break  # 仅跳出内层循环
    if found:
        break      # 根据标志位跳出外层循环
```
