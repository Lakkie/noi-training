```c
#include <iostream>
#include <iomanip> // 用于格式化输出

using namespace std;

// 辅助函数：获取控制字符的描述
string getControlCharDesc(int ascii) {
    switch (ascii) {
        case 0: return "NUL (空)";
        case 1: return "SOH";
        case 2: return "STX";
        case 3: return "ETX";
        case 4: return "EOT";
        case 5: return "ENQ";
        case 6: return "ACK";
        case 7: return "BEL (响铃)";
        case 8: return "BS (退格)";
        case 9: return "HT (制表符)";
        case 10: return "LF (换行)";
        case 11: return "VT";
        case 12: return "FF (换页)";
        case 13: return "CR (回车)";
        case 14: return "SO";
        case 15: return "SI";
        case 16: return "DLE";
        case 17: return "DC1";
        case 18: return "DC2";
        case 19: return "DC3";
        case 20: return "DC4";
        case 21: return "NAK";
        case 22: return "SYN";
        case 23: return "ETB";
        case 24: return "CAN";
        case 25: return "EM";
        case 26: return "SUB";
        case 27: return "ESC";
        case 28: return "FS";
        case 29: return "GS";
        case 30: return "RS";
        case 31: return "US";
        case 127: return "DEL (删除)";
        default: return "";
    }
}

int main() {
    cout << left << setw(6) << "Dec" 
         << setw(6) << "Hex" 
         << setw(15) << "Char/Desc" 
         << "Note" << endl;
    cout << string(40, '-') << endl;

    for (int i = 0; i <= 127; ++i) {
        cout << left << setw(6) << i 
             << setw(6) << hex << uppercase << i << dec // 切换回十进制以便后续使用
             << setw(15);
        
        if (i < 32 || i == 127) {
            // 控制字符：显示描述
            cout << getControlCharDesc(i);
        } else if (i == 32) {
            cout << "(空格)";
        } else {
            // 可打印字符：直接输出
            cout << static_cast<char>(i);
        }
        cout << endl;
    }

    return 0;
}

```

## 2.5 数据类型的转换

### 隐式类型转换 Implicit Conversion

隐式转换由编译器自动完成，通常发生在赋值或运算过程中。其核心原则是：‌低精度类型向高精度类型转换‌，以避免数据丢失。

```c
// 整型转换为浮点型 
#include<iostream>
using namespace std;
int main(){
    int a = 5;
    double b = a;
    cout << b << endl; // b 的值为 5.0
    return 0;
}
```

```c
// 整型转换为字符型
#include<iostream>
using namespace std;
int main(){
    int a = 65;
    char b = a;
    cout << b << endl; // b 的值为 A
    return 0;
}
```

```c
// 整型转换为布尔型 非零即真，零即假
#include<iostream>
using namespace std;
int main(){
    int a = 0;
    bool b = a;
    cout << b << endl; // b 的值为 0
    return 0;
}
```

```c
// 浮点型转换为整型
#include<iostream>
using namespace std;
int main(){
    double d = 3.14;
    int b = d;
    cout << b << endl; // b 的值为 3
    return 0;
}
```

```c
// 浮点型转换为整型
#include<iostream>
using namespace std;
int main(){
    double a = 5 / 2;         // 结果 2
    double b = 5.0 / 2;       // 结果 2.5
    double c = 5 / 2.0;       // 结果 2.5
    double d = (double)5 / 2; // 结果 2.5
    cout << a << " " << b << " " << c << " " << d << endl;
    return 0;
}
```

```c
// 浮点型转换为布尔型
#include<iostream>
using namespace std;
int main(){
    double d = 3.14;
    bool b = d;
    cout << b << endl; // b 的值为 1
    return 0;
}
```

```c
// 字符型转换为整型
#include<iostream>
using namespace std;
int main(){
    char c = 'A';
    int b = c;
    cout << b << endl; // b 的值为 65
    return 0;
}
```

```c
// 布尔型转换为整型
#include<iostream>
using namespace std;
int main(){
    bool t = true;
    int b = t;
    cout << b << endl; // b 的值为 1
    return 0;
}
```

### 显式类型转换 Explicit Conversion

显式转换由程序员手动指定。C++ 提供了四种标准转换运算符，每种都有特定的适用场景和安全级别。

#### 静态转换 static_cast

```c
double d = 3.14;
int i = static_cast<int>(d); // i 变为 3

Base* basePtr = new Derived();
Derived* derivedPtr = static_cast<Derived*>(basePtr); // 需确保 basePtr 确实指向 Derived
```

#### 动态转换 dynamic_cast

```c
Base* basePtr = getPointer(); 
// 尝试安全转换
if (Derived* derivedPtr = dynamic_cast<Derived*>(basePtr)) {
    derivedPtr->specificMethod(); // 安全调用
} else {
    // 转换失败，处理错误
}
```

#### 常量性转换 const_cast

```c
void legacyFunc(char* str); // 旧接口不接受 const

const char* myStr = "Hello";
// 谨慎使用：仅当确定 legacyFunc 不会修改字符串，且原对象非真正 const 时
legacyFunc(const_cast<char*>(myStr)); 
```

#### 重解释转换 reinterpret_cast

```c
int i = 42;
int* p = &i;
// 将指针地址 reinterpret 为 long 类型整数
long addr = reinterpret_cast<long>(p);
```

#### C 风格强制转换 C-style Cast

C 风格强制转换是 C 语言遗留下来的一种类型转换方式，在 C++ 中依然被支持，但因其缺乏安全性和明确的语义，在现代 C++ 编程中通常‌不推荐使用‌。

```c
// 浮点型转换为整型
#include <iostream>
#include <string>
using namespace std;
int main() {
    double d = 3.14;
    int i = (int)d;      //传统写法
    int j = int(d);      //函数式写法
    cout << i << endl;
    cout << j << endl;
    
    return 0;
}
```

## 2.6 多层分支/循环结构

### 多层分支结构 Multi-level Branching Structure  

最佳实践建议  
条件互斥与顺序‌：确保条件之间逻辑清晰，范围判断时要注意顺序（如先判断 >=90 再判断 >=80），防止高分被低分条件提前捕获。  
保留 Else 兜底‌：始终建议保留 else 分支，用于处理意外输入或异常情况，提高程序的健壮性。  
避免过度嵌套‌：如果嵌套层数超过 3 层，代码可读性会大幅下降。此时应考虑提取函数或使用字典映射、策略模式等重构手段。  
利用短路特性‌：在复杂条件中，将计算成本低或更容易为假的条件放在前面，可以提升执行效率。  

```c
#include <iostream>

int main() {
    int score = 85;
    bool is_final_exam = true;

    // 1. 多分支链式结构 (if-else if-else)
    // 适用于范围判断或互斥条件
    if (score >= 90) {
        std::cout << "等级: A" << std::endl;
    } else if (score >= 60) {
        std::cout << "等级: B" << std::endl;
    } else {
        std::cout << "等级: C (不及格)" << std::endl;
    }

    // 2. 嵌套结构 (Nested If)
    // 适用于需要进一步细化条件的场景
    if (score >= 60) {
        std::cout << "恭喜，通过考试！" << std::endl;
        
        // 内层判断
        if (is_final_exam && score > 80) {
            std::cout << "并且获得优秀奖！" << std::endl;
        }
    } else {
        std::cout << "很遗憾，未通过。" << std::endl;
    }

    return 0;
}
```

```c
#include <iostream>

int main() {
    char grade = 'B';

    switch (grade) {
        case 'A':
            std::cout << "优秀" << std::endl;
            break; // 必须加 break，否则会继续执行下一个 case (fall-through)
        case 'B':
            std::cout << "良好" << std::endl;
            break;
        case 'C':
            std::cout << "及格" << std::endl;
            break;
        default: // 处理所有其他情况
            std::cout << "未知等级" << std::endl;
            break;
    }

    return 0;
}
```

### 多层循环结构 Multi-level Loop Structure‌

最佳实践建议
控制嵌套深度‌：建议嵌套层数不超过 3 层。过深的嵌套会导致代码难以阅读和维护，应考虑提取函数或使用递归。  
优化内层逻辑‌：内层循环执行次数最多，应将复杂的计算或判断尽量移至外层，或在内层中使用 break 提前终止无效搜索。  
变量命名清晰‌：多层循环中常用 i, j, k 作为索引，但在处理具体业务数据时，建议使用语义化名称（如 row, col 或 user, order）以提升可读性。  

```c
#include <iostream>
#include <vector>

int main() {
    // 示例：打印 3x3 矩阵坐标
    for (int i = 0; i < 3; ++i) {       // 外层循环：控制行
        for (int j = 0; j < 3; ++j) {   // 内层循环：控制列
            std::cout << "(" << i << "," << j << ") ";
        }
        std::cout << std::endl; // 每行结束后换行
    }

    return 0;
}
```

## 2.7 常用数学函数（绝对值函数、平方根函数、max 函数、min 函数）

```c
// Absolute Value Function 绝对值函数
#include <iostream>
#include <cmath>   // std::abs, std::fabs
#include <cstdlib> // std::abs (int overload in older standards)

int main() {
    int i = -42;
    double d = -3.14;
    
    // 整数绝对值
    std::cout << "Int abs: " << std::abs(i) << std::endl;      // 输出 42
    
    // 浮点数绝对值 (std::abs 会自动匹配重载，std::fabs 也可用)
    std::cout << "Double abs: " << std::abs(d) << std::endl;   // 输出 3.14
    std::cout << "Float fabs: " << std::fabs(-2.7f) << std::endl; // 输出 2.7

    return 0;
}
```

```c
// Square Root Function 平方根函数
#include <iostream>
#include <cmath>
#include <stdexcept>

int main() {
    double x = 45.35;
    double result = std::sqrt(x);
    std::cout << "Sqrt of " << x << " is " << result << std::endl; // 输出 ~6.73

    // 处理负数
    double neg_result = std::sqrt(-1.0);
    if (std::isnan(neg_result)) {
        std::cout << "Result is NaN (Not a Number)" << std::endl;
    }

    return 0;
}
```

```c
// 最大值与最小值函数
#include <iostream>
#include <algorithm>
#include <vector>
#include <string>

int main() {
    int a = 10, b = 20;
    
    // 1. 比较两个值
    int max_val = std::max(a, b); // 返回 20
    int min_val = std::min(a, b); // 返回 10
    
    // 2. 比较多个值 (C++11 起，使用初始化列表)
    int multi_max = std::max({1, 5, 3, 9, 2}); // 返回 9
    int multi_min = std::min({10, -1, 8, 3});  // 返回 -1
    
    // 3. 自定义比较规则 (例如按字符串长度比较)
    std::string s1 = "apple";
    std::string s2 = "hi";
    // 返回较短的字符串
    std::string shortest = std::min(s1, s2, [](const std::string& x, const std::string& y) {
        return x.size() < y.size();
    });
    std::cout << "Shortest: " << shortest << std::endl; // 输出 "hi"

    return 0;
}
```
