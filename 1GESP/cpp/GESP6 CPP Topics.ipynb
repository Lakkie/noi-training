## 6.1 树的定义，构造与遍历

### 链式存储

链式存储（二叉链表）‌  
用两个数组lc[]和rc[]分别记录每个节点的左、右孩子编号，0表示该位置无节点，适合任意普通二叉树的构造。

```c
int lc[105], rc[105]; // lc[i]存储节点i的左孩子编号
// 手动构造根为1、左孩子2、右孩子3的二叉树
lc[1] = 2; rc[1] = 3;
lc[2] = 0; rc[2] = 0; // 节点2是叶子
lc[3] = 0; rc[3] = 0; // 节点3是叶子
```

### 顺序存储

顺序存储（数组表示法）‌
仅适用于完全二叉树，节点从下标1开始连续存储，无需额外记录孩子编号，通过下标运算直接定位父子节点，是近年GESP6级编程题的高频考点。
必背下标公式：
任意节点i的父节点：i / 2
任意节点i的左孩子：2 * i
任意节点i的右孩子：2 * i + 1

### 前序遍历‌ Preorder Traversal

前序遍历‌：访问顺序为「根节点 → 左子树 → 右子树」，先处理根，再递归遍历左右子树。

```c
void preorder(TreeNode* root, vector<int>& res) {
    if (root == nullptr) return; // 终止条件：节点为空
    
    res.push_back(root->val);    // 1. 中：访问根节点
    preorder(root->left, res);   // 2. 左：递归左子树
    preorder(root->right, res);  // 3. 右：递归右子树
}
```

中序遍历‌：访问顺序为「左子树 → 根节点 → 右子树」，二叉排序树的中序遍历结果一定是有序序列。

后序遍历‌：访问顺序为「左子树 → 右子树 → 根节点」，适合先处理所有子节点再处理根的场景，比如删除整棵树。

层序遍历：从上到下、同一层从左到右依次访问节点，基于队列实现，完全二叉树的层序遍历序列就是数组顺序存储的结果。

```c
#include <iostream>
#include <vector>
#include <unordered_map>
using namespace std;

// 定义二叉树节点
struct TreeNode {
    int val;
    TreeNode *left;
    TreeNode *right;
    TreeNode(int x) : val(x), left(nullptr), right(nullptr) {}
};

class Solution {
private:
    unordered_map<int, int> indexMap; // 哈希表：值 -> 中序索引

public:
    TreeNode* buildTree(vector<int>& preorder, vector<int>& inorder) {
        int n = preorder.size();
        // 1. 构建哈希表，加速中序查找
        for (int i = 0; i < n; ++i) {
            indexMap[inorder[i]] = i;
        }
        // 2. 启动递归构建
        return myBuildTree(preorder, inorder, 0, n - 1, 0, n - 1);
    }

    // preL, preR: 前序遍历的左右边界
    // inL, inR:   中序遍历的左右边界
    TreeNode* myBuildTree(vector<int>& preorder, vector<int>& inorder, 
                          int preL, int preR, int inL, int inR) {
        // 终止条件：区间非法
        if (preL > preR) {
            return nullptr;
        }

        // 1. 前序的第一个节点是根
        int rootVal = preorder[preL];
        TreeNode* root = new TreeNode(rootVal);

        // 2. 在中序中找到根的位置
        int inRootIndex = indexMap[rootVal];

        // 3. 计算左子树的节点数量
        int leftSubtreeSize = inRootIndex - inL;

        // 4. 递归构建左子树
        // 左子树在前序中的范围：[preL+1, preL+leftSubtreeSize]
        // 左子树在中序中的范围：[inL, inRootIndex-1]
        root->left = myBuildTree(preorder, inorder, 
                                 preL + 1, preL + leftSubtreeSize, 
                                 inL, inRootIndex - 1);

        // 5. 递归构建右子树
        // 右子树在前序中的范围：[preL+leftSubtreeSize+1, preR]
        // 右子树在中序中的范围：[inRootIndex+1, inR]
        root->right = myBuildTree(preorder, inorder, 
                                  preL + leftSubtreeSize + 1, preR, 
                                  inRootIndex + 1, inR);

        return root;
    }
};

// 辅助函数：前序遍历打印验证
void printPreorder(TreeNode* node) {
    if (!node) return;
    cout << node->val << " ";
    printPreorder(node->left);
    printPreorder(node->right);
}

int main() {
    // 示例数据
    vector<int> preorder = {3, 9, 20, 15, 7};
    vector<int> inorder = {9, 3, 15, 20, 7};

    Solution sol;
    TreeNode* root = sol.buildTree(preorder, inorder);

    cout << "还原后的前序遍历: ";
    printPreorder(root); // 应输出: 3 9 20 15 7
    cout << endl;

    return 0;
}
```

## 6.11 面向对象的思想

面向对象思想（Object Oriented，简称OO）是一种完全贴合人类自然思维的程序设计范式，主张直接将现实世界中的客观事物抽象为程序中的“对象”，用对象之间的交互来构建整个软件系统，彻底区别于传统“按步骤分解功能”的面向过程编程。

面向对象的核心哲学是：世界由无数个独立的对象组成，每个对象都有自己的内部状态和行为能力，不同对象之间通过消息传递完成协作，最终构成完整的系统。它不再围绕“功能步骤”拆分代码，而是直接映射现实世界的事物结构，让程序逻辑和真实世界的逻辑完全对齐，大幅降低复杂系统的开发和维护难度。

对象‌：现实世界中任何可被研究的事物，小到一个整数、大到一架飞机，都可以抽象为对象。它同时包含“状态数据”和“操作行为”两部分，是程序的基本组成单元。

类‌：对大量具有相同属性和行为的对象的抽象模板。比如“学生”是一个类，而具体的“张三同学”就是这个类实例化出来的一个对象。类是对象的“设计图”，对象是类的具体实例。

消息传递‌：对象之间交互的唯一方式，通过向目标对象发送消息，触发它执行对应的方法，完成特定操作，这是面向对象系统中对象协作的基础机制。

抽象‌：从大量相似事物中提取出共同的核心特征，忽略无关细节，最终形成通用类的过程。比如从“猫、狗、鸟”中抽象出“动物”类，提取出“进食、移动”等通用行为。

封装‌：把对象的属性数据和操作这些数据的方法捆绑成一个独立单元，通过访问控制权限（private、protected、public）隐藏内部实现细节，只对外暴露必要的安全接口。比如银行账户类会把“余额”设为私有属性，只开放“存款、取款”两个公共接口，避免外部代码随意篡改余额，从根源上提升数据安全性和系统稳定性。

继承‌：子类可以完全复用父类已有的属性和方法，同时还能扩展自己的新功能，甚至重写父类的旧方法。它完美适配现实世界的“is-a”关系，比如“猫是一种动物”，“猫”作为子类可以直接继承“动物”父类的“进食”方法，不用重复编写相同代码，大幅减少冗余，实现代码的层次化复用。

多态‌：同一个操作作用在不同的子类对象上，会产生完全不同的执行结果。它分为两种典型形式：重载（同一个类中同名方法通过不同参数实现不同逻辑）、重写（子类覆盖父类的同名方法）。比如“绘图”方法，调用“圆形对象”就画圆，调用“方形对象”就画方形，同一个接口可以适配不同场景，让程序的扩展性大幅提升。

## 6.12 类的创建

```c
#include <iostream>
#include <string>
using namespace std;

// 1. 类的定义
class Student {
private:
    // 成员变量（属性）：通常设为 private 以实现封装
    string name;
    int age;
    double score;

public:
    // 2. 构造函数：用于初始化对象
    // 默认构造函数
    Student() : name("Unknown"), age(0), score(0.0) {
        cout << "默认构造函数被调用" << endl;
    }

    // 带参构造函数：使用初始化列表高效赋值
    Student(string n, int a, double s) : name(n), age(a), score(s) {
        cout << "带参构造函数被调用: " << name << endl;
    }

    // 3. 成员函数（方法）：操作数据的行为
    void showInfo() const {
        cout << "姓名: " << name 
             << ", 年龄: " << age 
             << ", 成绩: " << score << endl;
    }

    // Setter 方法：提供受控的数据修改接口
    void setScore(double s) {
        if (s >= 0 && s <= 100) {
            score = s;
        } else {
            cout << "成绩无效！" << endl;
        }
    }

    // Getter 方法：获取私有数据
    double getScore() const {
        return score;
    }

    // 4. 析构函数：对象销毁时自动调用，用于清理资源
    ~Student() {
        cout << "析构函数被调用: " << name << endl;
    }
};

int main() {
    // 5. 对象的实例化
    
    // 方式一：栈上创建（自动管理内存）
    // 调用带参构造函数
    Student stu1("Alice", 18, 95.5);
    stu1.showInfo();

    // 方式二：堆上创建（需手动释放内存）
    // 使用 new 关键字，返回指针
    Student* stu2 = new Student("Bob", 20, 88.0);
    stu2->showInfo();
    
    // 修改数据
    stu2->setScore(92.0);
    cout << "Bob的新成绩: " << stu2->getScore() << endl;

    // 释放堆内存，触发析构函数
    delete stu2;

    // stu1 在 main 函数结束时自动销毁，触发析构函数
    return 0;
}
```

## 6.13 栈、队列、循环队列

### 栈 Stack

```c
#include <iostream>
#include <stack>
using namespace std;

int main() {
    stack<int> s;

    // 1. 入栈 (Push)
    s.push(10);
    s.push(20);
    s.push(30);

    // 2. 查看栈顶元素 (Top) - 不删除
    cout << "栈顶元素: " << s.top() << endl; // 输出 30

    // 3. 出栈 (Pop) - 删除栈顶元素
    s.pop(); 
    cout << "出栈后栈顶: " << s.top() << endl; // 输出 20

    // 4. 判断是否为空 & 获取大小
    if (!s.empty()) {
        cout << "栈的大小: " << s.size() << endl; // 输出 2
    }

    return 0;
}
```

```c
// 数组模拟栈
const int MAX_SIZE = 100;
int stackArr[MAX_SIZE];
int topIndex = -1; // 栈顶指针，-1表示空

void push(int val) {
    if (topIndex >= MAX_SIZE - 1) {
        cout << "栈满！" << endl;
        return;
    }
    stackArr[++topIndex] = val;
}

void pop() {
    if (topIndex == -1) {
        cout << "栈空！" << endl;
        return;
    }
    topIndex--;
}

int top() {
    if (topIndex == -1) return -1;
    return stackArr[topIndex];
}
```

### 队列 Queue

```c
#include <iostream>
#include <queue>
using namespace std;

int main() {
    queue<int> q;

    // 1. 入队 (Push)
    q.push(10);
    q.push(20);
    q.push(30);

    // 2. 查看队头元素 (Front) - 不删除
    cout << "队头元素: " << q.front() << endl; // 输出 10
    
    // 3. 查看队尾元素 (Back)
    cout << "队尾元素: " << q.back() << endl;  // 输出 30

    // 4. 出队 (Pop) - 删除队头元素
    q.pop();
    cout << "出队后队头: " << q.front() << endl; // 输出 20

    // 5. 判断是否为空 & 获取大小
    if (!q.empty()) {
        cout << "队列大小: " << q.size() << endl; // 输出 2
    }

    return 0;
}
```

### 循环队列 Circular Queue

```c
#include <iostream>
using namespace std;

class CircularQueue {
private:
    int* data;      // 动态数组
    int front;      // 指向队头元素
    int rear;       // 指向队尾元素的下一个位置
    int capacity;   // 数组总容量（实际可用空间为 capacity - 1）

public:
    // 构造函数：初始化大小为 k 的循环队列
    // 注意：为了区分空和满，我们申请 k+1 的空间，或者约定容量为 k+1
    CircularQueue(int k) {
        capacity = k + 1; 
        data = new int[capacity];
        front = 0;
        rear = 0;
    }

    // 析构函数：释放内存
    ~CircularQueue() {
        delete[] data;
    }

    // 判断队列是否为空
    bool isEmpty() {
        return front == rear;
    }

    // 判断队列是否已满
    bool isFull() {
        return (rear + 1) % capacity == front;
    }

    // 入队操作
    bool enQueue(int value) {
        if (isFull()) {
            return false; // 队列满，入队失败
        }
        data[rear] = value;
        // 核心：通过取模实现指针循环移动
        rear = (rear + 1) % capacity;
        return true;
    }

    // 出队操作
    bool deQueue() {
        if (isEmpty()) {
            return false; // 队列空，出队失败
        }
        // 核心：通过取模实现指针循环移动
        front = (front + 1) % capacity;
        return true;
    }

    // 获取队头元素
    int getFront() {
        if (isEmpty()) {
            return -1; // 或抛出异常
        }
        return data[front];
    }

    // 获取队尾元素
    int getRear() {
        if (isEmpty()) {
            return -1;
        }
        // rear 指向的是下一个空位，所以队尾元素在 rear 的前一位
        // 如果 rear 为 0，前一位是 capacity - 1
        int index = (rear - 1 + capacity) % capacity;
        return data[index];
    }
    
    // 获取当前元素个数
    int size() {
        return (rear - front + capacity) % capacity;
    }
};

// 测试主函数
int main() {
    CircularQueue cq(3); // 创建一个最多存3个元素的队列

    cq.enQueue(1);
    cq.enQueue(2);
    cq.enQueue(3);
    
    cout << "队头: " << cq.getFront() << endl; // 1
    cout << "队尾: " << cq.getRear() << endl;   // 3
    cout << "是否满: " << cq.isFull() << endl;   // 1 (true)

    cq.deQueue(); // 弹出 1
    cq.enQueue(4); // 插入 4，此时利用了之前弹出的空间
    
    cout << "新队头: " << cq.getFront() << endl; // 2
    cout << "新队尾: " << cq.getRear() << endl;   // 4
    cout << "当前大小: " << cq.size() << endl;     // 3

    return 0;
}
```
