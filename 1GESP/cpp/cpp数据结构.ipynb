## 链表‌  
实现方式‌：数据元素分散存储，通过指针串联各个节点。
核心特性‌：插入删除操作无需移动大量元素，仅修改指针即可，不支持随机访问。

‌链表 (Linked List)‌：  
‌特点‌：由节点组成，每个节点包含数据和指向下一个节点的指针，内存非连续。  
‌优势‌：插入和删除操作效率高（O(1)，前提是已知位置），动态调整大小灵活。  
‌劣势‌：不支持随机访问，查找指定位置元素需遍历，速度较慢（O(n)）；额外消耗内存存储指针。  

# 栈和队列

‌## 栈 (Stack)‌
‌原则‌：后进先出（LIFO, Last In First Out）。  
‌应用‌：函数调用堆栈、表达式求值、括号匹配等。  
‌C++实现‌：可使用 std::stack 适配器，底层通常基于 std::deque 或 std::vector。  

‌## 队列 (Queue)‌
‌原则‌：先进先出（FIFO, First In First Out）。
‌应用‌：任务调度、广度优先搜索（BFS）、缓冲区管理。
‌C++实现‌：使用 std::queue 容器适配器。

# 数组和特殊矩阵

‌数组 (Array)‌：  
‌特点‌：内存连续分配，大小固定。  
‌优势‌：支持随机访问，通过索引获取元素速度快（O(1)）。  
‌劣势‌：插入和删除元素需要移动大量数据，效率较低；大小不可动态扩展。

# 树 Tree

特点‌：层次结构，每个节点可能有多个子节点。
‌二叉树‌：每个节点最多有两个子节点（左子和右子）。  
‌常见变体‌：二叉搜索树（BST）、平衡二叉树（AVL）、红黑树（Red-Black Tree，std::map 和 std::set 的底层实现）、堆（Heap，优先队列底层）。  

# 图 Graph

组成‌：由顶点（Vertex）和边（Edge）组成。  
‌应用‌：社交网络关系、地图导航最短路径、网络拓扑等。  
‌存储方式‌：邻接矩阵、邻接表。
