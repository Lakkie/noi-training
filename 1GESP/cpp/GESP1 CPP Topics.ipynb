## 1.3 变量的定义与使用

在 C++ 中，变量是程序存储和操作数据的基本单元。它本质上是一块具有名称的内存空间，用于保存程序运行过程中可能变化的值。以下是关于 C++ 变量定义、初始化、赋值及命名规则的详细指南。

### 变量的定义

```c
int age;        // 定义一个整型变量
double salary;  // 定义一个双精度浮点型变量
char grade;     // 定义一个字符型变量
```

### 变量的初始化

初始化是指在定义变量的同时赋予其初始值。未初始化的局部变量包含的是内存中的垃圾值，直接使用可能导致不可预知的错误，因此建议始终初始化变量。

```c
//拷贝初始化（使用 =）
int num = 10;
double pi = 3.14159;

//直接初始化（使用 ()）这种写法源自 C++ 早期标准，常用于调用构造函数。
int num(10);

//列表初始化/统一初始化（使用 {}）推荐方式‌。从 C++11 开始引入，适用于所有类型，且能防止窄化转换（例如不允许将 double 隐式转换为 int 而丢失精度）。
int num{10};
char ch{'A'};
```

### 变量的赋值

赋值是指改变已定义变量的值。赋值可以发生多次，而初始化仅在定义时发生一次。

```c
int num;         // 定义
num = 10;        // 第一次赋值
num = 20;        // 第二次赋值，此时 num 的值变为 20
num = 30;        // 第三次赋值，此时 num 的值变为 30

//链式赋值 multiple assignment
int a, b, c;
a = b = c = 100; // 等价于 a=100; b=100; c=100;
```

### 变量的命名规则

变量名（标识符）必须遵循以下严格规则，否则会导致编译错误：

组成字符‌：只能包含字母（a-z, A-Z）、数字（0-9）和下划线（_）。  
首字符限制‌：必须以‌字母‌或‌下划线‌开头，‌不能以数字开头‌。  
合法：value, _temp, var1  
非法：123var, 9abc  
关键字冲突‌：不能使用 C++ 的保留关键字（如 int, class, return, if, while 等）。  
大小写敏感‌：C++ 区分大小写。Age, age, AGE 是三个完全不同的变量。  
特殊约定‌：  
避免以单个下划线开头后跟大写字母（如 _Value）或双下划线开头（如 __val），这些通常被编译器或标准库保留。  
建议使用具有语义的名称（如 studentCount 而非 sc），以提高代码可读性。  

## 1.4 基本数据类型（整型、浮点型、字符型、布尔型）

### 整型 Integer Types 

用于存储整数（没有小数部分的数字）。

| 类型 | 典型大小 (字节) | 说明 |
| --- | --- | --- |
| short | 2 | 短整型，范围较小，节省内存。 |
| int | 4 | 基本整型，最常用的整数类型。 |
| long long | 8 | 长长整型，用于存储非常大的整数。 |

注意‌：具体大小取决于编译器和操作系统，可以使用 sizeof() 运算符查询当前环境下的实际大小。

### 浮点型 Floating-Point Types 

用于存储带有小数部分的实数。

| 类型 | 典型大小 (字节) | 精度 | 说明 |
| --- | --- | --- | --- |
| float | 4 | 约 7 位有效数字 | 单精度浮点数。定义常量时需加后缀 f，如 3.14f。 |
| double | 8 | 约 15-16 位有效数字 | 双精度浮点数，C++ 中浮点字面量默认为此类型。 |
| long double | 8, 10, 12 或 16 | 更高 | 扩展精度浮点数，具体大小和精度依平台而定。 |

### 字符型 Character Type 

用于存储单个字符。

char‌：通常占用 ‌1 字节‌。它在底层实际上是一个小整数（ASCII 码值），因此既可以当作字符处理（如 'A'），也可以当作整数进行运算。  
signed char：有符号字符。  
unsigned char：无符号字符，常用于处理二进制数据或字节流。  
wchar_t / char16_t / char32_t：用于支持 Unicode 宽字符。  

### 布尔型 Boolean Type 

用于表示逻辑真或假。

bool‌：占用通常为 ‌1 字节‌。只有两个值：true（真，内部通常表示为 1）和 false（假，内部通常表示为 0）。

### 空类型 Void Type 

‌void‌：表示“无类型”或“无值”。  
函数返回值：表示函数不返回任何值（如 void func()）。  
指针：void* 是通用指针，可以指向任何类型的数据地址（但不能直接解引用，需转换）。  
不能定义 void 类型的变量。  

### 类型修饰符

signed‌：有符号（默认）。  
unsigned‌：无符号（仅适用于整型和字符型，不能用于浮点型）。  
const‌：常量，定义后值不可修改。  
volatile‌：易变变量，告诉编译器不要优化对该变量的访问（常用于嵌入式或多线程）。  

## 1.5 控制语句结构（顺序、循环、选择）

### 顺序结构 Sequential Structure

```c
#include<iostream>
using namespace std;
int main() {
    int a = 10;                     // 第一步：声明并初始化
    int b = 20;                     // 第二步：声明并初始化
    int sum = a + b;                // 第三步：计算
    cout << "Sum: " << sum << endl; // 第四步：输出
    return 0;                       // 第五步：返回
}
```

### 循环结构 Loop Structure

```c
// for循环 适用于已知循环次数或遍历序列（如列表、字符串、范围等）。
for (int i = 0; i < 5; i++) {
    cout << i << " ";
}

// C++11 范围 for 循环 (Range-based for)
int arr[] = {10, 20, 30};
for (int val : arr) {
    cout << val << " ";
}

```

```c
// while循环 适用于未知循环次数，只要条件为真 (True) 就持续执行。‌务必在循环体内修改条件变量‌，否则可能导致死循环。
int i = 1;
while (i <= 5) {
    cout << i << " ";
    i++; // 更新计数器，防止死循环
}
// 输出: 1 2 3 4 5
```

```c
// do-while 循环 ("直到型"循环) 先执行一次循环体，再判断条件。‌保证循环体至少执行一次‌。
int num;
do {
    cout << "Enter a positive number: ";
    cin >> num;
} while (num <= 0); // 如果输入负数，会再次提示
```

```c
// 循环控制语句 Jump Statements
for (int i = 0; i < 10; i++) {
    if (i == 3) continue; // 跳过 3
    if (i == 7) break;    // 到 7 时彻底结束循环
    cout << i << " ";
}
// 输出: 0 1 2 4 5 6
```

| 语句 | 作用 | 示例场景 |
| --- | --- | --- |
| ‌break‌ | ‌立即终止‌当前循环，跳出循环体。 | 找到目标值后提前退出搜索循环。 |
| ‌continue‌ | ‌跳过本次‌循环剩余的代码，直接进入下一次条件判断。 | 处理数组时跳过无效数据（如负数）。 |
| ‌goto‌ | 无条件跳转到指定标签。 | ‌不推荐‌使用，容易破坏代码结构，仅在极少数底层优化或错误处理中使用。 |

### 选择结构 Selection Structure

```c
// 单分支 (if)‌：满足条件则执行，否则跳过。
bool flag = true;
if (flag == true) {
    cout << "Done" << endl;
}

// 双分支 (if-else)‌：满足条件执行 A，不满足执行 B。
int score = 85;
if (score >= 60) {
    cout << "Pass" << endl;
} else {
    cout << "Fail" << endl;
}

// 多分支 (if-elif-else)‌：多个条件依次判断，执行第一个满足条件的分支。
if (score >= 90) {
    cout << "Excellent" << endl;
} else if (score >= 60) {
    cout << "Pass" << endl;
} else {
    cout << "Fail" << endl;
}
```

## 1.6 基本运算（算术运算、关系运算、逻辑运算）

### 算术运算 Arithmetic operations

| 运算符 | 含义 | 示例 | 说明 |
| --- | --- | --- | --- |
| + | 加法 | a + b | 也可用于字符串拼接或指针偏移 |
| - | 减法 | a - b |  |
| * | 乘法 | a * b |  |
| / | 除法 | a / b | ‌注意‌：整数相除结果仍为整数（截断小数） |
| % | 取模（求余） | a % b | 仅适用于整数，结果为余数 |
| ++ | 自增 | ++a 或 a++ | 前缀先加后用，后缀先用后加 |  |
| -- | 自减 | --a 或 a-- | 同上 |

```c
#include <iostream>
using namespace std;

int main() {
    int a = 10, b = 3;
    
    cout << "加法: " << a + b << endl;   // 13
    cout << "除法(整除): " << a / b << endl; // 3 (小数部分被丢弃)
    cout << "取模: " << a % b << endl;   // 1 (10除以3余1)
    
    double x = 10.0, y = 3.0;
    cout << "除法(浮点): " << x / y << endl; // 3.33333
    
    // 自增演示
    int i = 5;
    cout << "后置++: " << i++ << endl; // 输出 5，然后 i 变为 6
    cout << "前置++: " << ++i << endl; // i 先变为 7，然后输出 7
    
    return 0;
}
```

### 关系运算 Relational operations

| 运算符 | 含义 | 示例 | 结果 |
| --- | --- | --- | --- |
| > | 大于 | 5 > 3 | true |
| < | 小于 | 5 < 3 | false |
| >= | 大于等于 | 5 >= 5 | true |
| <= | 小于等于 | 5 <= 3 | false |
| == | 等于 | 5 == 5 | true |
| != | 不等于 | 5 != 3 | true |

```c
#include <iostream>
using namespace std;

int main() {
    int a = 10, b = 20;
    
    cout << boolalpha; // 设置输出 true/false 而不是 1/0
    
    cout << "a > b: " << (a > b) << endl;   // false
    cout << "a != b: " << (a != b) << endl;  // true
    cout << "a == 10: " << (a == 10) << endl; // true
    
    // 关系运算结果参与算术运算
    cout << "(a < b) + 5: " << (a < b) + 5 << endl; // true(1) + 5 = 6
    
    return 0;
}

```

### 逻辑运算 Logical operations

| 运算符 | 名称 | 规则 | 示例 |
| --- | --- | --- | --- |
| ! | 逻辑非 | 真变假，假变真 | !true → false |
| && | 逻辑与 | 两者都为真，结果才为真 | true && false → false |
| \|\| | 逻辑或 | 只要有一个为真，结果即为真 | true \|\| false → true |

短路求值 Short-circuit Evaluation：  
&&‌：如果左侧表达式为 false，右侧表达式‌不会执行‌（因为结果已确定为假）。  
||‌：如果左侧表达式为 true，右侧表达式‌不会执行‌（因为结果已确定为真）。  

```c
#include <iostream>
using namespace std;

int main() {
    int age = 25;
    bool hasID = true;
    
    // 逻辑与：必须同时满足
    if (age >= 18 && hasID) {
        cout << "允许进入" << endl;
    }
    
    // 逻辑或：满足其一即可
    bool isWeekend = true;
    bool isHoliday = false;
    if (isWeekend || isHoliday) {
        cout << "今天休息" << endl;
    }
    
    // 逻辑非：取反
    bool isEmpty = false;
    if (!isEmpty) {
        cout << "容器不为空" << endl;
    }
    
    // 短路求值演示
    int x = 0;
    // 左侧 false，右侧 ++x 不执行，x 仍为 0
    if (false && (++x > 0)) { } 
    cout << "x after short-circuit AND: " << x << endl; // 0
    
    // 左侧 true，右侧 ++x 不执行，x 仍为 0
    if (true || (++x > 0)) { }
    cout << "x after short-circuit OR: " << x << endl;  // 0
    
    return 0;
}
```

## 1.7 输入输出语句

### 标准输入输出 iostream cin / cout

这是 C++ 最推荐的 I/O 方式，类型安全且易于扩展。

头文件‌：#include \<iostream>  
命名空间‌：通常使用 using namespace std; 或显式指定 std::cin / std::cout。  
运算符‌：\>> (提取运算符)：用于输入，<< (插入运算符)：用于输出。   
‌注意‌：cin \>> 读取字符串时，遇到空格、制表符或换行符会停止。如果需要读取包含空格的整行文本，应使用 getline(cin, name)。  

```c
#include <iostream>
using namespace std;

int main() {
    int age;
    double height;
    string name;

    // 输出提示
    cout << "请输入姓名: ";
    cin >> name; // 读取字符串（遇到空格停止）

    cout << "请输入年龄和身高(用空格隔开): ";
    cin >> age >> height; // 连续读取多个变量

    // 输出结果
    cout << "姓名: " << name << endl; // endl 表示换行并刷新缓冲区
    cout << "年龄: " << age << ", 身高: " << height << "米" << endl;

    return 0;
}

```

### 格式化输出 iomanip

当需要控制输出的宽度、精度、对齐方式时，需引入 \<iomanip> 头文件。  
常用的操纵符包括  
setw（设置域宽）  
setprecision（设置精度）  
fixed（固定点表示法）  
left/right（对齐）  

```c
#include <iostream>
#include <iomanip> // 必须包含此头文件
using namespace std;

int main() {
    double pi = 3.1415926535;
    int num = 42;

    // 1. 设置浮点数精度
    cout << "默认精度: " << pi << endl;       // 可能输出 3.14159
    cout << fixed << setprecision(2);          // 固定小数点后2位
    cout << "固定2位小数: " << pi << endl;     // 输出 3.14

    // 2. 设置宽度和对齐
    cout << setw(10) << left << "Item"         // 左对齐，占10位
         << setw(10) << right << "Price" << endl; // 右对齐，占10位
    
    cout << setw(10) << left << "Apple"
         << setw(10) << right << "$1.20" << endl;

    // 3. 设置填充字符
    cout << setfill('*');                      // 将空白处填充为 *
    cout << setw(10) << right << num << endl;  // 输出 &zwnj;********&zwnj;42

    return 0;
}
```

### C 风格输入输出 cstdio printf / scanf

C++ 完全兼容 C 语言的 I/O 函数。在某些对性能要求极高或需要复杂格式控制（如特定进制、前导零）的场景下，printf 可能更简洁。  
‌头文件‌：#include \<cstdio> (或 <stdio.h>)

```c
#include <cstdio>

int main() {
    int a = 10;
    float b = 3.14f;

    // 格式化输出
    // %d: 整数, %f: 浮点数, %.2f: 保留两位小数, %5d: 占5位宽度右对齐
    printf("整数: %d, 浮点数: %.2f\n", a, b);
    printf("右对齐演示: [%5d]\n", a); // 输出: [   10]

    // 输入示例
    int x;
    printf("请输入一个整数: ");
    scanf("%d", &x); // 注意：必须使用取地址符 &
    printf("你输入了: %d\n", x);

    return 0;
}
```

### 文件输入输出 fstream

如果需要从文件读取数据或将结果保存到文件，需使用 \<fstream> 库中的 ifstream（输入文件流）和 ofstream（输出文件流）。

```c
#include <iostream>
#include <fstream>
#include <string>
using namespace std;

int main() {
    // --- 写入文件 ---
    ofstream outFile("test.txt");
    if (outFile.is_open()) {
        outFile << "Hello, File!" << endl;
        outFile << 12345 << endl;
        outFile.close(); // 关闭文件
        cout << "文件写入成功。" << endl;
    } else {
        cerr << "无法打开文件进行写入。" << endl;
    }

    // --- 读取文件 ---
    ifstream inFile("test.txt");
    if (inFile.is_open()) {
        string line;
        while (getline(inFile, line)) { // 逐行读取
            cout << "读取内容: " << line << endl;
        }
        inFile.close();
    } else {
        cerr << "无法打开文件进行读取。" << endl;
    }

    return 0;
}
```

| 特性 | cin/cout | printf/scanf |
| --- | --- | --- |
| ‌类型安全‌ | ✅ 高（自动推导类型） | ❌ 低（需手动匹配格式符，易出错） |
| ‌可扩展性‌ | ✅ 支持自定义类重载 | ❌ 仅支持基本类型 |
| ‌性能‌ | 略慢（可通过 ios::sync_with_stdio(false) 优化） | 较快 |
| ‌格式化能力‌ | 需配合 <iomanip>，较繁琐 | 强大且简洁（如 %05d） |
| ‌推荐场景‌ | ‌绝大多数 C++ 场景‌ | 嵌入式、高性能日志、复杂格式输出 |

最佳实践‌：

日常开发优先使用 cin/cout。  
若需大量输入输出且担心性能，可在 main 函数开头添加 ios::sync_with_stdio(false); cin.tie(NULL); 以加速 cin/cout。  
涉及文件操作时，使用 RAII 风格的 fstream 对象，或利用作用域自动关闭文件。  

```c
#include<iostream>
using namespace std;
int main(){
    // 大量输入输出时，性能优化
    ios::sync_with_stdio(false);
    cin.tie(NULL);

    // 暂时忽略的函数体

    return 0;
}
```
