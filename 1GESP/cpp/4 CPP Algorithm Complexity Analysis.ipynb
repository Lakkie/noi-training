## 2 对数阶 O(logn)‌

每次循环将问题规模减半（如二分查找）。

```c
while (i < n) {
    i = i * 2; // i 呈指数增长，循环次数为 log₂n
}
```

## 3 线性阶 O(n)

单层循环，执行次数与 n 成正比。

```c
for (int i = 0; i < n; i++) {
    // 执行 n 次
}
```

## 5 平方阶 $O(n^2)‌$

双层嵌套循环，每层都与 n 相关。

```c
for (int i = 0; i < n; i++) {
    for (int j = 0; j < n; j++) {
        // 执行 n * n 次
    }
}
```
