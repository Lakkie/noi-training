### 数的整除性

#### 整除性

用Z表示整数集合，整数之间进行加法、减法和乘法运算，所得结果仍然为整数。但是两个整数相除不一定是整数。

定义1.1.1 设$a,b \in Z, b \neq 0$，  
如果存在$c \in Z$，使得$a = bc$，则称为b整除a，记成$b \mid a$，并且称b为a的一个因子，称a为b的倍数。  
如果不存在$c \in Z$，使得$a = bc$，则称b不整除a，记成$b \nmid a$。  

引理 1.1.2 如果$x \mid y$，均指$x,y \in Z$，且$x \neq 0$。  
若$a \mid b, b \mid c$，则$a \mid c$。（传递性）  
若$a \mid b, b \mid a$，则$a = \pm b$。（绝对值相等性）  
若$a \mid b, a \mid c$，则对任意$x,y \in Z, a \mid (bx+cy)$。（线性组合性）  

定义1.1.3 设a为任意实数，我们用$[a]$表示不超过a的最大整数，叫做a的整数部分，而$a-[a]$叫做实数a的小数部分，表示乘${a}$。于是每个实数a可惟一表示成$a=[a]+{a},[a] \in Z,0 \leq {a} \lt 1$。

定理1.1.4 （带余除法） 设$a,b \in Z$, 且$b \geq 1$，则存在惟一决定的整数q和r，使得a=qb+r,$0 \leq r < b$。

引理1.1.5 设S是一个非空整数集合，并且满足以下条件：  
若$a,b \in S$，则$a \pm b \in S$；  
若$a \in S$，则对任意整数c，均有$ac \in S$。  
则存在惟一的整数$d \geq 0$，使得S由d的所有倍数构成，即$s=dZ=\{da:a \in Z\}$。

定义1.1.6 设p是大于1的整数。如果p的正整数因子只有1和p，称p为素数（也叫做质数）。

定理1.1.7 素数有无限多个。

### 素数判定算法与优化

```c
//基础素数判定方法
bool isPrime(int n){
    if(n<=1) return false;
    for(int i=2;i<n;++i){
        if(n%i==0) return false;
    }
    return true;
}
```

```c
//优化方案：平方根上限
bool isPrime(int n){
    if(n<=1) return false;
    for(int i=2;i*i<=n;++i){
        if(n%i==0) return false;
    }
    return true;
}
```

```c
//埃拉托斯特尼筛法
vector<bool> sieve(int max_num){
    vector<bool> is_prime(max_num+1,true);
    is_prime[0]=is_prime[1]=false;
    for(int i=2;i*i<=max_num;++i){
        if(is_prime[i]){
            for(int j=i*i;j<=max_num;j+=i){
                is_prime[j]=false;
            }
        } 
    }
    return is_prime;
}
```

### 最大公约数 Greatest Common Divisor

最大公约数是指能够同时整除两个或多个整数的最大正整数。

欧几里得算法，gcd(a, b) = gcd(b, a mod b)

```c
//递归
int gcd(int a,int b){
    if(b==0) return a;
    return gcd(b,a%b);
}
```

```c
//迭代
int gcd(int a,int b){
    while(b!=0){
        int t=b;
        b=a%b;
        a=t;
    }
    return a;
}
```

### 最小公倍数 Least Common Multiple‌

lcm(a, b) = |a * b| / gcd(a, b)

```c
int gcd(int a,int b){
    if(b==0) return a;
    return gcd(b,a%b);
}
int lcm(int a,int b){
    return a/gcd(a,b)*b;//先除后乘避免溢出
}
```

### 模运算与同余方程

(a + b) mod m = (a mod m + b mod m)

(a * b) mod m = (a mod m * b mod m)

```c
//快速幂取模
int fasrPow(int a,int b,int m){
    int r=1;
    a=a%m;
    while(b>0){
        if(b%2==1){
            r=(r*a)%m;
        }
        a=(a*a)%m;
        b=b/2;
    }
    return r;
}
```

```c
//线性同余方程
int extended_gcd(int a,int b,int &x,int &y){
    if(b==0){
        x=1;
        y=0;
        return a;
    }
    int gcd=extended_gcd(b,a%b,y,x);
    y=y-a/b*x;
    return gcd;
}
bool solve_congruence(int a,int b,int m,int &x){
    int x0,y0;
    int d=extended_gcd(a,m,x0,y0);
    if(b%d!=0) return false;
    x=(x0*(b/d))%m;
    if(x<0) x+=m;
    return true;
}
```

## 5.2 数组模拟高精度加法、减法、乘法、除法

在 C++ 中，内置的整数类型（如 int, long long）都有固定的表示范围。当数值超过10^18或需要极高精度时，必须使用‌高精度运算（Big Integer Arithmetic）‌。其核心思想是‌用数组模拟竖式运算‌，将大数的每一位拆分存储，逐位进行计算并处理进位或借位。

### 数据存储与结构

为了简化进位和借位的处理，高精度数通常采用‌倒序存储‌（Little-Endian）：
‌数组下标0‌存储个位。  
‌数组下标1‌存储十位。  
...  
‌数组下标k‌存储10^k位。  
这样，当最高位产生进位时，只需在数组末尾追加元素，无需移动整个数组，效率更高。

```c
#include <iostream>
#include <vector>
#include <string>
#include <algorithm>
#include <cstring>

using namespace std;

// 定义最大位数，可根据题目需求调整
const int MAXN = 1005;

struct BigInt {
    int d[MAXN]; // 存储每一位数字，d[0]是个位
    int len;     // 有效位数

    BigInt() {
        memset(d, 0, sizeof(d));
        len = 1;
    }

    // 从字符串初始化
    void init(string s) {
        memset(d, 0, sizeof(d));
        len = s.size();
        for (int i = 0; i < len; ++i) {
            // 倒序存储：字符串最后一位(个位)存入d[0]
            d[i] = s[len - 1 - i] - '0';
        }
        // 处理前导零，例如 "00123" -> len=3, d=[3,2,1]
        while (len > 1 && d[len - 1] == 0) {
            len--;
        }
    }

    // 输出
    void print() const {
        for (int i = len - 1; i >= 0; --i) {
            cout << d[i];
        }
        cout << endl;
    }
};
```

### 高精度加法

```c
BigInt add(const BigInt& A, const BigInt& B) {
    BigInt C;
    int carry = 0;
    // 遍历较长的位数
    int maxLen = max(A.len, B.len);
    
    for (int i = 0; i < maxLen || carry; ++i) {
        if (i < A.len) carry += A.d[i];
        if (i < B.len) carry += B.d[i];
        
        C.d[C.len++] = carry % 10; // 当前位
        carry /= 10;               // 更新进位
    }
    
    // 如果最后还有进位，len已经在上一步++中包含了
    // 注意：如果结果是0，len至少为1
    if (C.len == 0) C.len = 1; 
    
    return C;
}
```

### 高精度减法

```c
BigInt subtract(const BigInt& A, const BigInt& B) {
    BigInt C;
    int borrow = 0;
    
    for (int i = 0; i < A.len; ++i) {
        int temp = A.d[i] - borrow;
        if (i < B.len) temp -= B.d[i];
        
        if (temp < 0) {
            temp += 10;
            borrow = 1;
        } else {
            borrow = 0;
        }
        C.d[C.len++] = temp;
    }
    
    // 去除前导零
    while (C.len > 1 && C.d[C.len - 1] == 0) {
        C.len--;
    }
    
    return C;
}

// 辅助函数：比较两个高精度数的大小 (A >= B 返回 true)
bool compare(const BigInt& A, const BigInt& B) {
    if (A.len != B.len) return A.len > B.len;
    for (int i = A.len - 1; i >= 0; --i) {
        if (A.d[i] != B.d[i]) return A.d[i] > B.d[i];
    }
    return true; // 相等也算 A >= B
}

```

### 高精度乘法

```c
// 高精乘低精
BigInt multiplyLow(const BigInt& A, int b) {
    BigInt C;
    long long carry = 0; // 使用 long long 防止中间溢出
    
    for (int i = 0; i < A.len || carry; ++i) {
        if (i < A.len) carry += (long long)A.d[i] * b;
        C.d[C.len++] = carry % 10;
        carry /= 10;
    }
    
    while (C.len > 1 && C.d[C.len - 1] == 0) C.len--;
    return C;
}
```

```c
// 高精乘高精
BigInt multiplyHigh(const BigInt& A, const BigInt& B) {
    BigInt C;
    C.len = A.len + B.len; // 结果最大位数
    
    for (int i = 0; i < A.len; ++i) {
        for (int j = 0; j < B.len; ++j) {
            C.d[i + j] += A.d[i] * B.d[j];
        }
    }
    
    // 统一处理进位
    int carry = 0;
    for (int i = 0; i < C.len; ++i) {
        C.d[i] += carry;
        carry = C.d[i] / 10;
        C.d[i] %= 10;
    }
    
    // 处理最高位可能的剩余进位
    while (carry) {
        C.d[C.len++] = carry % 10;
        carry /= 10;
    }
    
    // 去除前导零
    while (C.len > 1 && C.d[C.len - 1] == 0) {
        C.len--;
    }
    
    return C;
}
```

### 高精度除法

```c
// 高精除低精
BigInt divideLow(const BigInt& A, int b, int& remainder) {
    BigInt C;
    C.len = A.len;
    remainder = 0;
    
    // 从最高位开始除
    for (int i = A.len - 1; i >= 0; --i) {
        remainder = remainder * 10 + A.d[i];
        C.d[i] = remainder / b;
        remainder %= b;
    }
    
    // 去除前导零
    while (C.len > 1 && C.d[C.len - 1] == 0) {
        C.len--;
    }
    
    return C;
}
```

```c
// 高精除高精
BigInt divide(const BigInt& A, const BigInt& B, BigInt& remainder) {
    // 特殊情况处理
    if (B.compare(BigInt("0")) == 0) {
        cerr << "Error: Division by zero!" << endl;
        exit(1);
    }
    
    // 初始化余数为 A
    remainder = A;
    BigInt quotient("0");
    quotient.d.clear();
    quotient.d.resize(A.size(), 0); // 商的最大位数不会超过被除数位数
    
    if (A.compare(B) < 0) {
        // A < B, 商为0, 余数为A
        return BigInt("0");
    }
    
    // 核心算法：从高位到低位确定商的每一位
    // 为了提高效率，我们不逐位试商，而是利用位移
    
    int diff = A.size() - B.size();
    
    // 临时变量用于计算
    BigInt currentRem = A;
    BigInt q(A.size(), 0); // 初始化商，全0
    
    for (int i = diff; i >= 0; --i) {
        // 构造 B * 10^i
        BigInt shiftedB = B.shiftLeft(i);
        
        // 只要 currentRem >= shiftedB，就减去 shiftedB，并在商的第 i 位加 1
        // 注意：这里可以用循环减法，但因为 shiftedB 已经很大，循环次数很少（最多9次）
        while (currentRem.compare(shiftedB) >= 0) {
            currentRem.subtract(shiftedB);
            q.d[i]++;
        }
    }
    
    remainder = currentRem;
    q.removeLeadingZeros();
    return q;
}
```

## 5.3 单链表、双链表、循环链表

在 C++ 中，单链表、双链表和循环链表是三种基础的线性数据结构。它们的核心区别在于‌节点指针的指向方式‌以及‌遍历的逻辑‌。

### 单链表 Singly Linked List

单链表是线性数据结构最基础的形式。节点只包含数据和指向下一个节点的指针。C++ STL 中的 std::forward_list 就是单向链表。

插入‌：新节点的 next 指向原位置的下一个节点，原位置节点的 next 指向新节点。  
删除‌：找到待删节点的前驱节点 prev，令 prev->next = target->next，然后释放 `target。  

```c
struct SinglyNode {
    int data;
    SinglyNode* next;
    SinglyNode(int val) : data(val), next(nullptr) {}
};

// 头部插入示例
void pushFront(SinglyNode*& head, int val) {
    SinglyNode* newNode = new SinglyNode(val);
    newNode->next = head;
    head = newNode;
}
```

### 双链表 Doubly Linked List

双链表每个节点包含前驱和后继指针，支持双向遍历。C++ STL 中的 std::list 就是双向链表。

插入‌：需要修改四个指针（新节点的 prev/next，前驱节点的 next，后继节点的 prev）。  
优势‌：删除当前节点时，不需要从头遍历寻找前驱节点，直接通过 curr->prev 即可获取，效率极高。  

```c
struct DoublyNode {
    int data;
    DoublyNode* prev;
    DoublyNode* next;
    DoublyNode(int val) : data(val), prev(nullptr), next(nullptr) {}
};

// 在节点 curr 之后插入 newNode
void insertAfter(DoublyNode* curr, int val) {
    if (!curr) return;
    DoublyNode* newNode = new DoublyNode(val);
    
    newNode->next = curr->next;
    newNode->prev = curr;
    
    if (curr->next) {
        curr->next->prev = newNode;
    }
    curr->next = newNode;
}
```

### 循环链表 Circular Linked List

循环链表尾节点的 next 不再指向 nullptr，而是指向头节点（Head），形成环。可以是单向循环或双向循环。STL 没有直接的循环链表容器。如果需要循环行为，通常使用 std::list 或 std::vector，并通过迭代器取模或手动重置迭代器到 begin() 来模拟循环逻辑。  

‌判空/判尾‌：遍历结束的条件不再是 ptr == nullptr，而是 ptr->next == head 或 ptr == head。  
‌遍历方法‌必须使用 do-while 循环，以确保至少执行一次并正确判断回到起点的时刻。  

```c
struct CircularNode {
    int data;
    CircularNode* next;
    CircularNode(int val) : data(val), next(nullptr) {}
};

// 遍历循环链表
void printCircular(CircularNode* head) {
    if (!head) return;
    
    CircularNode* temp = head;
    do {
        std::cout << temp->data << " ";
        temp = temp->next;
    } while (temp != head); // 当再次回到头节点时停止
    std::cout << std::endl;
}

// 尾部插入
void append(CircularNode*& head, int val) {
    CircularNode* newNode = new CircularNode(val);
    if (!head) {
        head = newNode;
        head->next = head; // 自环
        return;
    }
    
    // 找到尾节点（其 next 指向 head）
    CircularNode* tail = head;
    while (tail->next != head) {
        tail = tail->next;
    }
    
    tail->next = newNode;
    newNode->next = head;
}
```

## 5.4 辗转相除法 Euclidean Algorithm

辗转相除法‌（也称欧几里得算法）是用于计算两个正整数最大公约数的经典算法。该算法以其高效性和简洁性著称，是数论和计算机科学中的基础算法之一。

```c
#include <iostream>
#include <cmath> // 用于 std::abs

// 迭代实现
int gcd_iterative(int a, int b) {
    // 1. 处理负数：最大公约数定义为正整数，先取绝对值
    a = std::abs(a);
    b = std::abs(b);
    
    // 2. 核心逻辑：辗转相除
    while (b != 0) {
        int temp = b;
        b = a % b;
        a = temp;
    }
    
    return a;
}

int main() {
    int x = 48, y = -18;
    std::cout << "GCD(" << x << ", " << y << ") = " << gcd_iterative(x, y) << std::endl;
    return 0;
}
```

```c
#include <iostream>
#include <cmath>

// 递归写法代码最短，逻辑直观，但在处理极大数值或深层递归时可能导致‌栈溢出‌。仅在确定输入范围较小或开启编译器尾递归优化时使用。
int gcd_recursive(int a, int b) {
    a = std::abs(a);
    b = std::abs(b);
    
    if (b == 0) {
        return a;
    }
    return gcd_recursive(b, a % b);
}

```

## 5.5 素数表的埃氏筛法和线性筛法

### 埃氏筛法 Eratosthenes Sieve

```c
//从小到大遍历每个数，如果当前数i未被标记为合数，则它是一个质数。随后，将该质数i的所有倍数（2i,3i,…）标记为合数。
void eratosthenes_sieve(int n, vector<bool>& is_prime) {
    fill(is_prime.begin(), is_prime.end(), true);
    is_prime[0] = is_prime[1] = false;
    for (int i = 2; i * i <= n; ++i) {
        if (is_prime[i]) {
            for (int j = i * i; j <= n; j += i) {
                is_prime[j] = false;
            }
        }
    }
}
```

### 线性筛法 Linear Sieve / Euler Sieve

```c
//每个合数只被其最小质因子筛掉一次
vector<int> primes;
vector<bool> st; // st[i] == true 表示 i 是合数

void linear_sieve(int n) {
    st.resize(n + 1, false);
    for (int i = 2; i <= n; ++i) {
        if (!st[i]) {
            primes.push_back(i);
        }
        for (int j = 0; j < primes.size() && i * primes[j] <= n; ++j) {
            st[i * primes[j]] = true;
            if (i % primes[j] == 0) {
                break; // 关键：保证每个合数只被最小质因子筛掉
            }
        }
    }
}
```

## 5.6 唯一分解定理 Unique Factorization Theorem

唯一分解定理（也叫算术基本定理 Fundamental Theorem of Arithmetic）的核心内容是：
任何一个大于 1 的正整数，都可以‌唯一地‌表示为若干个质数的乘积，且质因数按从小到大排列。
公式形式为：$𝑛=𝑝_1^{𝑒_1}×𝑝_2^{𝑒_2}×𝑝_3^{𝑒_3}×⋯×𝑝_𝑘^{𝑒_𝑘}
其中𝑝_1<𝑝_2<⋯<𝑝_𝑘都是质数，指数𝑒_𝑖是正整数。  
举个例子：$120=2^3×3^1×5^1$，这个分解方式是唯一的，不存在其他不同的质数组合能相乘得到120。

```c
// 质因数分解
#include <iostream>
using namespace std;

int main() {
    int n;
    cin >> n;
    
    for (int i = 2; i <= n; ++i) {
        if (n % i == 0) { // i 一定是质数（因为小于i的因子都已经被除干净了）
            int cnt = 0;
            while (n % i == 0) {
                cnt++;
                n /= i;
            }
            cout << i << "^" << cnt << " ";
        }
    }
    
    return 0;
}
```

```c
// 阶乘分解
#include <iostream>
#include <vector>
using namespace std;

vector<int> primes;
vector<bool> st;

// 线性筛预处理质数
void linear_sieve(int n) {
    st.resize(n + 1, false);
    for (int i = 2; i <= n; ++i) {
        if (!st[i]) primes.push_back(i);
        for (int j = 0; j < primes.size() && i * primes[j] <= n; ++j) {
            st[i * primes[j]] = true;
            if (i % primes[j] == 0) break;
        }
    }
}

int main() {
    int n;
    cin >> n;
    linear_sieve(n);
    
    // 遍历所有质数，计算在n!中的指数
    for (int p : primes) {
        int e = 0;
        int temp = n;
        while (temp > 0) {
            temp /= p;
            e += temp;
        }
        cout << p << "^" << e << " ";
    }
    
    return 0;
}
```

## 5.7 二分查找/二分答案（也称二分枚举法）

### 二分查找 Binary Search

二分查找（Binary Search）是 C++ 中一种极其高效的基础算法，其核心思想是通过不断将搜索区间对半分割，从而快速定位目标值。相比于线性查找的 O(n) 时间复杂度，二分查找的时间复杂度仅为O(logn)，在处理大规模有序数据时优势显著。

```c
#include <vector>
#include <iostream>

// 在升序数组中查找 target，返回索引，未找到返回 -1
int binarySearch(const std::vector<int>& nums, int target) {
    int left = 0;
    int right = nums.size() - 1; // 闭区间 [left, right]

    while (left <= right) {
        // 防止 (left + right) 整数溢出
        int mid = left + (right - left) / 2;

        if (nums[mid] == target) {
            return mid; // 找到目标
        } else if (nums[mid] < target) {
            left = mid + 1; // 目标在右半部分，排除 mid
        } else {
            right = mid - 1; // 目标在左半部分，排除 mid
        }
    }
    return -1; // 未找到
}

// 查找左边界（第一个等于 target 的位置）
int findLeftBound(const std::vector<int>& nums, int target) {
    int left = 0;
    int right = nums.size() - 1;
    int result = -1;

    while (left <= right) {
        int mid = left + (right - left) / 2;
        if (nums[mid] == target) {
            result = mid;   // 记录当前找到的位置
            right = mid - 1; // 继续向左搜索，看是否有更早的出现
        } else if (nums[mid] < target) {
            left = mid + 1;
        } else {
            right = mid - 1;
        }
    }
    return result;
}

// 查找右边界（最后一个等于 target 的位置）
int findRightBound(const std::vector<int>& nums, int target) {
    int left = 0;
    int right = nums.size() - 1;
    int result = -1;

    while (left <= right) {
        int mid = left + (right - left) / 2;
        if (nums[mid] == target) {
            result = mid;   // 记录当前找到的位置
            left = mid + 1; // 继续向右搜索，看是否有更晚的出现
        } else if (nums[mid] < target) {
            left = mid + 1;
        } else {
            right = mid - 1;
        }
    }
    return result;
}
```

```c
#include <algorithm>
#include <vector>
#include <iostream>

void stlExample() {
    std::vector<int> nums = {1, 3, 5, 7, 7, 7, 9, 11};
    int target = 7;

    // 1. 判断是否存在
    bool exists = std::binary_search(nums.begin(), nums.end(), target);
    
    // 2. 获取第一个 >= target 的位置 (即左边界)
    auto itLow = std::lower_bound(nums.begin(), nums.end(), target);
    
    // 3. 获取第一个 > target 的位置 (即右边界的下一个位置)
    auto itHigh = std::upper_bound(nums.begin(), nums.end(), target);

    if (itLow != nums.end() && *itLow == target) {
        int index = std::distance(nums.begin(), itLow);
        std::cout << "Found at index: " << index << std::endl;
        std::cout << "Count: " << (itHigh - itLow) << std::endl;
    }
}
```

### 二分答案 Binary Search on Answer

‌二分答案是二分查找思想的一种高级应用。它不直接在有序数组中查找某个具体的值，而是将‌最优化问题‌（求最大值、最小值）转化为‌判定性问题‌（判断某个解是否可行）。其核心逻辑是：‌猜测一个答案，验证其可行性，然后根据验证结果缩小猜测范围。

```c
#include <iostream>
#include <vector>
#include <algorithm>

using namespace std;

bool check(int dist) {
    if (conditions) {
        return true;
    } else {
        return false;
    }
}

// 求最小可行解
int minResult(int left, int right) {
    while (left < right) {
        int mid = left + (right - left) / 2; // 向下取整
        if (check(mid)) { 
            // mid 可行，但我们要找最小的，所以保留 mid，排除右半部分
            right = mid; 
        } else {
            // mid 不可行，解必须在右边
            left = mid + 1;
        }
    }
    // 循环结束时 left == right，即为答案
    return left;    
}

int main() {
    int L,R;
    cin >> L >> R;
    
    cout << minResult(L, R) << endl;
    
    return 0;
}
```

```c
#include <iostream>
#include <vector>
#include <algorithm>

using namespace std;

bool check(int dist) {
    if (conditions) {
        return true;
    } else {
        return false;
    }
}

// 求最大可行解
int maxResult(int left, int right) {
    while (left < right) {
        // 注意：这里必须 +1 向上取整，防止死循环
        int mid = left + (right - left + 1) / 2; 
        if (check(mid)) {
            // mid 可行，但我们要找最大的，所以保留 mid，排除左半部分
            left = mid;
        } else {
            // mid 不可行，解必须在左边
            right = mid - 1;
        }
    }
    // 循环结束时 left == right，即为答案
    return left;    
}

int main() {
    int L,R;
    cin >> L >> R;

    cout << maxResult(L, R) << endl;
    
    return 0;
}
```

```c
// 有N个牛棚在一条直线上，位置分别为𝑥1,...,𝑥𝑁。要把C头牛放入这些牛棚，使得任意两头牛之间的最小距离最大。求这个最大距离。
#include <iostream>
#include <vector>
#include <algorithm>

using namespace std;

int N, C;
vector<int> stalls;

// 检查在最小间距为 dist 时，能否放下 C 头牛
bool check(int dist) {
    int count = 1; // 第一头牛放在第一个牛棚
    int last_pos = stalls[0];
    
    for (int i = 1; i < N; ++i) {
        if (stalls[i] - last_pos >= dist) {
            count++;
            last_pos = stalls[i];
            if (count >= C) return true; // 提前退出
        }
    }
    return false;
}

int main() {
    // 优化IO
    ios::sync_with_stdio(false);
    cin.tie(NULL);

    if (!(cin >> N >> C)) return 0;
    
    stalls.resize(N);
    for (int i = 0; i < N; ++i) {
        cin >> stalls[i];
    }
    
    // 1. 排序，因为牛棚位置可能是无序的
    sort(stalls.begin(), stalls.end());
    
    // 2. 确定边界
    // 最小可能距离是 1 (或者0，视题目而定，通常至少为1)
    // 最大可能距离是 (最大坐标 - 最小坐标)
    int left = 1;
    int right = stalls[N - 1] - stalls[0];
    
    // 3. 二分答案 (求最大值模板)
    while (left < right) {
        // 向上取整，防止死循环
        int mid = left + (right - left + 1) / 2;
        
        if (check(mid)) {
            // 当前距离 mid 可行，尝试更大的距离
            left = mid;
        } else {
            // 当前距离 mid 不可行，必须减小距离
            right = mid - 1;
        }
    }
    
    cout << left << endl;
    
    return 0;
}
```

## 5.8 贪心算法 Greedy Algorithm

贪心算法是一种经典的算法设计技术，核心逻辑是在每一步决策时都做出‌当前状态下的局部最优选择‌，通过逐步累积这些局部最优解，期望最终得到整个问题的全局最优解。它采用‌自顶向下、无回溯‌的迭代方式推进，每做出一次贪心选择，就将原问题简化为规模更小的子问题，全程不回溯、不推翻之前的决策，省去了穷尽所有可能性的复杂计算，运行效率极高。

### 核心要素

贪心算法并非对所有问题都能得到全局最优解，只有同时满足以下两个特性时，才能保证结果正确：

贪心选择性质‌：问题的全局最优解可以通过一系列互不依赖的局部最优选择推导出来，每一步决策只依赖之前的选择，不需要预判后续所有可能性。
最优子结构性质‌：整个问题的最优解，必然包含它所有子问题的最优解，子问题的局部最优是全局最优的组成部分。

### 优缺点

优点：代码实现简单，时间复杂度通常为O(n)或O(nlogn)，远低于动态规划的复杂度，适合大规模数据场景。  

缺点：无法保证所有场景下都得到全局最优解。比如经典的0-1背包问题，直接用贪心策略选单价最高的物品，很可能导致剩余空间浪费，最终得到的只是近似最优解而非全局最优。

### 区间调度模型（最大不相交区间数）

```c
#include <iostream>
#include <algorithm>
#include <vector>

using namespace std;

struct Interval {
    int start, end;
};

// 核心：按结束时间升序排序；若结束时间相同，可按开始时间升序（非必须，但规范）
bool cmp(const Interval& a, const Interval& b) {
    if (a.end != b.end) return a.end < b.end;
    return a.start < b.start;
}

int main() {
    int n;
    cin >> n;
    vector<Interval> intervals(n);
    for (int i = 0; i < n; ++i) {
        cin >> intervals[i].start >> intervals[i].end;
    }

    // 1. 排序
    sort(intervals.begin(), intervals.end(), cmp);

    int count = 0;
    int last_end = -1e9; // 记录上一个选中区间的结束时间

    // 2. 贪心选择
    for (const auto& it : intervals) {
        // 如果当前区间的开始时间 >= 上一个选中区间的结束时间，则不冲突
        if (it.start >= last_end) {
            count++;
            last_end = it.end; // 更新结束时间
        }
    }

    cout << count << endl;
    return 0;
}
```

### 区间分组模型（最少资源数）

```c
#include <iostream>
#include <algorithm>
#include <vector>
#include <queue>

using namespace std;

struct Interval {
    int l, r;
    // 按开始时间升序排序
    bool operator<(const Interval& other) const {
        return l < other.l;
    }
};

int main() {
    int n;
    cin >> n;
    vector<Interval> intervals(n);
    for (int i = 0; i < n; ++i) {
        cin >> intervals[i].l >> intervals[i].r;
    }

    // 1. 按开始时间排序
    sort(intervals.begin(), intervals.end());

    // 2. 小根堆：存储每个组目前最后一个区间的结束时间
    // greater<int> 使 priority_queue 变为小根堆
    priority_queue<int, vector<int>, greater<int>> min_heap;

    for (const auto& interval : intervals) {
        // 如果堆不为空，且堆顶（最早结束的组）的结束时间 <= 当前区间开始时间
        // 说明可以复用该组
        if (!min_heap.empty() && min_heap.top() <= interval.l) {
            min_heap.pop(); // 弹出旧结束时间
        }
        
        // 将当前区间的结束时间加入堆（无论是新开组还是复用组）
        min_heap.push(interval.r);
    }

    // 堆的大小即为最少需要的组数
    cout << min_heap.size() << endl;

    return 0;
}
```

### 排队接水模型（最小化平均等待时间）

```c
#include <iostream>
#include <algorithm>
#include <vector>
#include <iomanip>

using namespace std;

int main() {
    int n;
    cin >> n;
    vector<int> times(n);
    for (int i = 0; i < n; ++i) {
        cin >> times[i];
    }

    // 1. 核心：按服务时间升序排序
    sort(times.begin(), times.end());

    long long total_wait = 0; // 总等待时间
    long long current_sum = 0; // 当前累计花费的时间

    // 2. 累加计算
    // 第 i 个人的等待时间 = 前面所有人服务时间之和
    for (int i = 0; i < n; ++i) {
        total_wait += current_sum;
        current_sum += times[i];
    }

    // 输出平均等待时间（保留两位小数）
    double avg_wait = (double)total_wait / n;
    cout << fixed << setprecision(2) << avg_wait << endl;

    return 0;
}
```

### 合并果子模型（最小代价合并）

```c
#include <iostream>
#include <queue>
#include <vector>

using namespace std;

int main() {
    int n;
    cin >> n;
    
    // 小根堆
    priority_queue<long long, vector<long long>, greater<long long>> min_heap;
    
    for (int i = 0; i < n; ++i) {
        long long x;
        cin >> x;
        min_heap.push(x);
    }

    long long total_cost = 0;

    // 当堆中元素大于1时，持续合并
    while (min_heap.size() > 1) {
        // 取出最小的两个
        long long a = min_heap.top(); min_heap.pop();
        long long b = min_heap.top(); min_heap.pop();
        
        long long sum = a + b;
        total_cost += sum; // 累加代价
        
        min_heap.push(sum); // 将新堆放回
    }

    cout << total_cost << endl;

    return 0;
}
```

## 5.9 分治算法 Divide and Conquer

分治算法是一种经典的递归式算法设计范式，核心思想是“分而治之”：将一个复杂的大问题，拆解为若干个结构相同、规模更小的独立子问题，递归求解子问题后，再将子问题的解逐层合并，最终得到原问题的完整答案。

### 核心要素

所有分治算法的实现都严格遵循这三个递归流程：  
分解（Divide）‌将原问题拆分为若干个‌规模更小、相互独立、结构与原问题完全相同‌的子问题，通常优先将问题对半拆分，保证子问题规模均衡，最大化算法效率。  
解决（Conquer）‌递归求解每个子问题，当子问题的规模缩小到阈值以下时，直接给出简单解，终止递归，避免无限调用。  
合并（Combine）‌按照原问题的逻辑规则，将所有子问题的解逐层整合，最终拼接出原问题的全局最优解。合并的效率直接决定了分治算法的整体时间复杂度。  

只有同时满足以下特征的问题，才能用分治算法高效求解，否则可能无法得到正确解或效率低下：  
可直接求解的小规模子问题‌：当问题规模缩小到一定程度后，无需继续拆分，能直接得到答案，这是递归终止的基础。  
最优子结构性质‌：原问题的全局最优解，必然包含所有子问题的局部最优解，子问题的最优解是全局解的组成部分。  
解可合并性‌：子问题的解可以通过简单操作高效拼接，最终得到原问题的完整解。如果子问题解无法合并，分治将失去意义。  
子问题完全独立‌：拆分出的子问题之间没有公共重叠部分，不存在重复的公共子问题。这是分治和动态规划最核心的区别，子问题不独立会导致大量重复计算，此时动态规划是更优选择。  

### 优缺点

优点  
大幅降低时间复杂度‌：通过将问题对半拆分，能把原本O(n²)复杂度的问题优化到O(n log n)级别，典型如快速排序、归并排序，甚至能将大整数乘法从O(n²)降至约O(n^1.585)，快速傅里叶变换（FFT）更是实现了量级上的效率飞跃。  
天然支持并行计算‌：由于拆分出的子问题完全独立、互不依赖，多个子问题可以同时在不同计算单元上并行执行，非常适配多核CPU、分布式计算场景，能充分利用硬件算力。  
代码逻辑清晰易维护‌：将复杂大问题拆解为结构相同的小问题，递归实现的代码结构简洁直观，大幅降低了处理大规模复杂问题的思维难度，比如汉诺塔、最近点对等问题，用分治实现比纯迭代逻辑更易理解。  
稳定性与鲁棒性强‌：只要问题满足分治的四大适用特征，算法就能稳定输出全局最优解，不会出现贪心算法可能陷入局部最优的问题。  

缺点  
递归栈溢出风险‌：分治依赖多层递归调用，当问题规模过大、递归深度过深时，会持续占用系统调用栈空间，很容易超出栈内存上限，直接触发栈溢出错误，导致程序崩溃。  
合并步骤实现难度高‌：分治算法的性能完全依赖“合并”环节的效率，不同问题的合并逻辑没有统一模板，部分复杂场景下合并操作的实现难度极高，甚至可能抵消分治带来的效率提升。  
子问题不独立时效率骤降‌：如果问题存在大量重叠的公共子问题，分治会重复计算大量完全相同的子任务，此时分治的效率远低于动态规划，完全不适用这类场景。  
额外内存开销‌：部分分治实现（如归并排序）需要额外开辟临时数组存储中间结果，空间复杂度高于原地排序算法，对内存资源紧张的场景不够友好。  

### 归并排序

```c
#include <iostream>
#include <vector>
using namespace std;

// 合并两个有序子数组
void merge(vector<int>& arr, int left, int mid, int right) {
    vector<int> temp(right - left + 1);
    int i = left, j = mid + 1, k = 0;
    
    // 双指针合并两个有序区间
    while (i <= mid && j <= right) {
        if (arr[i] <= arr[j]) temp[k++] = arr[i++];
        else temp[k++] = arr[j++];
    }
    // 补全剩余元素
    while (i <= mid) temp[k++] = arr[i++];
    while (j <= right) temp[k++] = arr[j++];
    
    // 将合并结果拷贝回原数组
    for (int p = 0; p < temp.size(); ++p) {
        arr[left + p] = temp[p];
    }
}

// 分治递归主体
void mergeSort(vector<int>& arr, int left, int right) {
    if (left >= right) return; // 递归终止：子数组只剩1个元素，天然有序
    
    int mid = left + (right - left) / 2;
    // 1. 分解：递归排序左右两半
    mergeSort(arr, left, mid);
    mergeSort(arr, mid + 1, right);
    // 2. 合并：将两个有序子数组合并
    merge(arr, left, mid, right);
}

int main() {
    vector<int> arr = {3, 1, 4, 1, 5, 9, 2, 6};
    mergeSort(arr, 0, arr.size() - 1);
    
    for (int num : arr) cout << num << " ";
    return 0;
}
```

### 快速排序

```c
#include <iostream>
#include <vector>
#include <algorithm> // 用于 swap

using namespace std;

/**
 * 分区函数 (Partition)
 * 功能：选取基准，将数组分为“小于基准”和“大于基准”两部分
 * 参数：arr-待排序数组, left-左边界, right-右边界
 * 返回：基准元素最终所在的索引位置
 */
int partition(vector<int>& arr, int left, int right) {
    // 1. 选择基准：这里选择最右侧元素作为基准 (Pivot)
    int pivot = arr[right]; 
    
    // i 指向“小于基准区域”的最后一个元素
    // 初始时，小于基准的区域为空，所以 i 在 left 之前
    int i = left - 1; 

    // 2. 遍历数组，将小于 pivot 的元素交换到左侧
    for (int j = left; j < right; j++) {
        if (arr[j] <= pivot) {
            i++;                // 扩展小于基准的区域
            swap(arr[i], arr[j]); // 将当前小元素换到区域末尾
        }
    }

    // 3. 将基准元素放到正确的位置（即小于区域和大于区域之间）
    swap(arr[i + 1], arr[right]);
    
    // 返回基准元素的最终索引
    return i + 1;
}

/**
 * 快速排序主函数 (Quick Sort)
 * 功能：递归地对左右子数组进行排序
 */
void quickSort(vector<int>& arr, int left, int right) {
    // 递归终止条件：左边界 >= 右边界时，子数组长度为0或1，无需排序
    if (left < right) {
        // 1. 分区操作，获取基准元素的最终位置
        int pi = partition(arr, left, right);

        // 2. 递归排序基准左边的子数组
        quickSort(arr, left, pi - 1);

        // 3. 递归排序基准右边的子数组
        quickSort(arr, pi + 1, right);
    }
}

int main() {
    // 优化 IO 效率
    ios::sync_with_stdio(false);
    cin.tie(nullptr);

    vector<int> arr = {10, 7, 8, 9, 1, 5, 3, 2, 4, 6};
    
    cout << "排序前: ";
    for (int num : arr) cout << num << " ";
    cout << endl;

    // 调用快速排序
    quickSort(arr, 0, arr.size() - 1);

    cout << "排序后: ";
    for (int num : arr) cout << num << " ";
    cout << endl;

    return 0;
}

```

## 5.10 递归 Recursion

递归是一种在函数或过程的定义中调用自身来解决问题的编程方法与数学思维。其核心逻辑是将一个复杂的大问题，逐步分解为结构相同但规模更小的子问题，直到子问题简化到可以直接求解的程度，再逐层返回结果以解决原问题。

### 核心要素

基线条件（终止条件）‌  
这是递归的“出口”，指不再进行递归调用、直接给出答案的最简单情况。
如果没有基线条件，程序将陷入无限循环（死递归），最终导致栈溢出（Stack Overflow）。
例如：计算阶乘时，n=1 或 n=0 时直接返回 1。

递归步骤（递推关系）‌  
这是递归的“过程”，指将当前问题转化为一个或多个规模更小的同类子问题，并调用自身来解决这些子问题。
每次调用都必须使问题规模向基线条件靠近。
例如：n! = n * (n-1)!，将求 n 的阶乘转化为求 n-1 的阶乘。

### 优缺点

优点  
代码简洁直观‌：用极少量代码就能表达复杂逻辑，结构清晰，大幅减少冗余循环代码，比如阶乘、汉诺塔的递归实现仅需数行核心代码。  
贴合自然思维‌：处理树形结构遍历、分治算法等场景时，思路完全契合人类拆解问题的直觉，和数学归纳法的表达高度一致，理解门槛更低。  
适配特殊场景‌：对树的前中后序遍历、图的深度优先搜索（DFS）、快速排序等递归结构问题，递归实现比迭代更自然高效，无需手动维护复杂栈结构。  

缺点  
栈溢出风险‌：每次递归调用都要将函数上下文保存到调用栈中，递归深度过大时，很容易超出系统栈的内存上限，直接触发栈溢出错误。  
额外性能损耗‌：函数调用存在参数传递、现场保护、栈帧维护的额外开销，运行效率通常低于等价的迭代（循环）实现。  
调试难度高‌：层层嵌套的调用关系，让追踪函数执行路径、定位变量异常变化变得十分困难，排错成本远高于普通循环代码。  
重复计算问题‌：未做优化的朴素递归（比如无记忆化的斐波那契数列求解），会反复计算大量完全相同的子问题，时间复杂度指数级飙升。  

### 通用递归模板

```c
void recursion(int level, int param) {
    // 1. 递归终结条件 (Terminator)
    // 当达到最大层级或满足基本条件时，停止递归并处理结果
    if (level > MAX_LEVEL) {
        // process result
        return;
    }

    // 2. 处理当前层逻辑 (Process Current Logic)
    // 处理当前层级需要做的事情，比如打印、计算、修改局部变量
    process(level, param);

    // 3. 下探到下一层 (Drill Down)
    // 调用自身，传入更新后的参数（如 level + 1），解决子问题
    recursion(level + 1, newParam);

    // 4. 清理当前层状态 (Reverse Status) - 可选
    // 如果当前层修改了全局状态或共享数据，需要在返回前恢复原状
    // reverse the current level status if needed
}
```

### 回溯算法

```c
vector<vector<int>> res; // 最终结果集合
vector<int> path;        // 当前路径/选择列表

void backtrack(int start, vector<int>& nums) {
    // 1. 终止条件：满足题目要求时，保存结果
    if (path.size() == target_size) { // 例如：路径长度达到要求
        res.push_back(path);
        return;
    }

    // 2. 选择列表：遍历当前层所有可能的选择
    for (int i = start; i < nums.size(); i++) {
        
        // 剪枝操作（可选）：如果当前选择不符合条件，直接跳过
        // if (invalid_condition) continue;

        // 3. 做选择
        path.push_back(nums[i]);

        // 4. 进入下一层决策
        // 注意参数变化：
        // - 如果是组合/子集（不可重复选同一元素）：传 i + 1
        // - 如果是完全背包/可重复选：传 i
        // - 如果是全排列（需打乱顺序）：通常不带 start，配合 used 数组
        backtrack(i + 1, nums); 

        // 5. 撤销选择（回溯）
        // 恢复现场，以便尝试下一个选择
        path.pop_back();
    }
}
```

### 分治递归

```c
void hanoi(int n, char from, char to, char aux) {
    // 1. 终止条件：只剩一个盘子，直接移动
    if (n == 1) {
        cout << "Move disk 1 from " << from << " to " << to << endl;
        return;
    }
    
    // 2. 分解问题：
    // 第一步：将上面 n-1 个盘子从 from 移到 aux (借助 to)
    hanoi(n - 1, from, aux, to);
    
    // 第二步：将最大的第 n 个盘子从 from 移到 to
    cout << "Move disk " << n << " from " << from << " to " << to << endl;
    
    // 第三步：将 n-1 个盘子从 aux 移到 to (借助 from)
    hanoi(n - 1, aux, to, from);
}
```
