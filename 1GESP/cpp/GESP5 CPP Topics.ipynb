## 5.1 初等数论

初等数论是研究整数性质的数学分支‌，主要用初等方法（不依赖高等数学工具）研究整除、同余、不定方程等问题。

### 素数判定算法与优化

```c
//基础素数判定方法
bool isPrime(int n){
    if(n<=1) return false;
    for(int i=2;i<n;++i){
        if(n%i==0) return false;
    }
    return true;
}
```

```c
//优化方案：平方根上限
bool isPrime(int n){
    if(n<=1) return false;
    for(int i=2;i*i<=n;++i){
        if(n%i==0) return false;
    }
    return true;
}
```

```c
//埃拉托斯特尼筛法
vector<bool> sieve(int max_num){
    vector<bool> is_prime(max_num+1,true);
    is_prime[0]=is_prime[1]=false;
    for(int i=2;i*i<=max_num;++i){
        if(is_prime[i]){
            for(int j=i*i;j<=max_num;j+=i){
                is_prime[j]=false;
            }
        } 
    }
    return is_prime;
}
```

### 最大公约数

最大公约数是指能够同时整除两个或多个整数的最大正整数。

欧几里得算法，gcd(a, b) = gcd(b, a mod b)

```c
//递归
int gcd(int a,int b){
    if(b==0) return a;
    return gcd(b,a%b);
}
```

```c
//迭代
int gcd(int a,int b){
    while(b!=0){
        int t=b;
        b=a%b;
        a=t;
    }
    return a;
}
```

### 最小公倍数

lcm(a, b) = |a * b| / gcd(a, b)

```c
int gcd(int a,int b){
    if(b==0) return a;
    return gcd(b,a%b);
}
int lcm(int a,int b){
    return a/gcd(a,b)*b;//先除后乘避免溢出
}
```

### 模运算与同余方程

(a + b) mod m = (a mod m + b mod m)

(a * b) mod m = (a mod m * b mod m)

```c
//快速幂取模
int fasrPow(int a,int b,int m){
    int r=1;
    a=a%m;
    while(b>0){
        if(b%2==1){
            r=(r*a)%m;
        }
        a=(a*a)%m;
        b=b/2;
    }
    return r;
}
```

```c
//线性同余方程
int extended_gcd(int a,int b,int &x,int &y){
    if(b==0){
        x=1;
        y=0;
        return a;
    }
    int gcd=extended_gcd(b,a%b,y,x);
    y=y-a/b*x;
    return gcd;
}
bool solve_congruence(int a,int b,int m,int &x){
    int x0,y0;
    int d=extended_gcd(a,m,x0,y0);
    if(b%d!=0) return false;
    x=(x0*(b/d))%m;
    if(x<0) x+=m;
    return true;
}
```

### 质因数分解与约数

质因数分解是将一个合数表示为若干素数乘积的过程。

```c
//质因数分解算法
vector<int> prime_factors(int n){
    vector<int> factors;
    while(n%2==0){
        factors.push_back(2);
        n/=2;
    }
    for(int i=3;i*i<=n;i+=2){
        while(n%i==0){
            factors.push_back(i);
            n/=i;
        }
    }
    if(n>2){
        factors.push_back(n);
    }
    return factors;
}
```

```c
vector<int> prime_factors(int n){
    vector<int> factors;
    while(n%2==0){
        factors.push_back(2);
        n/=2;
    }
    for(int i=3;i*i<=n;i+=2){
        while(n%i==0){
            factors.push_back(i);
            n/=i;
        }
    }
    if(n>2){
        factors.push_back(n);
    }
    return factors;
}
//计算约数个数
int count_divisors(int n){
    auto factors=prime_factors(n);
    if(factors.empty()) return 1;
    int c=1;
    int current=factors[0],power=1;
    for(size_t i=1;i<factors.size();++i){
        if(factors[i]==current){
            power++;
        }else{
            c*=(power+1);
            current=factors[i];
            power=1;
        }
    }
    c*=(power+1);
    return c;
}
```

```c
vector<int> prime_factors(int n){
    vector<int> factors;
    while(n%2==0){
        factors.push_back(2);
        n/=2;
    }
    for(int i=3;i*i<=n;i+=2){
        while(n%i==0){
            factors.push_back(i);
            n/=i;
        }
    }
    if(n>2){
        factors.push_back(n);
    }
    return factors;
}
//计算约数之和
int count_divisors(int n){
    auto factors=prime_factors(n);
    if(factors.empty()) return 1;
    int sum=1;
    int current=factors[0],current_sum=current+1;
    for(size_t i=1;i<factors.size();++i){
        if(factors[i]==current){
            current_sum=current_sum*current+1;
        }else{
            sum*=current_sum;
            current=factors[i];
            current_sum=current+1;
        }
    }
    sum*=current_sum;
    return sum;
}
```

### 奇偶性分析与位运算

```c
//判断偶数
bool is_even(int n){
    if((n&1)==0){
        return true;
    }else{
        return false;
    } 
}
```

```c
//位运算
int calculator(int n){
    n<<1;// n*2
    n>>1;// n/2
    return n;
}
```

```c
//交换两个数,按位异或赋值
void change(int a,int b){
    a^=b;
    b^=a;
    a^=b;
}
```

```c
//快速计算2的幂次
bool power2(int exp){
    return 1<<exp;
}
//快速判断是否为2的幂
bool is_power_of_two(int n){
    return n>0&&(n&(n-1))==0;
}
```

```c
//计算Hamming重量
int hamming_weight(uint32_t n){
    int c=0;
    while(n!=0){
        n&=n-1;
        c++;
    }
    return c;
}
```
