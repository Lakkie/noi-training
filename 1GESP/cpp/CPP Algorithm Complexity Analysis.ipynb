# 常数阶 O(1)

```c
int getFirstElement(const vector<int>& arr) {
    if (!arr.empty()) return arr; // 仅一次访问
    return -1;
}
```

# 线性阶 O(N)

```c
int linearSearch(const vector<int>& arr, int target) {
    for (int num : arr) { // 遍历 N 次
        if (num == target) return num;
    }
    return -1;
}
```
