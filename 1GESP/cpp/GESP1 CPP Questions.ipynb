```c
//B4495 [GESP202603 一级] 交朋友 顺序结构
#include<iostream>
using namespace std;
int main(){
    //输入
    int h1,h2,h3,h4;          //存放班上4个小朋友的身高。
    cin>>h1>>h2>>h3>>h4;
    //计算
    int min_diff=10000;       //存放最小身高差，初始给一个足够大的数。（比所有可能的身高差都要大）
    int best_h=0;             //存放最佳身高，即身高最接近Alice的小朋友的身高。
    int diff=0;               //存放身高差，即其他小朋友和Alice的身高差值。
    
    diff=h2-h1;               //计算2号小朋友和Alice的身高差。
    if(diff<0){
        diff=diff*-1;
    }
    if(diff<min_diff){        //如果这次的身高差比之前的最小身高差要小，就更新数据。
        min_diff=diff;        //更新最小身高差
        best_h=h2;            //更新最佳身高
    }else if(diff==min_diff){ //如果这次的身高差跟之前的最小身高差一样
        if(h2<best_h){        //选更矮的小朋友
            best_h=h2;        //更新最佳身高
        }
    }

    diff=h3-h1;               //计算3号小朋友和Alice的身高差。
    if(diff<0){
        diff=diff*-1;
    }
    if(diff<min_diff){
        min_diff=diff;
        best_h=h3;
    }else if(diff==min_diff){
        if(h3<best_h){
            best_h=h3;
        }
    }

    diff=h4-h1;               //计算4号小朋友和Alice的身高差。
    if(diff<0){
        diff=diff*-1;
    }
    if(diff<min_diff){
        min_diff=diff;
        best_h=h4;        
    }else if(diff==min_diff){
        if(h4<best_h){
            best_h=h4;
        }
    }
    //输出
    cout<<best_h;
}
```

```c
//B4495 [GESP202603 一级] 交朋友 循环结构
#include<iostream>
using namespace std;
int main(){
    //输入
    int h[4];                     //存放班上4个小朋友的身高。
    for(int i=0;i<4;++i){
        cin>>h[i];                //h[0]是Alice的身高，h[1]、h[2]、h[3]是其他三个小朋友的身高。
    }
    //计算
    int min_diff=10000;           //存放最小身高差，初始给一个足够大的数。（比所有可能的身高差都要大）
    int best_h=0;                 //存放最佳身高，即身高最接近Alice的小朋友的身高。
    int diff=0;                   //存放身高差，即其他小朋友和Alice的身高差值。
    for(int i=1;i<4;++i){         //循环计算其他三个小朋友和Alice的身高差。
        diff=h[i]-h[0];               
        if(diff<0){
            diff=diff*-1;
        }
        if(diff<min_diff){        //如果这次的身高差比之前的最小身高差要小，就更新数据。
            min_diff=diff;        //更新最小身高差
            best_h=h[i];          //更新最佳身高
        }else if(diff==min_diff){ //如果这次的身高差跟之前的最小身高差一样
            if(h[i]<best_h){      //选更矮的小朋友
                best_h=h[i];      //更新最佳身高
            }
        }
    }
    //输出
    cout<<best_h;
}
```

```c
//B4496 [GESP202603 一级] 数字替换 分解数字
#include<iostream>
using namespace std;
int main(){
    //输入
    int n;
    cin>>n;
    //计算
    int result=0;              //用于存储替换后的新数字
    int k=1;                   //位权，初始为1（代表个位），每次循环乘10
    if(n==0){                  //特殊情况：如果输入是0，直接输出0
        cout<<0<<endl;
        return 0;
    }
    while(n>0){
        int digit=n%10;        //取出个位数字
        if(digit==4){          //判断并替换
            digit=8;
        }
        result=result+digit*k; //累加到替换后的新数字
        k*=10;                 //位权进位（1->10->100...）
        n/=10;                 //去掉已处理的个位
    }
    //输出
    cout<<result<<endl;
    return 0;
}
```
