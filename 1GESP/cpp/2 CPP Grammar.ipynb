# 什么是C语言？

C语言‌是一种通用的、面向过程的计算机编程语言，由丹尼斯·里奇（Dennis Ritchie）于1972年在贝尔实验室开发，最初用于编写UNIX操作系统。

核心定位‌：
它被誉为“中级语言”，因为它既具有高级语言的易读性和结构化特征，又保留了汇编语言直接操作硬件和内存的能力。

主要特点‌：
高效灵活‌：编译后生成的机器码执行效率极高，接近汇编语言。
简洁紧凑‌：关键字少（仅30余个），语法简洁，适合底层开发。
可移植性强‌：标准C代码可以在多种硬件平台和操作系统上编译运行。
手动内存管理‌：没有垃圾回收机制，程序员需手动分配和释放内存，这带来了高性能，但也要求开发者具备严谨的资源管理能力。

应用场景‌：
广泛用于操作系统内核（如Linux、Windows部分核心）、设备驱动程序、嵌入式系统（如单片机、物联网设备）、编译器开发以及对性能要求极高的科学计算领域。

# 什么是C++语言？

C++‌语言是由本贾尼·斯特劳斯特卢普（Bjarne Stroustrup）于1979年在贝尔实验室开始研发，并于1983年正式命名的编程语言。它是在C语言的基础上扩展而来的，旨在保留C语言高效性的同时，增加对大型软件复杂性的管理能力。

## 核心定位‌：  
C++是一门多范式编程语言，支持面向过程、面向对象（OOP）和泛型编程。名字中的“++”源自C语言的自增运算符，寓意“在C的基础上向前推进一步”。

## 主要特点‌：  
兼容C语言‌：C++几乎完全兼容C，这意味着大多数C代码可以在C++编译器中运行。  
面向对象特性‌：引入了类（Class）、对象、继承、多态和封装等概念，使得代码更易于模块化、复用和维护，特别适合大规模软件开发。  
丰富的功能库‌：拥有强大的标准模板库（STL），提供了容器、算法和迭代器等高效工具。  
现代特性演进‌：随着C++11、C++14、C++17及后续标准的发布，引入了智能指针、Lambda表达式、自动类型推导等现代特性，提升了开发安全性和便捷性。  

## 应用场景‌：  
广泛应用于游戏引擎（如Unreal Engine）、高性能服务器后端、图形图像处理、金融交易系统、浏览器内核（如Chrome）、数据库系统以及各类桌面应用软件。

# 什么是C语言标准？

C语言标准是由国际标准化组织（ISO）和国际电工委员会（IEC）共同制定的一系列规范，旨在统一C语言的语法、语义及标准库，确保代码在不同编译器和硬件平台上的可移植性。

# 什么是C++语言标准？

C++语言标准是由ISO/IEC联合技术委员会下属的C++标准化工作组（WG21）制定的国际规范，用于统一C++的语法、语义和标准库定义，保障代码跨平台可移植性。

# 什么是C++的标准模板库（Standard Template Library）？

Standard Template Library‌（简称 ‌STL‌），中文译为‌标准模板库‌，是 C++ 标准库的核心组成部分，也是 C++ 语言中最具影响力的创新之一。它由 Alexander Stepanov 等人开发，最终被纳入 ISO C++ 标准，旨在提供一套高效、通用且可复用的数据结构和算法框架。

## 1. 常用容器头文件  
\<vector>‌：动态数组，支持随机访问，最常用的序列容器。  
\<string>‌：字符串类，提供便捷的文本处理功能。  
\<list>‌：双向链表，支持高效的插入和删除操作。  
\<deque>‌：双端队列，支持两端高效插入删除。  
\<stack>‌：栈适配器，遵循后进先出（LIFO）原则。  
\<queue>‌：队列适配器，遵循先进先出（FIFO）原则。  
\<priority_queue>‌：优先队列，元素按优先级排序。  
\<map> / \<set>‌：基于红黑树的关联容器，自动排序且键值唯一。  
\<unordered_map> / \<unordered_set>‌：基于哈希表的关联容器，查找效率更高。  

## 2. 算法与函数对象头文件  
‌\<algorithm>‌：核心算法库，包含 sort、find、max、reverse 等通用算法。  
‌\<functional>‌：函数对象库，提供 greater、less、bind 等仿函数和绑定器。  
‌\<numeric>‌：数值算法库，包含 accumulate（求和）、iota（递增填充）等。

## 3. 迭代器与工具头文件  
‌\<iterator>‌：迭代器工具，提供 back_inserter、istream_iterator 等特殊迭代器。  
‌\<utility>‌：实用工具库，包含 pair、swap、move 等基础功能。  
‌\<memory>‌：内存管理库，包含智能指针 shared_ptr、unique_ptr 等。  

# 什么是C++的标准库命名空间‌（std）？  

在 C++ 编程语言中，‌std‌ 是英文单词 ‌standard‌（标准）的缩写。

它代表 ‌C++ 标准库命名空间‌。C++ 标准库中的所有类、函数和对象（例如 cout、cin、string、vector 等）都被定义在这个名为 std 的命名空间中。使用命名空间的主要目的是为了防止不同库或用户代码之间出现名称冲突，确保标识符的唯一性和代码的组织性。

# 什么是变量？

在 C++ 中，变量是用于在内存中存储数据的容器。程序通过变量来保存、访问和操作数据。

## 变量的定义（声明）

数据类型 变量名 = 初始值;

数据类型，决定了变量能存储什么样的数据以及占用多少内存。
变量名，变量的唯一标识符。
可选的初始值。

```c

int age;          // 声明一个整型变量，未初始化
int score = 100;  // 声明并初始化为 100
int x, y, z;      // 同时声明多个同类型的变量


```

## 变量的使用

变量的命名规则（必须遵守）：‌

名称只能包含字母（a-z, A-Z）、数字（0-9）和下划线（_）。
名称必须以字母或下划线开头，不能以数字开头。
名称区分大小写（myVar 和 myvar 是不同的变量）。
不能使用 C++ 关键字（如 int, return, class 等）作为变量名。

变量命名的最佳实践：‌

使用有意义的名称‌：例如使用 studentAge 而不是 a，以提高代码可读性。
驼峰命名法‌：通常变量名首字母小写，后续单词首字母大写（如 myVariableName）。

# 基本运算

## 算术运算 Arithmetic operations

```c

int a = 10, b = 3;

cout << "加法: " << a + b << endl;
cout << "减法: " << a - b << endl;
cout << "乘法: " << a * b << endl;

cout << "除法: " << (double) a / b << endl;
cout << "整除: " << a / b << endl;
cout << "取模: " << a % b << endl;

```

## 关系运算 Relational operations

```c

int age = 20;
double score = 85.5;
char grade = 'A';

// 等于
cout << "score == 90: " << (score == 90) << endl;
// 不等于
cout << "score != 90: " << (score != 90) << endl;

// 大小比较
cout << "age > 18: " << (age > 18) << endl;
cout << "age < 18: " << (age < 18) << endl;
cout << "score >= 90: " << (score >= 90) << endl;
cout << "score <= 90: " << (score <= 90) << endl;

// 字符比较 (基于 ASCII 码)
cout << "grade >= 'B': " << (grade >= 'B') << endl; // 输出 1 (true, 'A'的ASCII值大于'B')

// 注意 浮点数比较的特殊性
double a = 0.1 + 0.2;
double b = 0.3;
cout << "\n--- 浮点陷阱 ---" << endl;
// 由于精度问题，直接比较可能不相等
cout << "0.1+0.2 == 0.3 ? " << (a == b) << endl; // 通常输出 0 (false)
// 推荐做法：比较差值的绝对值是否极小
cout << "近似相等 ? " << (abs(a - b) < 1e-9) << endl; // 输出 1 (true)

```

## 逻辑运算 Logical operations

```c

int age = 25;
bool hasTicket = true;
bool isMember = false;

// 1. 逻辑与 (&&)：必须同时满足
// 场景：年满18岁 且 有票才能入场
if (age >= 18 && hasTicket) {
    cout << "允许入场" << endl;
} else {
    cout << "禁止入场" << endl;
}

// 2. 逻辑或 (||)：满足其一即可
// 场景：是会员 或 持有优惠券均可打折
if (isMember || hasTicket) {
    cout << "享受折扣" << endl;
}

// 3. 逻辑非 (!)：取反操作
// 场景：如果“不是”会员，则提示注册
if (!isMember) {
    cout << "您还不是会员，建议注册" << endl;
}

// 4. 短路求值 (Short-circuit Evaluation)
int x = 0;
// 在 A && B 中，若 A 为假，B 不会被执行
if (x != 0 && (10 / x > 1)) { 
    cout << "计算成功" << endl;
} else {
    cout << "跳过除法，避免除零错误" << endl;
}

```

# 输入输出

## 基础输入输出

```c

#include <iostream>
using namespace std; // 实际项目中建议直接使用 std::cin/std::cout

int main(){
    // 输入 (cin)‌：使用右移运算符 >> 从键盘读取数据。
	int n,s=0;
	cin>>n;

    // 计算
    for(int i=0;i<n;i++){
        s+=i;
    }
    
    // 输出 (cout)‌：使用左移运算符 << 将数据发送到屏幕。
	cout<<s<<endl;

    return 0;
}

```

```c

#include <iostream>

int main(){
    // 输入
	int n,s=0;
	std::cin>>n;

    // 计算
    for(int i=0;i<n;i++){
        s+=i;
    }
    
    // 输出
	std::cout<<s<<std::endl;

    return 0;
}

```

## 格式化输出

```c

#include <iostream>
#include <iomanip>
using namespace std;

int main(){

    double price = 19.5;
    cout << fixed << setprecision(2) << price << endl; // 保留两位小数，输出 19.50
    cout << setw(10) << setfill('*') << price << endl; // 以右对齐的方式输出，总宽度固定为 10 个字符，不足的部分用星号 * 填充。

    return 0;
}

```
