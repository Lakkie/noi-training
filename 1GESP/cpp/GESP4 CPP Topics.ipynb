# 4.2 形参与实参、作用域  

形参（Formal Parameter）‌

定义‌：在‌函数定义‌时括号内声明的变量名。
特点‌：它只是一个占位符，没有实际值，仅在函数内部有效。
示例‌：def add(a, b): 中的 a 和 b 就是形参。

实参（Actual Argument）‌

定义‌：在‌函数调用‌时传递给函数的具体数据或变量。
特点‌：具有实际的值，用于初始化对应的形参。
示例‌：add(3, 5) 中的 3 和 5 就是实参。

作用域（Scope）
作用域决定了变量在代码中的‌可见范围‌和‌生命周期‌。  

全局作用域‌：定义在所有函数之外的标识符，生命周期贯穿整个程序运行，所有函数都可访问。  
局部作用域‌：函数内部定义的变量，仅在当前函数内可见，函数执行完毕后自动销毁。  
块作用域‌：{}包裹的代码块内定义的变量，仅在该代码块内有效，出块后立即销毁。  
命名空间作用域‌：命名空间内的标识符，需通过命名空间名::标识符的方式访问。  
类作用域‌：类的成员变量和成员函数，需通过类对象或类名访问。  

C++的变量查找遵循自定义的作用域优先级逻辑：  

优先查找当前局部作用域‌：函数内部、代码块内定义的变量优先级最高，找到即停止查找。  
逐级向外层作用域回溯‌：若当前局部域未找到，会向外层嵌套的块作用域、函数作用域逐层查找。  
全局作用域兜底‌：外层所有局部域都未找到时，才会查找全局作用域的变量。  
特殊域需显式指定‌：命名空间域、类域的变量不会被编译器自动向上查找，必须通过命名空间名::变量名、对象.成员名的方式显式指定访问。  

# 4.3 C++指针类型的概念及基本应用  

指针(Pointer)是 C++ 中的一种变量类型，它的作用是存储另一个变量的内存地址。简单来说，指针可以“指向”其他变量的内存位置。每个变量都有一个固定的内存地址，指针的值就是该内存地址。  

取地址符 &：用于获取变量的内存地址。例如，&a 表示获取变量 a 的内存地址。  
解引用符 *：用于访问指针所指向的内存位置中的值。例如，*p 表示访问指针 p 所指向的变量的值。

指针的声明与普通变量相似，但需要在变量类型前加上 * 符号（解引用符）。比如，int* p 声明了一个指向 int 类型的指针变量 p。  

```c
#include <iostream>
using namespace std;

int main()
{
    int a = 10;               // 声明一个整数变量a并赋值为10
    int* p = &a;              // 声明一个指针p，存储a的地址
    
    cout << *p << endl;       // 通过指针p解引用，输出a的值（即10）
    cout << a << endl;        // 输出变量a的值（即10）
    cout << "---------" << endl;
    cout << &p << endl;       // 输出指针p存储的地址，即指针p本身的地址
    cout << "---------" << endl;
    cout << &a << endl;       // 输出变量a的地址
    cout << *&p << endl;      // 解引用指针p的地址，输出指针p存储的值，即a的地址
    cout << p << endl;        // 输出指针p存储的值，即a的地址
    
    return 0;	
}
```

```c
// 控制台输出  
10                            // *p：通过解引用指针 p，访问指针所指向地址的值，即 a 的值，输出为 10。
10                            // a：直接输出变量 a 的值，结果为 10。
---------
0x70fe10                      // &p：输出指针 p 存储的地址，即指针变量 p 在内存中的位置。
---------
0x70felc                      // &a：输出变量 a 的地址，表示 a 存储在内存中的位置。
0x70felc                      // *&p：这实际上是先取 p 的地址，再解引用其值。结果是输出 p 所指向的地址，即 a 的地址。
0x70felc                      // p：直接输出指针 p 存储的值，表示 a 的内存地址。
```

空指针‌：一个指针初始化为nullptr（C++11及以后版本）或NULL（在C++中通常等同于0或nullptr），表示它不指向任何东西。这是一种好的编程习惯，可以避免野指针的问题。

‌野指针‌：未初始化或已经释放的指针称为野指针。使用野指针访问内存可能会导致未定义行为。始终确保在使用指针之前初始化它们。

# 4.4 函数参数传递的概念（C++值传递、引用传递、指针传递）  

值传递‌：实参将数据副本传给形参，函数内修改形参完全不影响外部实参。

引用传递‌：形参作为实参的别名，函数内对形参的修改会直接同步到外部实参。

‌指针传递‌：形参接收实参的内存地址，通过解引用操作间接修改外部实参的值。

# 4.5 C++结构体  

C++结构体是一种自定义复合数据类型，核心作用是将不同类型的数据成员打包成一个整体，方便统一管理和操作。

## 结构体定义语法

基础定义格式‌：使用struct关键字声明结构体，花括号内定义成员列表，末尾必须加分号。

```c

struct 结构体名 {
    成员类型 成员名;
    成员类型 成员名;
    // 可包含任意数量不同类型的成员
};

```

匿名结构体语法‌：省略结构体名，直接在定义后声明变量，仅能在当前作用域使用。

```c

struct {
    成员类型 成员名;
    成员类型 成员名;
    // 可包含任意数量不同类型的成员
} 变量名; // 直接定义匿名结构体变量


```

带成员函数的扩展定义‌：C++允许在结构体内部直接定义成员方法，和类的语法特性一致。

```c

struct 结构体名 {
    成员类型 成员名;
    成员类型 成员名;
    // 可包含任意数量不同类型的成员
    返回类型 函数名(参数列表) { // 结构体内部定义成员函数
        函数体;
        return 返回值;
    }
}; // 直接定义匿名结构体变量


```

## 结构体变量声明语法

先定义结构体类型，再单独声明变量：Student s1;，C++中声明时可省略struct关键字。  
定义结构体的同时直接声明变量：struct Student { ... } s2;。  
使用别名声明变量：通过typedef或using定义类型别名后，用别名声明变量。  

结构体指针：Student* p = &s1;，指向已存在的结构体变量地址。  
动态分配结构体：Student* p = new Student;，在堆区开辟结构体内存。  
结构体数组：Student stuArr;，一次性定义多个同类型结构体变量。
