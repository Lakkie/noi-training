```c
//B4360 [GESP202506 四级] 画布裁剪
#include<iostream>
#include<vector>
#include<string>
using namespace std;
int main(){
    //输入
    ios::sync_with_stdio(false);
    cin.tie(NULL);
    int H,W;
    if(!(cin>>H>>W)){
        return 0;
    }
    int x1,x2,y1,y2;
    cin>>x1>>x2>>y1>>y2;
    vector<string> grid(H);
    for(int i=0;i<H;++i){
        cin>>grid[i];
    }
    //计算&输出
    int start_row=x1-1;
    int end_row=x2-1;
    int start_col=y1-1;
    int sub_len=y2-y1+1;//
    for (int i=start_row;i<=end_row;++i){
        cout<<grid[i].substr(start_col,sub_len)<<endl;
    }
    return 0;
}
```
