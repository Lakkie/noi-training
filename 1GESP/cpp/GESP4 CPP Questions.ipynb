```c
//B4501 [GESP202603 四级] 山之谷 标准边界判断
#include<iostream>
#include<vector>
using namespace std;
const int MAXN=105;//定义最大行列数，稍微开大一点防止越界。
int n,m;
int h[MAXN][MAXN];
//8个方向的偏移量：dx和dy对应
//顺序可以是任意，只要覆盖8个方向即可
//这里对应：左上, 上, 右上, 左, 右, 左下, 下, 右下
int dx[]={-1, -1, -1, 0, 0, 1, 1, 1};
int dy[]={-1, 0, 1, -1, 1, -1, 0, 1};
bool isValley(int x,int y){
    for(int k=0;k<8;++k){//遍历8个方向
        int nx=x+dx[k];
        int ny=y+dy[k];
        if(nx>=1&&nx<=n&&ny>=1&&ny<=m){//检查邻居是否在网格范围内。
            if(h[x][y]>h[nx][ny]){//如果当前单元格海拔大于邻居海拔，则不满足“不高于所有邻居”的条件。
                return false;
            }
        }
    }
    return true;//所有合法邻居都不低于当前单元格，则是山谷。
}
int main(){
	//输入
    ios::sync_with_stdio(false);
    cin.tie(NULL);
    if(!(cin>>n>>m)){
		return 0;
	}
    for(int i=1;i<=n;++i){//读入实际数据，从下标1开始。
        for(int j=1;j<=m;++j){
            cin>>h[i][j];
        }
    }
    //计算
    int count=0;
    for(int i=1;i<=n;++i){//遍历每个单元格。
        for(int j=1;j<=m;++j){
            if(isValley(i,j)){
                count++;
            }
        }
    }
	//输出
    cout<<count<<endl;
    return 0;
}
```

```c
//B4501 [GESP202603 四级] 山之谷 哨兵算法‌
#include<iostream>
#include<algorithm>
using namespace std;
const int MAXN=105;
const int INF=1e5+5;//足够大的数，大于任何可能的海拔。
int n,m;
int h[MAXN][MAXN];
int dx[]={-1,-1,-1, 0, 0, 1, 1, 1};
int dy[]={-1, 0, 1,-1, 1,-1, 0, 1};
bool isValley(int i,int j){
    for(int k=0;k<8;++k){
        int nx=i+dx[k];
        int ny=j+dy[k];
        if(h[i][j]>h[nx][ny]){//因为周围已经填了INF，所以不需要判断边界，直接比较。
            return false;
        }
    }
	return true;
}
int main(){
	//输入
    ios::sync_with_stdio(false);
    cin.tie(NULL);
    cin>>n>>m;
    for(int i=0;i<=n+1;++i){//初始化整个数组为INF，包括边界外的“围墙”。注意：数组大小要能容纳 n+2 行 m+2 列
        for(int j=0;j<=m+1;++j){
            h[i][j]=INF;
        }
    }
    for(int i=1;i<=n;++i){//读入实际数据，从下标1开始。
        for(int j=1;j<=m;++j){
            cin>>h[i][j];
        }
    }
    //计算
    int count=0;
    for(int i=1;i<=n;++i){//遍历实际网格区域。
        for(int j=1;j<=m;++j){
            bool is_valley=true;
            is_valley=isValley(i,j);
            if(is_valley){
                count++;
            }
        }
    }
    //输出
    cout<<count<<endl;
    return 0;
}
```

```c
//B4502 [GESP202603 四级] 礼盒排序
#include<iostream>
#include<vector>
#include<algorithm>
using namespace std;
struct Box{      //定义礼盒结构体
    int id;      //礼盒编号
    int sum;     //总价格
    int max_val; //最贵价格
    int min_val; //最便宜价格
};
//自定义比较函数
bool compareBoxes(const Box& a,const Box& b){
    if(a.sum!=b.sum){//按总价格从小到大排序
        return a.sum<b.sum;
    }
    if(a.max_val!=b.max_val){//总价格相同，则按最贵价格从小到大排序
        return a.max_val<b.max_val;
    }
    if(a.min_val!=b.min_val){//总价格和最贵价格都相同，则按最便宜价格从小到大排序
        return a.min_val<b.min_val;
    }
    return a.id<b.id;//总价格、最贵价格和最便宜价格都相同，则按礼盒编号从小到大排序
}
int main(){
    //输入
    ios::sync_with_stdio(false);
    cin.tie(nullptr);
    int n,k;
    if(!(cin>>n>>k)){
        return 0;
    }
    vector<Box> boxes(n);
    for(int i=0;i<n;++i){
        int current_sum=0;
        int current_max=0;     //商品价格 >= 1，初始化为0即可
        int current_min=10005; //商品价格 <= 10^4，初始化为一个较大值
        for(int j=0;j<k;++j){
            int price;
            cin>>price;
            current_sum+=price;
            if(price>current_max){
                current_max=price;
            }
            if(price<current_min){
                current_min=price;
            }
        }
        boxes[i].id=i+1; // 编号从1开始
        boxes[i].sum=current_sum;
        boxes[i].max_val=current_max;
        boxes[i].min_val=current_min;
    }
    //计算
    sort(boxes.begin(),boxes.end(),compareBoxes);
    //输出
    for(int i=0;i<n;++i){
        if(i>0){
            cout<<" ";
        }
        cout<<boxes[i].id;
    }
    cout<<endl;
    return 0;
}
```

```c
//B4557 [GESP202606 四级] 扫雷
#include<iostream>
#include<vector>
using namespace std;
const int MAXN=505;
int mine[MAXN][MAXN];
int dx[]={-1,-1,-1, 0, 0, 1, 1, 1};
int dy[]={-1, 0, 1,-1, 1,-1, 0, 1};
int main(){
    //输入
    ios::sync_with_stdio(false);
    cin.tie(NULL);
    int n,m,q;
    if(!(cin>>n>>m>>q)){
        return 0;
    }
    for(int i=0;i<q;++i){
        int x,y;
        cin>>x>>y;
        mine[x][y]=1;
    }
    //计算&输出
    for(int i=1; i<=n;++i){
        for(int j=1;j<=m;++j){
            if(mine[i][j]==1){//如果当前位置是雷
                cout<<"* ";
            }else{//如果当前位置不是雷
                int cnt=0;//计数器
                for(int k=0;k<8;++k){//统计周围8个方向的雷数
                    int nx=i+dx[k];
                    int ny=j+dy[k];
                    if(nx>=1&&nx<=n&&ny>=1&&ny<=m){//判断是否越界
                        if(mine[nx][ny]==1){//如果相邻位置是雷，计数器加1
                            cnt++;
                        }
                    }
                }
                cout<<cnt<<" ";
            }
        }
        cout<<endl;
    }
    return 0;
}
```

```c
//B4558 [GESP202606 四级] 身高体重指数
#include<iostream>
#include<vector>
#include<algorithm>
#include<iomanip>
using namespace std;
struct Person {    //定义结构体存储小朋友信息
    int id;        //编号
    int weight;    //体重
    double height; //身高
    double bmi;    //身高体重指数
};
//自定义比较函数
bool comparePersons(const Person& a,const Person& b){
    return a.bmi>b.bmi;//按 BMI 从高到低排序
}
int main(){
    //输入
    ios::sync_with_stdio(false);
    cin.tie(nullptr);
    int n;
    if(!(cin>>n)){
        return 0;
    }
    vector<Person> people(n);
    for(int i=0;i<n;++i){
        cin>>people[i].weight;
        people[i].id=i+1;//分配编号，从1开始
    }
    for(int i=0;i<n;++i){
        cin>>people[i].height;
        people[i].bmi=static_cast<double>(people[i].weight)/(people[i].height*people[i].height);
    }
    //计算
    sort(people.begin(),people.end(),comparePersons);
    //输出
    for(int i=0;i<n;++i){
        if(i>0){
            cout<<" ";
        }
        cout<<people[i].id;
    }
    cout<<endl;
    return 0;
}
```

```c
//B4263 [GESP202503 四级] 荒地开垦
#include<iostream>
#include<vector>
#include<algorithm>
#include<string>
using namespace std;
//方向数组：上、下、左、右
const int dx[]={-1, 1,  0, 0};
const int dy[]={ 0, 0, -1, 1};
int n,m;
vector<string> grid;
//判断在当前的 grid 状态下，位置 (x, y) 是否可开垦
bool isReclaimable(int x,int y){
    if(grid[x][y]=='#'){//如果当前位置是杂物，直接不可开垦
        return false;
    }
    for(int k=0;k<4;++k){//检查四个方向
        int nx=x+dx[k];
        int ny=y+dy[k];
        if(nx>=0&&nx<n&&ny>=0&&ny<m){//如果相邻格子在地图范围内
            if(grid[nx][ny]=='#'){//如果相邻格子是杂物，则当前格子不可开垦
                return false;
            }
        }
    }
    return true;
}
int main(){
    //输入
    ios::sync_with_stdio(false);
    cin.tie(nullptr);
    if(!(cin>>n>>m)){
        return 0;
    }
    grid.resize(n);
    for(int i=0;i<n;++i){
        cin>>grid[i];
    }
    //计算
    int base=0;//初始基础可开垦数量
    vector<vector<bool>> valid(n,vector<bool>(m,false));//valid[i][j]记录初始状态下(i,j)是否可开垦
    for(int i=0;i<n;++i){
        for(int j=0;j<m;++j){
            if(isReclaimable(i,j)){
                valid[i][j]=true;
                base++;
            }
        }
    }
    int max_gain=0;
    for(int i=0;i<n;++i){
        for(int j=0;j<m;++j){
            if(grid[i][j]!='#'){
                continue;//只处理杂物
            }
            vector<pair<int,int>> affected;//收集受影响的5个位置：自身和4个邻居
            affected.push_back({i,j});//自身
            for(int k=0;k<4;++k){//4个邻居
                int nx=i+dx[k];
                int ny=j+dy[k];
                if(nx>=0&&nx<n&&ny>=0&&ny<m){
                    affected.push_back({nx,ny});
                }
            }
            int old_count=0;//计算这5个位置在初始状态下的可开垦数量
            for(auto& pos:affected){
                if(valid[pos.first][pos.second]){
                    old_count++;
                }
            }
            grid[i][j]='.';//模拟清除杂物：临时改为'.'
            int new_count=0;//计算这5个位置在新状态下的可开垦数量
            for(auto& pos:affected){
                if(isReclaimable(pos.first,pos.second)){
                    new_count++;
                }
            }
            grid[i][j]='#';//恢复现场：临时改为'.'的还原为'#'
            int gain=new_count-old_count;//计算增益
            if(gain>max_gain){
                max_gain=gain;
            }
        }
    }
    //输出
    cout<<base+max_gain<<endl;
    return 0;
}
```

```c
//B4264 [GESP202503 四级] 二阶矩阵
#include<iostream>
#include<vector>
using namespace std;
int main(){
    //输入
    ios::sync_with_stdio(false);
    cin.tie(nullptr);
    int n,m;
    if(!(cin>>n>>m)){
        return 0;
    }
    vector<vector<int>> a(n,vector<int>(m));
    for(int i=0;i<n;++i){
        for(int j=0;j<m;++j){
            cin>>a[i][j];
        }
    }
    //计算
    int ans=0;
    for(int i=0;i<=n-2;++i){
        for(int j=0;j<=m-2;++j){
            int d11=a[i][j];     //左上
            int d12=a[i][j+1];   //右上
            int d21=a[i+1][j];   //左下
            int d22=a[i+1][j+1]; //右下
            if(d11*d22==d12*d21){
                ans++;
            }
        }
    }
    //输出
    cout << ans << endl;
    return 0;
}
```

```c
//B4361 [GESP202506 四级] 排序
#include<iostream>
#include<vector>
#include<algorithm>
using namespace std;
struct Student{  //定义学生结构体
    long long h; //身高
    long long w; //体重
};
//自定义比较函数
bool isBefore(const Student& a,const Student& b){
    if(a.h!=b.h){
        return a.h>b.h; //身高降序
    }
    return a.w>b.w;     //体重降序
}
int main(){
    //输入
    ios::sync_with_stdio(false);
    cin.tie(nullptr);
    int n;
    if(!(cin>>n)){
        return 0;
    }
    vector<Student> students(n);
    for(int i=0;i<n;++i){
        cin>>students[i].h>>students[i].w;
    }
    //计算
    long long swaps=0; 
    for(int i=0;i<n;++i){
        for (int j=i+1;j<n;++j){
            if(isBefore(students[j],students[i])){
                swaps++;
            }
        }
    }
    //输出
    cout<<swaps<<endl;
    return 0;
}
```

```c
//B4415 [GESP202509 四级] 排兵布阵 暴力枚举
#include<iostream>
#include<algorithm>
using namespace std;
const int N=15;
int a[N][N];
int main(){
    //输入
    ios::sync_with_stdio(false);
    cin.tie(nullptr);
    int n,m;
    cin>>n>>m;
    for(int i=1;i<=n;++i){
        for(int j=1;j<=m;++j){
            cin>>a[i][j];
        }
    }
    //计算
    int max_area=0;
    for(int i=1;i<=n;++i){//遍历左上角(i,j)
        for(int j=1;j<=m;++j){
            for(int p=i;p<=n;++p){//遍历右下角(p,q)
                for(int q=j;q<=m;++q){
                    bool is_all_one=true;
                    for(int x=i;x<=p;++x){//检查矩形(i,j)到(p,q)是否全为1
                        for(int y=j;y<=q;++y){
                            if(a[x][y]==0){
                                is_all_one=false;
                                break;//发现一个0，立即跳出内层循环
                            }
                        }
                        if(!is_all_one){//跳出外层检查循环
                            break;
                        }
                    }
                    if(is_all_one){
                        int area=(p-i+1)*(q-j+1);
                        max_area=max(max_area,area);
                    }
                }
            }
        }
    }
    //输出
    cout<<max_area<<endl;
    return 0;
}

```

```c
//B4415 [GESP202509 四级] 排兵布阵 二维前缀和
#include<iostream>
#include<algorithm>
using namespace std;
const int N=15;
int a[N][N];//原始矩阵
int s[N][N];//二维前缀和
int main(){
    //输入
    ios::sync_with_stdio(false);
    cin.tie(nullptr);
    int n,m;
    if(!(cin>>n>>m)){
        return 0;
    }
    //计算
    for(int i=1;i<=n;++i){
        for(int j=1;j<=m;++j){
            cin>>a[i][j];
            s[i][j]=s[i-1][j]+s[i][j-1]-s[i-1][j-1]+a[i][j];//前缀和公式: s[i][j]=上方+左方-左上角+当前值
        }
    }
    int max_area=0;
    for(int x1=1;x1<=n;++x1){//左上角坐标(x1,y1)
        for(int y1=1;y1<=m;++y1){
            for(int x2=x1;x2<=n;++x2){//右下角坐标(x2,y2)
                for(int y2=y1;y2<=m;++y2){
                    int area=(x2-x1+1)*(y2-y1+1);//计算当前子矩形的面积
                    int sum_ones=s[x2][y2]-s[x1-1][y2]-s[x2][y1-1]+s[x1-1][y1-1];//利用前缀和计算子矩形内1的个数
                    if(sum_ones==area){//如果1的个数等于面积，说明全是1
                        max_area=max(max_area,area);
                    }
                }
            }
        }
    }
    //输出
    cout<<max_area<<endl;
    return 0;
}
```
