```c
//B4501 [GESP202603 四级] 山之谷 标准边界判断
#include<iostream>
#include<vector>
using namespace std;
const int MAXN=105;//定义最大行列数，稍微开大一点防止越界。
int n,m;
int h[MAXN][MAXN];
//8个方向的偏移量：dx和dy对应
//顺序可以是任意，只要覆盖8个方向即可
//这里对应：左上, 上, 右上, 左, 右, 左下, 下, 右下
int dx[]={-1, -1, -1, 0, 0, 1, 1, 1};
int dy[]={-1, 0, 1, -1, 1, -1, 0, 1};
bool isValley(int x,int y){
    for(int k=0;k<8;++k){//遍历8个方向
        int nx=x+dx[k];
        int ny=y+dy[k];
        if(nx>=1&&nx<=n&&ny>=1&&ny<=m){//检查邻居是否在网格范围内。
            if(h[x][y]>h[nx][ny]){//如果当前单元格海拔大于邻居海拔，则不满足“不高于所有邻居”的条件。
                return false;
            }
        }
    }
    return true;//所有合法邻居都不低于当前单元格，则是山谷。
}
int main(){
	//输入
    ios::sync_with_stdio(false);
    cin.tie(NULL);
    if(!(cin>>n>>m)){
		return 0;
	}
    for(int i=1;i<=n;++i){//读入实际数据，从下标1开始。
        for(int j=1;j<=m;++j){
            cin>>h[i][j];
        }
    }
    //计算
    int count=0;
    for(int i=1;i<=n;++i){//遍历每个单元格。
        for(int j=1;j<=m;++j){
            if(isValley(i,j)){
                count++;
            }
        }
    }
	//输出
    cout<<count<<endl;
    return 0;
}
```

```c
//B4502 [GESP202603 四级] 礼盒排序
#include<iostream>
#include<vector>
#include<algorithm>
using namespace std;
struct Box{      //定义礼盒结构体
    int id;      //礼盒编号
    int sum;     //总价格
    int max_val; //最贵价格
    int min_val; //最便宜价格
};
//自定义比较函数
bool compareBoxes(const Box& a,const Box& b){
    if(a.sum!=b.sum){//按总价格从小到大排序
        return a.sum<b.sum;
    }
    if(a.max_val!=b.max_val){//总价格相同，则按最贵价格从小到大排序
        return a.max_val<b.max_val;
    }
    if(a.min_val!=b.min_val){//总价格和最贵价格都相同，则按最便宜价格从小到大排序
        return a.min_val<b.min_val;
    }
    return a.id<b.id;//总价格、最贵价格和最便宜价格都相同，则按礼盒编号从小到大排序
}
int main(){
    //输入
    ios::sync_with_stdio(false);
    cin.tie(nullptr);
    int n,k;
    if(!(cin>>n>>k)){
        return 0;
    }
    vector<Box> boxes(n);
    for(int i=0;i<n;++i){
        int current_sum=0;
        int current_max=0;     //商品价格 >= 1，初始化为0即可
        int current_min=10005; //商品价格 <= 10^4，初始化为一个较大值
        for(int j=0;j<k;++j){
            int price;
            cin>>price;
            current_sum+=price;
            if(price>current_max){
                current_max=price;
            }
            if(price<current_min){
                current_min=price;
            }
        }
        boxes[i].id=i+1; // 编号从1开始
        boxes[i].sum=current_sum;
        boxes[i].max_val=current_max;
        boxes[i].min_val=current_min;
    }
    //计算
    sort(boxes.begin(),boxes.end(),compareBoxes);
    //输出
    for(int i=0;i<n;++i){
        if(i>0){
            cout<<" ";
        }
        cout<<boxes[i].id;
    }
    cout<<endl;
    return 0;
}
```

```c
//B4263 [GESP202503 四级] 荒地开垦
#include<iostream>
#include<vector>
#include<algorithm>
#include<string>
using namespace std;
//方向数组：上、下、左、右
const int dx[]={-1, 1,  0, 0};
const int dy[]={ 0, 0, -1, 1};
int n,m;
vector<string> grid;
//判断在当前的 grid 状态下，位置 (x, y) 是否可开垦
bool isReclaimable(int x,int y){
    if(grid[x][y]=='#'){//如果当前位置是杂物，直接不可开垦
        return false;
    }
    for(int k=0;k<4;++k){//检查四个方向
        int nx=x+dx[k];
        int ny=y+dy[k];
        if(nx>=0&&nx<n&&ny>=0&&ny<m){//如果相邻格子在地图范围内
            if(grid[nx][ny]=='#'){//如果相邻格子是杂物，则当前格子不可开垦
                return false;
            }
        }
    }
    return true;
}
int main(){
    //输入
    ios::sync_with_stdio(false);
    cin.tie(nullptr);
    if(!(cin>>n>>m)){
        return 0;
    }
    grid.resize(n);
    for(int i=0;i<n;++i){
        cin>>grid[i];
    }
    //计算
    int base=0;//初始基础可开垦数量
    vector<vector<bool>> valid(n,vector<bool>(m,false));//valid[i][j]记录初始状态下(i,j)是否可开垦
    for(int i=0;i<n;++i){
        for(int j=0;j<m;++j){
            if(isReclaimable(i,j)){
                valid[i][j]=true;
                base++;
            }
        }
    }
    int max_gain=0;
    for(int i=0;i<n;++i){
        for(int j=0;j<m;++j){
            if(grid[i][j]!='#'){
                continue;//只处理杂物
            }
            vector<pair<int,int>> affected;//收集受影响的5个位置：自身和4个邻居
            affected.push_back({i,j});//自身
            for(int k=0;k<4;++k){//4个邻居
                int nx=i+dx[k];
                int ny=j+dy[k];
                if(nx>=0&&nx<n&&ny>=0&&ny<m){
                    affected.push_back({nx,ny});
                }
            }
            int old_count=0;//计算这5个位置在初始状态下的可开垦数量
            for(auto& pos:affected){
                if(valid[pos.first][pos.second]){
                    old_count++;
                }
            }
            grid[i][j]='.';//模拟清除杂物：临时改为'.'
            int new_count=0;//计算这5个位置在新状态下的可开垦数量
            for(auto& pos:affected){
                if(isReclaimable(pos.first,pos.second)){
                    new_count++;
                }
            }
            grid[i][j]='#';//恢复现场：临时改为'.'的还原为'#'
            int gain=new_count-old_count;//计算增益
            if(gain>max_gain){
                max_gain=gain;
            }
        }
    }
    //输出
    cout<<base+max_gain<<endl;
    return 0;
}
```

```c
//B4415 [GESP202509 四级] 排兵布阵 暴力枚举
#include<iostream>
#include<algorithm>
using namespace std;
const int N=15;
int a[N][N];
int main(){
    //输入
    ios::sync_with_stdio(false);
    cin.tie(nullptr);
    int n,m;
    cin>>n>>m;
    for(int i=1;i<=n;++i){
        for(int j=1;j<=m;++j){
            cin>>a[i][j];
        }
    }
    //计算
    int max_area=0;
    for(int i=1;i<=n;++i){//遍历左上角(i,j)
        for(int j=1;j<=m;++j){
            for(int p=i;p<=n;++p){//遍历右下角(p,q)
                for(int q=j;q<=m;++q){
                    bool is_all_one=true;
                    for(int x=i;x<=p;++x){//检查矩形(i,j)到(p,q)是否全为1
                        for(int y=j;y<=q;++y){
                            if(a[x][y]==0){
                                is_all_one=false;
                                break;//发现一个0，立即跳出内层循环
                            }
                        }
                        if(!is_all_one){//跳出外层检查循环
                            break;
                        }
                    }
                    if(is_all_one){
                        int area=(p-i+1)*(q-j+1);
                        max_area=max(max_area,area);
                    }
                }
            }
        }
    }
    //输出
    cout<<max_area<<endl;
    return 0;
}

```

```c
//B4415 [GESP202509 四级] 排兵布阵 二维前缀和
#include<iostream>
#include<algorithm>
using namespace std;
const int N=15;
int a[N][N];//原始矩阵
int s[N][N];//二维前缀和
int main(){
    //输入
    ios::sync_with_stdio(false);
    cin.tie(nullptr);
    int n,m;
    if(!(cin>>n>>m)){
        return 0;
    }
    //计算
    for(int i=1;i<=n;++i){
        for(int j=1;j<=m;++j){
            cin>>a[i][j];
            s[i][j]=s[i-1][j]+s[i][j-1]-s[i-1][j-1]+a[i][j];//前缀和公式: s[i][j]=上方+左方-左上角+当前值
        }
    }
    int max_area=0;
    for(int x1=1;x1<=n;++x1){//左上角坐标(x1,y1)
        for(int y1=1;y1<=m;++y1){
            for(int x2=x1;x2<=n;++x2){//右下角坐标(x2,y2)
                for(int y2=y1;y2<=m;++y2){
                    int area=(x2-x1+1)*(y2-y1+1);//计算当前子矩形的面积
                    int sum_ones=s[x2][y2]-s[x1-1][y2]-s[x2][y1-1]+s[x1-1][y1-1];//利用前缀和计算子矩形内1的个数
                    if(sum_ones==area){//如果1的个数等于面积，说明全是1
                        max_area=max(max_area,area);
                    }
                }
            }
        }
    }
    //输出
    cout<<max_area<<endl;
    return 0;
}
```

```c
//B4416 [GESP202509 四级] 最长连续段 排序法
#include<iostream>
#include<vector>
#include<algorithm>
#include<set>
using namespace std;
int main(){
    //输入
    ios::sync_with_stdio(false);
    cin.tie(nullptr);
    int n;
    if(!(cin >> n)){
        return 0;
    }
    vector<long long> a(n);
    for(int i=0;i<n;++i){
        cin>>a[i];
    }
    //计算
    sort(a.begin(),a.end());
    auto last=unique(a.begin(),a.end());//unique将重复元素移到末尾，返回指向第一个重复元素的迭代器
    a.erase(last,a.end());//erase删除末尾的重复元素
    int m=a.size();//现在a中包含的是去重后且有序的元素
    if(m==0){
        cout<<0<<endl;
        return 0;
    }
    int max_len=1;//至少有一个元素，长度为1
    int current_len=1;
    for(int i=1;i<m;++i){//遍历寻找最长连续段
        if(a[i]==a[i-1]+1){//如果当前元素等于前一个元素 + 1，则连续
            current_len++;
        }else{//否则断开，更新最大值，并重置当前长度
            if(current_len>max_len){
                max_len=current_len;
            }
            current_len=1;
        }
    }
    if(current_len>max_len){//最后一次比较，防止最长序列在数组末尾
        max_len=current_len;
    }
    //输出
    cout<<max_len<<endl;
    return 0;
}
```

```c
//B4416 [GESP202509 四级] 最长连续段 哈希表法
#include<iostream>
#include<vector>
#include<unordered_set>
#include<algorithm>
using namespace std;
int main(){
    //输入
    ios::sync_with_stdio(false);
    cin.tie(nullptr);
    int n;
    if(!(cin>>n)){
        return 0;
    }
    vector<long long> a(n);
    unordered_set<long long> numSet;//使用unordered_set进行去重和存储，支持O(1)查找
    for(int i=0;i<n;++i){
        cin>>a[i];
        numSet.insert(a[i]);
    }
    //计算
    int max_len=0;
    for(long long num:numSet){//遍历集合中的每一个唯一元素
        if(numSet.find(num-1)==numSet.end()){
            long long current_num=num;
            int current_len=1;
            while(numSet.find(current_num+1)!=numSet.end()){//从起点开始，向后查找连续的数字
                current_num++;
                current_len++;
            }
            if(current_len>max_len){//更新最大长度
                max_len=current_len;
            }
        }
    }
    //输出
    cout<<max_len<<endl;
    return 0;
}
```

```c
//B4451 [GESP202512 四级] 建造
#include<iostream>
#include<vector>
#include<algorithm>
#include<climits>
using namespace std;
int main(){
    //输入
    ios::sync_with_stdio(false);
    cin.tie(nullptr);
    int M,N,H;
    if(!(cin>>M>>N>>H)){
        return 0;
    }
    vector<vector<int>> grid(M, vector<int>(N));//使用二维向量存储地形图，大小为MxN
    for(int i=0;i<M;++i){
        for(int j=0;j<N;++j){
            cin>>grid[i][j];
        }
    }
    //计算
    int max_sum=INT_MIN;//初始化为最小整数，确保能更新到第一个合法值
    bool found=false;//标记是否找到至少一个合法区域
    for(int i=0;i<=M-3;++i){//遍历所有可能的3x3子矩阵的左上角(i, j)
        for(int j=0;j<=N-3;++j){
            int current_sum=0;
            int min_val=INT_MAX;
            int max_val=INT_MIN;
            for(int x=0;x<3;++x){//遍历当前3x3区域
                for(int y=0;y<3;++y){
                    int val=grid[i+x][j+y];
                    current_sum+=val;
                    if(val<min_val){
                        min_val=val;
                    }
                    if(val>max_val){
                        max_val=val;
                    }
                }
            }
            if(max_val-min_val<=H){//检查约束条件：最大高度与最小高度之差不超过H
                found=true;
                if(current_sum>max_sum){
                    max_sum=current_sum;
                }
            }
        }
    }
    //输出
    cout<<max_sum<<endl;
    return 0;
}
```

```c
//B4452 [GESP202512 四级] 优先购买
#include<iostream>
#include<vector>
#include<algorithm>
#include<string>
using namespace std;
struct Item{     //商品结构体
    string name; //名称
    int price;   //价格
    int val;     //优先级
};
bool cmpBuy(const Item &a,const Item &b){//购买排序规则
    if(a.val!=b.val){
        return a.val<b.val;
    }
    if(a.price!=b.price){
        return a.price<b.price;
    }
    return a.name<b.name;
}
bool cmpOutput(const Item &a,const Item &b){//输出排序规则
    return a.name<b.name;
}
int main(){
    //输入
    ios::sync_with_stdio(false);
    cin.tie(nullptr);
    int M,N;
    if(!(cin>>M>>N)){
        return 0;
    }
    vector<Item> goods(N);
    for(int i=0;i<N;i++){
        cin>>goods[i].name>>goods[i].price>>goods[i].val;
    }
    //计算
    sort(goods.begin(),goods.end(),cmpBuy);
    vector<Item> bought;
    int rest_money=M;
    for(auto &it:goods){//贪心模拟购买过程
        if(rest_money>=it.price){
            bought.push_back(it);
            rest_money-=it.price;
        }
    }
    sort(bought.begin(),bought.end(),cmpOutput);
    //输出
    for(auto &t:bought){
        cout<<t.name<<endl;
    }
    return 0;
}
```

```c
//B3850 [GESP202306 四级] 幸运数
#include <iostream>
using namespace std;
//变换函数：将数字n乘以7，并不断求各位和直到结果<=9
int trans(int n){
    int k=n*7;
    while(k>9){
        int sum=0;
        while(k>0){
            sum+=k%10;
            k/=10;
        }
        k=sum;
    }
    return k;
}
bool isLuckyNumer(long long n){
    int sum=0;
    for(int d=1;n>0;++d){
        int t=(int)(n%10);
        if(d%2==0){
            sum+=t;
        }else{
            sum+=trans(t);
        }
        n/=10;
    }
    return (sum%8==0);
}
int main(){
    //输入
    int n=0;
    cin>>n;
    for(int i=0;i<n;++i){
        long long x=0;
        cin>>x;
        //计算&输出
        if(isLuckyNumer(x)){
            cout<<"T"<<endl;
        }else{
            cout<<"F"<<endl;
        }
    }
    return 0;
}
```

```c
//B3851 [GESP202306 四级] 图像压缩
#include<iostream>
#include<vector>
#include<algorithm>
#include<string>
#include<cmath>
using namespace std;
int palette[16];//存储选出的16个关键灰阶值
int mapToPalette[256];//存储映射关系，避免重复计算
int charToInt(char c){//辅助函数：字符转整数
    return (c>='0'&&c<='9')?c-'0':c-'A'+10;
}
char intToChar(int v){//辅助函数：整数(0-15)转十六进制字符
    return (v<10)?'0'+v:'A'+v-10;
}
int main(){
    //输入
    ios::sync_with_stdio(false);
    cin.tie(nullptr);
    int n;
    if(!(cin>>n)){
        return 0;
    }
    vector<string> lines(n);
    int freq[256]={0};//频率统计数组
    for(int i=0;i<n;++i){
        cin>>lines[i];
        //计算
        for(int j=0;j<lines[i].size();j+=2){//统计频率
            int val=charToInt(lines[i][j])*16+charToInt(lines[i][j+1]);
            freq[val]++;
        }
    }
    vector<pair<int,int>> candidates;//排序数据：{频率*-1, 灰度值}
    for(int i=0;i<256;++i){
        if(freq[i]>0){
            candidates.push_back({-freq[i],i});
        }
    }
    sort(candidates.begin(),candidates.end());//排序：频率高的在前（即负数小的在前），频率相同灰度值小的在前
    for(int i=0;i<16;++i){//提取前16个关键灰阶
        palette[i]=candidates[i].second;
        //输出第一行
        cout<<intToChar(palette[i]/16)<<intToChar(palette[i]%16);
    }
    cout<<endl;
    for(int i=0;i<256;++i){//预计算映射表：对于0-255每个值，找到最近的palette索引
        int minDist=1000;
        int bestIdx=0;
        for(int j=0;j<16;++j){
            int dist=abs(i-palette[j]);
            if(dist<minDist){//如果距离更小，更新最小距离
                minDist=dist;
                bestIdx=j;
            }
        }
        mapToPalette[i]=bestIdx;
    }
    //输出压缩后的图像
    for(const auto& line:lines){
        for(int j=0;j<line.size();j+=2){
            int val=charToInt(line[j])*16+charToInt(line[j+1]);
            cout<<intToChar(mapToPalette[val]);
        }
        cout<<endl;
    }
    return 0;
}
```
