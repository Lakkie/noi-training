# 什么是变量？

在 C++ 中，变量是用于在内存中存储数据的容器。程序通过变量来保存、访问和操作数据。

## 变量的定义（声明）

数据类型 变量名 = 初始值;

数据类型，决定了变量能存储什么样的数据以及占用多少内存。
变量名，变量的唯一标识符。
可选的初始值。

```c

int age;          // 声明一个整型变量，未初始化
int score = 100;  // 声明并初始化为 100
int x, y, z;      // 同时声明多个同类型的变量


```

## 变量的使用

变量的命名规则（必须遵守）：‌

名称只能包含字母（a-z, A-Z）、数字（0-9）和下划线（_）。
名称必须以字母或下划线开头，不能以数字开头。
名称区分大小写（myVar 和 myvar 是不同的变量）。
不能使用 C++ 关键字（如 int, return, class 等）作为变量名。

变量命名的最佳实践：‌

使用有意义的名称‌：例如使用 studentAge 而不是 a，以提高代码可读性。
驼峰命名法‌：通常变量名首字母小写，后续单词首字母大写（如 myVariableName）。

# 基本数据类型

## 整型 Integer Types 

用于存储整数（没有小数部分的数字）

```c

short number_short = 0;         // 2个字节，-32,768‌ 到 ‌32,767‌ 之间， 即-2^15 ~ 2^15-1
int number_int = 0;             // 4个字节，-2,147,483,648 到 2,147,483,647 之间， 即 -2^31 ~ 2^31-1
long long number_long_long = 0; // 8个字节，-9223372036854775808 到 9223372036854775807 之间， 即 -2^63 ~ 2^63-1

```

## 浮点型 Floating-Point Types 

用于存储带有小数点的实数。

```c

float number_float = 0.0f;  // 单精度浮点数 4个字节，约 7 位有效数字
double number_double = 0.0; // 双精度浮点数 8个字节，约 15 位有效数字

```

## 字符型 Character Type

```c

char grade = 'A'; // 1个字节

```

## 布尔型 Boolean Type

```c

bool t = true;  // 真，通常内部表示为 1 
bool f = false; // 假，通常内部表示为 0

```

# 控制语句结构

## 顺序结构 Sequential Structure

```c

int a = 10;      // 步骤1：定义并初始化
int b = 20;      // 步骤2：定义并初始化
int sum = a + b; // 步骤3：计算结果

```

## 循环结构 Loop Structure

```c

for (int i = 0; i < 5; ++i) {
    cout << i << " "; // 输出 0 1 2 3 4
}

```

```c

int count = 0; 
while (count < 5) { 
    cout << count << " "; 
    ++count; 
}

```

```c

int input; 
do { 
    cout << "请输入正数: "; 
    cin >> input; 
} while (input <= 0); // 如果输入负数，会再次提示

```

## 选择结构 Selection Structure

```c

// 单分支 if
if (condition) {
    // 条件为 true 时执行
}

```

```c

// 双分支 if-else
if (condition) {
    // 条件为 true 时执行
} else {
    // 以上条件都不满足时执行
}

```

```c

// 多分支 if-elif-else
if (condition) {
    // 条件为 true 时执行
} else if (other_condition) {
    // 其他条件为 true 时执行
} else {
    // 以上条件都不满足时执行
}

```

```c

// 多分支 switch
switch (day) {
    case 1:
        cout << "星期一" << endl;
        break; // 必须加 break，否则会继续执行下一个 case
    case 2:
        cout << "星期二" << endl;
        break;
    case 3:
        cout << "星期三" << endl;
        break;
    case 4:
        cout << "星期四" << endl;
        break;
    case 5:
        cout << "星期五" << endl;
        break;
    default: // 当所有 case 都不匹配时执行
        cout << "其他日子" << endl;
        break;
}

```

# 基本运算

## 算术运算 Arithmetic operations

```c

int a = 10, b = 3;

cout << "加法: " << a + b << endl;
cout << "减法: " << a - b << endl;
cout << "乘法: " << a * b << endl;

cout << "除法: " << (double) a / b << endl;
cout << "整除: " << a / b << endl;
cout << "取模: " << a % b << endl;

```

## 关系运算 Relational operations

```c

int age = 20;
double score = 85.5;
char grade = 'A';

// 等于
cout << "score == 90: " << (score == 90) << endl;
// 不等于
cout << "score != 90: " << (score != 90) << endl;

// 大小比较
cout << "age > 18: " << (age > 18) << endl;
cout << "age < 18: " << (age < 18) << endl;
cout << "score >= 90: " << (score >= 90) << endl;
cout << "score <= 90: " << (score <= 90) << endl;

// 字符比较 (基于 ASCII 码)
cout << "grade >= 'B': " << (grade >= 'B') << endl; // 输出 1 (true, 'A'的ASCII值大于'B')

// 注意 浮点数比较的特殊性
double a = 0.1 + 0.2;
double b = 0.3;
cout << "\n--- 浮点陷阱 ---" << endl;
// 由于精度问题，直接比较可能不相等
cout << "0.1+0.2 == 0.3 ? " << (a == b) << endl; // 通常输出 0 (false)
// 推荐做法：比较差值的绝对值是否极小
cout << "近似相等 ? " << (abs(a - b) < 1e-9) << endl; // 输出 1 (true)

```

## 逻辑运算 Logical operations

```c

int age = 25;
bool hasTicket = true;
bool isMember = false;

// 1. 逻辑与 (&&)：必须同时满足
// 场景：年满18岁 且 有票才能入场
if (age >= 18 && hasTicket) {
    cout << "允许入场" << endl;
} else {
    cout << "禁止入场" << endl;
}

// 2. 逻辑或 (||)：满足其一即可
// 场景：是会员 或 持有优惠券均可打折
if (isMember || hasTicket) {
    cout << "享受折扣" << endl;
}

// 3. 逻辑非 (!)：取反操作
// 场景：如果“不是”会员，则提示注册
if (!isMember) {
    cout << "您还不是会员，建议注册" << endl;
}

// 4. 短路求值 (Short-circuit Evaluation)
int x = 0;
// 在 A && B 中，若 A 为假，B 不会被执行
if (x != 0 && (10 / x > 1)) { 
    cout << "计算成功" << endl;
} else {
    cout << "跳过除法，避免除零错误" << endl;
}

```

# 输入输出

## 基础输入输出

```c

#include <iostream>
using namespace std; // 实际项目中建议直接使用 std::cin/std::cout

int main(){
    // 输入 (cin)‌：使用右移运算符 >> 从键盘读取数据。
	int n,s=0;
	cin>>n;

    // 计算
    for(int i=0;i<n;i++){
        s+=i;
    }
    
    // 输出 (cout)‌：使用左移运算符 << 将数据发送到屏幕。
	cout<<s<<endl;

    return 0;
}

```

```c

#include <iostream>

int main(){
    // 输入
	int n,s=0;
	std::cin>>n;

    // 计算
    for(int i=0;i<n;i++){
        s+=i;
    }
    
    // 输出
	std::cout<<s<<std::endl;

    return 0;
}

```

## 格式化输出

```c

#include <iostream>
#include <iomanip>
using namespace std;

int main(){

    double price = 19.5;
    cout << fixed << setprecision(2) << price << endl; // 保留两位小数，输出 19.50
    cout << setw(10) << setfill('*') << price << endl; // 以右对齐的方式输出，总宽度固定为 10 个字符，不足的部分用星号 * 填充。

    return 0;
}

```
