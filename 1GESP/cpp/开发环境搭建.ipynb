# Dev C++安装

建议安装路径为纯英文字母，不要有空格、中文或者其他语言的字符。（以免有些低版本的工具路径识别报错。）

# Dev C++配置

切换语言包  
Tools -> Environment Options -> General -> Language

# Dev C++使用

1. 新建代码文件
File -> New -> Source File

2. 编译和运行代码
Compile & Run

3. 设置断点和调试代码  
断点是调试过程中的一个重要工具，它允许程序在执行到某行代码时暂停，这样你就可以检查变量的值或者程序的执行流程。  
在你想要暂停执行的代码行旁边点击左侧的空白区域，或者在代码行号上双击，会出现一个红色的圆点，这就是一个断点。
Dev C++调试功能非常难用，建议直接使用输出命令打印出需要观察的变量值。
