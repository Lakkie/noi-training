# 什么是C语言？

C语言‌是一种通用的、面向过程的计算机编程语言，由丹尼斯·里奇（Dennis Ritchie）于1972年在贝尔实验室开发，最初用于编写UNIX操作系统。  

## 核心定位‌：  
它被誉为“中级语言”，因为它既具有高级语言的易读性和结构化特征，又保留了汇编语言直接操作硬件和内存的能力。

## 主要特点‌：
高效灵活‌：编译后生成的机器码执行效率极高，接近汇编语言。  
简洁紧凑‌：关键字少（仅30余个），语法简洁，适合底层开发。  
可移植性强‌：标准C代码可以在多种硬件平台和操作系统上编译运行。  
手动内存管理‌：没有垃圾回收机制，程序员需手动分配和释放内存，这带来了高性能，但也要求开发者具备严谨的资源管理能力。  

## 应用场景‌：  
广泛用于操作系统内核（如Linux、Windows部分核心）、设备驱动程序、嵌入式系统（如单片机、物联网设备）、编译器开发以及对性能要求极高的科学计算领域。  

# 什么是C++语言？

C++‌语言是由本贾尼·斯特劳斯特卢普（Bjarne Stroustrup）于1979年在贝尔实验室开始研发，并于1983年正式命名的编程语言。它是在C语言的基础上扩展而来的，旨在保留C语言高效性的同时，增加对大型软件复杂性的管理能力。

## 核心定位‌：  
C++是一门多范式编程语言，支持面向过程、面向对象（OOP）和泛型编程。名字中的“++”源自C语言的自增运算符，寓意“在C的基础上向前推进一步”。

## 主要特点‌：  
兼容C语言‌：C++几乎完全兼容C，这意味着大多数C代码可以在C++编译器中运行。  
面向对象特性‌：引入了类（Class）、对象、继承、多态和封装等概念，使得代码更易于模块化、复用和维护，特别适合大规模软件开发。  
丰富的功能库‌：拥有强大的标准模板库（STL），提供了容器、算法和迭代器等高效工具。  
现代特性演进‌：随着C++11、C++14、C++17及后续标准的发布，引入了智能指针、Lambda表达式、自动类型推导等现代特性，提升了开发安全性和便捷性。  

## 应用场景‌：  
广泛应用于游戏引擎（如Unreal Engine）、高性能服务器后端、图形图像处理、金融交易系统、浏览器内核（如Chrome）、数据库系统以及各类桌面应用软件。

# 什么是C语言标准？

C语言标准是由国际标准化组织（ISO）和国际电工委员会（IEC）共同制定的一系列规范，旨在统一C语言的语法、语义及标准库，确保代码在不同编译器和硬件平台上的可移植性。  

参考网站：https://www.c-language.org

# 什么是C++语言标准？

C++语言标准是由ISO/IEC联合技术委员会下属的C++标准化工作组（WG21）制定的国际规范，用于统一C++的语法、语义和标准库定义，保障代码跨平台可移植性。  

参考网站：https://isocpp.org/std/the-standard

在考点机器环境和官方评测系统中，C++代码通常使用 -std=c++14 进行编译。这是目前GESP C++级别考试（如一级至八级）的主要标准支持版本。

# 什么是C++的标准模板库（Standard Template Library）？

Standard Template Library‌（简称 ‌STL‌），中文译为‌标准模板库‌，是 C++ 标准库的核心组成部分，也是 C++ 语言中最具影响力的创新之一。它由 Alexander Stepanov 等人开发，最终被纳入 ISO C++ 标准，旨在提供一套高效、通用且可复用的数据结构和算法框架。

## 1. 常用容器头文件  
<vector>‌：动态数组，支持随机访问，最常用的序列容器。  
<string>‌：字符串类，提供便捷的文本处理功能。  
<list>‌：双向链表，支持高效的插入和删除操作。  
<deque>‌：双端队列，支持两端高效插入删除。  
<stack>‌：栈适配器，遵循后进先出（LIFO）原则。  
<queue>‌：队列适配器，遵循先进先出（FIFO）原则。  
<priority_queue>‌：优先队列，元素按优先级排序。  
<map> / <set>‌：基于红黑树的关联容器，自动排序且键值唯一。  
<unordered_map> / <unordered_set>‌：基于哈希表的关联容器，查找效率更高。  

## 2. 算法与函数对象头文件  
‌<algorithm>‌：核心算法库，包含 sort、find、max、reverse 等通用算法。  
‌<functional>‌：函数对象库，提供 greater、less、bind 等仿函数和绑定器。  
‌<numeric>‌：数值算法库，包含 accumulate（求和）、iota（递增填充）等。  

## 3. 迭代器与工具头文件  
‌<iterator>‌：迭代器工具，提供 back_inserter、istream_iterator 等特殊迭代器。  
‌<utility>‌：实用工具库，包含 pair、swap、move 等基础功能。  
‌<memory>‌：内存管理库，包含智能指针 shared_ptr、unique_ptr 等。  

# 什么是C++的标准库命名空间‌（std）？  

在 C++ 编程语言中，‌std‌ 是英文单词 ‌standard‌（标准）的缩写。

它代表 ‌C++ 标准库命名空间‌。C++ 标准库中的所有类、函数和对象（例如 cout、cin、string、vector 等）都被定义在这个名为 std 的命名空间中。使用命名空间的主要目的是为了防止不同库或用户代码之间出现名称冲突，确保标识符的唯一性和代码的组织性。
