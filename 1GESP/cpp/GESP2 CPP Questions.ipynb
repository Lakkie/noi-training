```c
//B4497 [GESP202603 二级] 数数
#include<iostream>
using namespace std;
bool isBeautiful(int n){
    int count=0;
    while(n>0){
        if(n%10==2){
            count++;
        }
        if(count>3){
            return false;
        }
        n/=10;
    }
    return count==3;
}
int main(){
    //输入
    ios::sync_with_stdio(false);
    cin.tie(NULL);
    int L,R;
    if(cin>>L>>R){
    	//计算
        int ans=0;
        for(int i=L;i<=R;++i){
            if(isBeautiful(i)){
                ans++;
            }
        }
        //输出
        cout<<ans<<endl;
    }
    return 0;
}
```
