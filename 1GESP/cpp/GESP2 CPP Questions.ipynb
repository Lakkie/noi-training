```c
//B4497 [GESP202603 二级] 数数
#include<iostream>
using namespace std;
bool isBeautiful(int n){
    int count=0;
    while(n>0){
        if(n%10==2){
            count++;
        }
        if(count>3){
            return false;
        }
        n/=10;
    }
    return count==3;
}
int main(){
    //输入
    ios::sync_with_stdio(false);
    cin.tie(NULL);
    int L,R;
    if(cin>>L>>R){
    	//计算
        int ans=0;
        for(int i=L;i<=R;++i){
            if(isBeautiful(i)){
                ans++;
            }
        }
        //输出
        cout<<ans<<endl;
    }
    return 0;
}
```

```c
//B4260 [GESP202503 二级] 时间跨越
#include<iostream>
using namespace std;
bool is_leap(int y){//判断是否为闰年
    if(y%4==0&&y%100!=0){
        return true;
    }else if(y%400==0){
        return true;
    }else{
        return false;
    }
}
int main(){
    //输入
    int y,m,d,h,k;
    cin>>y>>m>>d>>h>>k;
    //计算
    int yp=y,mp=m,dp=d,hp=h;
    int days[]={0,31,28,31,30,31,30,31,31,30,31,30,31};//平年各月天数数组，索引0仅用于占位
    hp=(h+k)%24;//1.处理小时进位
    dp=d+(h+k)/24;//增加天数，后续继续处理
    while(true){//循环处理日期、月份、年份进位
        int max_day;//2.处理天数进位
        if(mp==2){
            max_day=is_leap(yp)?29:days[mp];
        }else{
            max_day=days[mp];
        }
        if(dp<=max_day){
            break;//天数符合要求后，整个循环结束。
        }else{
            dp-=max_day;//3.处理月份进位
            mp++;
            if(mp>12){//4.处理年份进位
                mp=1;
                yp++;
            }
        }
    }
    //输出
    cout<<yp<<" "<<mp<<" "<<dp<<" "<<hp<<endl;
    return 0;
}
```

```c
//B4260 [GESP202503 二级] 时间跨越 优化
#include<iostream>
using namespace std;
bool is_leap(int y){//判断是否为闰年
    return (y%4==0&&y%100!=0)||(y%400==0);
}
int main(){
    //输入
    int y,m,d,h,k;
    cin>>y>>m>>d>>h>>k;
    //计算
    int days[]={31,28,31,30,31,30,31,31,30,31,30,31};//平年各月天数数组，月份=索引+1
    h+=k;//1.处理小时进位
    int add_day=h/24;//新增加的天数
    h%=24;//新的小时数
    d+=add_day;
    while(true){//循环处理日期、月份、年份进位
        int max_day;//2.处理天数进位
        if(m==2){//二月份
            max_day=is_leap(y)?29:days[m-1];
        }else{
            max_day=days[m-1];
        }
        if(d<=max_day) break;
        d-=max_day;//3.处理月份进位
        m++;
        if(m>12){//4.处理年份进位
            m=1;
            y++;
        }
    }
    //输出
    cout<<y<<" "<<m<<" "<<d<<" "<<h<<endl;
    return 0;
}
```
