```c
//B4497 [GESP202603 二级] 数数
#include<iostream>
using namespace std;
bool isBeautiful(int n){
    int count=0;
    while(n>0){
        if(n%10==2){
            count++;
        }
        if(count>3){
            return false;
        }
        n/=10;
    }
    return count==3;
}
int main(){
    //输入
    int L,R;
    if(cin>>L>>R){
    	//计算
        int ans=0;
        for(int i=L;i<=R;++i){
            if(isBeautiful(i)){
                ans++;
            }
        }
        //输出
        cout<<ans<<endl;
    }
    return 0;
}
```

```c
//B4553 [GESP202606 二级] 完全平方数计数 暴力枚举
#include<iostream>
#include<cmath>
using namespace std;
bool isPerfectSquare(int x){
    if(x<0){
        return false;
    }
    int s=sqrt(x);
    return s*s==x;
}
int main(){
    int l,r;
    cin>>l>>r;
    int ans=0;
    for(int i=l;i<=r;++i){
        if(isPerfectSquare(i)){
            ans++;
        }
    }
    cout<<ans<<endl;
    return 0;
}
```

```c
//B4554 [GESP202606 二级] 菱形 曼哈顿距离‌ Manhattan Distance
#include<iostream>
#include<cmath>
using namespace std;
int main(){
    int n;
    cin>>n;
    int center=n-1;//中心
    int radius=n-1;//半径
    for(int i=0;i<2*n-1;++i){//遍历每一行
        for(int j=0;j<2*n-1;++j){//遍历每一列
            int dist=abs(i-center)+abs(j-center);//计算当前点 (i, j) 到中心 (center, center) 的距离
            if(dist==radius){//如果距离等于半径，说明在菱形边界上
                cout<<"+";
            }else{
                cout<<".";
            }
        }
        cout<<endl;
    }
    return 0;
}
```

```c
//B4260 [GESP202503 二级] 时间跨越
#include<iostream>
using namespace std;
bool is_leap(int y){//判断是否为闰年
    if(y%4==0&&y%100!=0){
        return true;
    }else if(y%400==0){
        return true;
    }else{
        return false;
    }
}
int main(){
    //输入
    int y,m,d,h,k;
    cin>>y>>m>>d>>h>>k;
    //计算
    int yp=y,mp=m,dp=d,hp=h;
    int days[]={0,31,28,31,30,31,30,31,31,30,31,30,31};//平年各月天数数组，索引0仅用于占位
    hp=(h+k)%24;//1.处理小时进位
    dp=d+(h+k)/24;//增加天数，后续继续处理
    while(true){//循环处理日期、月份、年份进位
        int max_day;//2.处理天数进位
        if(mp==2){
            max_day=is_leap(yp)?29:days[mp];
        }else{
            max_day=days[mp];
        }
        if(dp<=max_day){
            break;//天数符合要求后，整个循环结束。
        }else{
            dp-=max_day;//3.处理月份进位
            mp++;
            if(mp>12){//4.处理年份进位
                mp=1;
                yp++;
            }
        }
    }
    //输出
    cout<<yp<<" "<<mp<<" "<<dp<<" "<<hp<<endl;
    return 0;
}
```

```c
//B4260 [GESP202503 二级] 时间跨越 优化
#include<iostream>
using namespace std;
bool is_leap(int y){//判断是否为闰年
    return (y%4==0&&y%100!=0)||(y%400==0);
}
int main(){
    //输入
    int y,m,d,h,k;
    cin>>y>>m>>d>>h>>k;
    //计算
    int days[]={31,28,31,30,31,30,31,31,30,31,30,31};//平年各月天数数组，月份=索引+1
    h+=k;//1.处理小时进位
    int add_day=h/24;//新增加的天数
    h%=24;//新的小时数
    d+=add_day;
    while(true){//循环处理日期、月份、年份进位
        int max_day;//2.处理天数进位
        if(m==2){//二月份
            max_day=is_leap(y)?29:days[m-1];
        }else{
            max_day=days[m-1];
        }
        if(d<=max_day) break;
        d-=max_day;//3.处理月份进位
        m++;
        if(m>12){//4.处理年份进位
            m=1;
            y++;
        }
    }
    //输出
    cout<<y<<" "<<m<<" "<<d<<" "<<h<<endl;
    return 0;
}
```

```c
//B4357 [GESP202506 二级] 幂和数
#include<iostream>
using namespace std;
int countSetBits(int n){//计算整数n的二进制表示中 1 的个数
    int count=0;
    while(n>0){
        if(n&1){//检查最低位是否为 1
            count++;
        }
        n>>=1;//右移一位，相当于n=n/2
    }
    return count;
}
int main(){
	//输入
    int l,r;
    cin>>l>>r;
    //计算
    int count=0;
    for(int i=l;i<=r;++i){
        if(i>=2&&countSetBits(i)<=2){
            count++;
        }
    }
    //输出
    cout<<count<<endl;
    return 0;
}
```

```c
//B4357 [GESP202506 二级] 幂和数
#include<iostream>
#include<vector>
#include<algorithm>
using namespace std;
int main(){
    //输入
    int l,r;
    cin>>l>>r;
    //计算
    vector<int> sums;
    for(int x=0;x<=14;++x){//2^13=8192<10000，2^14=16384>10000
        for(int y=0;y<=14;++y){
            int val=(1<<x)+(1<<y);
            if(val>=l&&val<=r){
                sums.push_back((int)val);
            }
        }
    }
    sort(sums.begin(), sums.end());//排序
    sums.erase(unique(sums.begin(),sums.end()),sums.end());//去重
    //输出
    cout<<sums.size()<<endl;
    return 0;
}
```

```c
//B4411 [GESP202509 二级] 优美的数字
#include<iostream>
using namespace std;
int main(){
    //输入
    int n;
    cin>>n;
    //计算
    int c=0;
    for(int i=0;i<=n;++i){
    	int x=i;
        int s=x%10;        //参考数字
        while(x>0){        //从个位开始逐个位置的数字
            if(x%10==s){   //与参考数字相同
                if(x<10){  //比10小代表已经取到最后一位数字
                    c++;   //确认当前的i符合要求，累加一次
                    break; //结束当前循环
                }else{     //比10大，则需要去掉最后一位
                    x/=10; //去掉个位
                }
            }else{         //与参考数字不同
                break;     //直接跳过当前循环
            }
        }
    }
    //输出
    cout<<c<<endl;
    return 0;
}
```

```c
//B4411 [GESP202509 二级] 优美的数字 数学优化
#include<iostream>
#include<string>
#include<cmath>
using namespace std;
int main(){
    //输入
    int n;
    cin>>n;
    //计算
    string s=to_string(n);
    int len=s.length();
    int ans=0;
    ans+=(len-1)*9;
    int base=0;
    for(int i=0;i<len;++i){
        base=base*10+1;
    }
    for(int d=1;d<=9;++d){
        int val=base*d;
        if(val<=n){
            ans++;
        }else{
            break;
        }
    }
    //输出
    cout<<ans<<endl;
    return 0;
}
```

```c
//B4448 [GESP202512 二级] 黄金格
#include<iostream>
using namespace std;
int main(){
    //输入
    int h,w,x;
    cin>>h>>w>>x;
    //计算
    int cnt=0;//用于统计黄金格的数量
    for(int r=1;r<=h;++r){//遍历所有格子
        for(int c=1;c<=w;++c){
            int val=x+r-c;//计算不等式右边的值
            if(val<0){//如果右边小于0，因为左边 sqrt(r^2+c^2) >= 0，所以不等式一定不成立
                continue;
            }
            int left_side=r*r+c*c;//左边平方
            int right_side=val*val;//右边平方
            if(left_side<=right_side){
                cnt++;
            }
        }
    }
    //输出
    cout<<cnt<<endl;
    return 0;
}
```
