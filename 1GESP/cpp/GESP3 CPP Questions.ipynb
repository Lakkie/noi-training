```c
//B4499 [GESP202603 三级] 二进制回文串
#include<iostream>
#include<string>
#include<algorithm>
using namespace std;
//判断一个数是否是二进制回文数
bool isBinaryPalindrome(int x){
    string s="";
    int temp=x;
    while(temp>0){
        s+=(temp%2)+'0';//逆序的二进制字符串（低位在前）
        temp/=2;
    }
    string t=s;
    reverse(t.begin(), t.end());//正序的二进制字符串
    return s==t;
}
int main(){
    //输入
    ios::sync_with_stdio(false);
    cin.tie(NULL);
    int n;
    if(!(cin>>n)){
        return 0;
    }
    //计算
    int count=0;
    for(int i=1;i<=n;++i){
        if(isBinaryPalindrome(i)){
            count++;
        }
    }
    //输出
    cout<<count<<endl;
    return 0;
}
```

```c
//B4262 [GESP202503 三级] 词频统计
#include<iostream>
#include<string>
#include<map>
#include<algorithm>
#include<cctype>
using namespace std;
int main(){
    //输入
    ios::sync_with_stdio(false);
    cin.tie(NULL);
    int n;
    if(!(cin>>n)){
        return 0;
    }
    //计算
    map<string, int> word_count;//使用 map 存储单词及其出现次数 key: 小写单词, value: 出现次数
    string word;
    for(int i=0;i<n;++i){
        cin>>word;
        transform(word.begin(), word.end(), word.begin(), ::tolower);//将单词转换为小写
        word_count[word]++;//统计次数
    }
    string max_word="";
    int max_count=0;
    for(const auto& pair:word_count){//遍历 map 寻找出现次数最多的单词
        if(pair.second>max_count){
            max_count=pair.second;
            max_word=pair.first;
        }
    }
    //输出
    cout<<max_word<<endl;
    return 0;
}
```

```c
//B4359 [GESP202506 三级] 分糖果
#include<iostream>
#include<vector>
#include<algorithm>
using namespace std;
int main(){
    //输入
    ios::sync_with_stdio(false);
    cin.tie(NULL);
    int n;
    if(!(cin>>n)){
        return 0;
    }
    vector<int> a(n);
    for(int i=0;i<n;++i){
        cin>>a[i];
    }
    //计算
    long long total_candies=0;
    int prev_candy=0;//前一个小朋友分到的糖果数
    for(int i=0;i<n;++i){
        int min_from_prev=prev_candy+1;//当前小朋友至少要比前一个多1颗
        int current_candy=max(a[i],min_from_prev);//同时不能低于自己的最低需求 a[i]
        total_candies+=current_candy;
        prev_candy = current_candy;//更新前一个小朋友的糖果数，供下一个小朋友参考
    }
    //输出
    cout<<total_candies<<endl;
    return 0;
}
```

```c
//B4358 [GESP202506 三级] 奇偶校验
#include<iostream>
#include<string>
using namespace std;
int main(){
    //输入&计算
    ios::sync_with_stdio(false);
    cin.tie(NULL);
    int n;
    if(!(cin>>n)){
        return 0;
    }
    int count=0;
    for(int i=0;i<n;++i){
        int c;
        cin>>c;
        while(c){
            if(c%2==1){
                count++;
            }
            c/=2;
        }
    }
    //输出
    cout<<count<<" "<<(count%2)<<endl;
    return 0;
}
```

```c
//B4414 [GESP202509 三级] 日历制作
#include<iostream>
#include<iomanip>
#include<string>
using namespace std;
//获取2025年某月的天数
int get_days_in_month(int m){
    if(m == 2){
        return 28;//2025是平年
    }else if(m==4||m==6||m==9||m==11){
        return 30;
    }else if(m==1||m==3||m==5||m==7||m==8||m==10||m==12){
        return 31;
    }else{
        return 0;
    }
}
//计算2025年第m月1号是星期几
int get_first_day_weekday(int m){
    int current_weekday=2;//已知 2025-01-01 是星期三 Wednesday
    for(int i=1;i<m;++i){
        int days=get_days_in_month(i);
        current_weekday=(current_weekday+days)%7;//Mon=0, Tue=1, Wed=2, Thu=3, Fri=4, Sat=5, Sun=6
    }
    return current_weekday;
}
int main(){
    //输入
    ios::sync_with_stdio(false);
    cin.tie(NULL);
    int m;
    if(!(cin>>m)){
        return 0;
    }
    //计算&输出
    cout<<"MON TUE WED THU FRI SAT SUN"<<endl;
    int start_idx=get_first_day_weekday(m);
    int total_days=get_days_in_month(m);
    for(int i=0;i<start_idx;++i){//输出前置空格
        cout<<"    ";
    }
    for(int day=1;day<=total_days;++day){
        int current_idx=(start_idx+day-1)%7;
        cout<<setw(3)<<day;
        if(current_idx!=6){
            cout<<" ";
        }else{
            cout<<endl;
        }
    }
    if((start_idx+total_days-1)%7!=6){
        cout<<endl;
    }
    return 0;
}
```

```c
//B4413 [GESP202509 三级] 数组清零
#include<iostream>
#include<vector>
using namespace std;
int main(){
    //输入
    ios::sync_with_stdio(false);
    cin.tie(NULL);
    int n;
    if(!(cin>>n)){
        return 0;
    }
    vector<int> a(n);
    for(int i=0;i<n;++i){
        cin>>a[i];
    }
    //计算
    int c=0;
    while(true){
        bool all_zero=true;
        int k=-1,max_val=-1,min_val=100;
        for(int i=0;i<n;++i){
            if(a[i]!=0){
                all_zero=false;
            }
            if(a[i]>max_val){//在数组中找到最大的整数，记其下标k
                max_val=a[i];
                k=i;
            }else if((a[i]==max_val)&&(i>k)){//如果有多个最大值，选择下标最大的
                k=i;
            }
            if((a[i]>0)&&(a[i]<min_val)){//从数组所有不为零的整数中找到最小的整数
                min_val=a[i];
            }
        }
        if(all_zero){
            break;
        }
        a[k]-=min_val;
        c++;
    }
    //输出
    cout<<c<<endl;
    return 0;
}
```

```c
//B4449 [GESP202512 三级] 密码强度
#include<iostream>
#include<string>
#include<cctype>
using namespace std;
int main(){
    //输入
    ios::sync_with_stdio(false);
    cin.tie(NULL);
    int t;
    if(!(cin>>t)){
        return 0;
    }
    for(int i=0;i<t;++i){
        string s;
        if(!(cin>>s)){
            return 0;
        }
        //计算&输出
        if(s.length()<8){
            cout<<"N"<<endl;
            continue;
        }
        bool has_upper=false,has_digit=false;
        for(char c:s){
            if(isupper(c)){
                has_upper=true;
            }else if(isdigit(c)){
                has_digit=true;
            }
            //如果两个条件都满足了，可以提前退出循环，提高效率
            if(has_upper&&has_digit){
                break;
            }
        }
        if(has_upper&&has_digit){
            cout<<"Y"<<endl;
        }else{
            cout<<"N"<<endl;
        }
    }
    return 0;
}
```

```c
//B4450 [GESP202512 三级] 小杨的智慧购物
#include<iostream>
#include<vector>
#include<algorithm>
using namespace std;
int main() {
    //输入
    ios::sync_with_stdio(false);
    cin.tie(NULL);
    int m,n;
    if(!(cin>>m>>n)){
        return 0;
    }
    vector<int> min_price(m+1,1001);//确保初始值>任何实际价格
    for(int i=0;i<n;++i){
        int k,p;
        cin>>k>>p;
        //计算
        if(p<min_price[k]){
            min_price[k]=p;
        }
    }
    int total_cost=0;
    for (int i=1;i<=m;++i){
        total_cost+=min_price[i];
    }
    //输出
    cout<<total_cost<<endl;
    return 0;
}

```
