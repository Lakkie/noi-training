```python
side=11
center=side//2+1
symbol_background="."
symbol_grash="+"
for i in range(0,center):
    a=i+1
    leading=symbol_background*(center-i-1)
    if(a==1):
        line=leading+symbol_grash+leading
    elif(a==center):
        line=leading+symbol_grash*(2*a-1)
    else:
        mid_bg=symbol_background*(2*a-3)
        line=leading+symbol_grash+mid_bg+symbol_grash+leading
    print(line)
```

```python
row=10                                      #row代表行数
column=10                                   #column代表列数
symbol_background="*"                       #背景的符号
for i in range(row):                        #外层循环处理行
    row_list=[]                             #存放每行要打印的符号
    for j in range(column):                 #内层循环处理列
        row_list.append(symbol_background)  #将符号存入列表
    row_str="".join(row_list)               #将列表转换为字符串
    print(row_str)                          #打印当前行
```

```python
row=10
column=10
symbol_background="*"
symbol_graph="x"
for i in range(row):
    row_list=[]
    for j in range(column):
        if(i==0 or i==row-1 or j==0 or j==column-1):
            row_list.append(symbol_graph)
        else:
            row_list.append(symbol_background)
    row_str="".join(row_list)
    print(row_str)
```

```python
row=10
column=10
symbol_background="*"
symbol_graph1="+"
symbol_graph2="-"
symbol_graph3="|"
for i in range(row):
    row_list=[]
    for j in range(column):
        if((i==0 and j==0)or(i==0 and j==column-1)or(i==column-1 and j==0)or(i==column-1 and j==column-1)):
            row_list.append(symbol_graph1)
        elif((i==0 and j!=0 and j!=column-1)or(i==column-1 and j!=0 and j!=column-1)):
            row_list.append(symbol_graph2)
        elif((j==0 and i!=0 and i!=column-1)or(j==column-1 and i!=0 and i!=column-1)):
            row_list.append(symbol_graph3)
        else:
            row_list.append(symbol_background)
    row_str="".join(row_list)
    print(row_str)
```
