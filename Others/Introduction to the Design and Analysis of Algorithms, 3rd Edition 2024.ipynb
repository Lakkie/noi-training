算法设计与分析基础（第3版详解版）  
ISBN：9787302625445

作者基于丰富的教学经验，开发了一套全新的算法分类方法。  
全书共12章，涵盖算法效率分析、数据结构及常用设计技术（蛮力法、分治法、动态规划等），通过排序、查找、图论等案例结合时间复杂度分析解析算法原理。
附录提供公式推导与递推关系指南，强化实践应用。本书以问题求解策略为主线，整合理论讲解与习题训练，适配计算机专业教学及算法研究者参考。
