第10章　知识的视觉表征  
Chapter 10: The Visual Representation of Knowledge  

知识的视觉表征是知识可视化的核心实践载体，指借助媒介工具将知识的意义封装于视觉文本之中，通过图形、图像等可感知的视觉形式，实现复杂知识的传递、重构与创新，是知识可视化落地的关键实践途径。  
常见的知识视觉表征工具包括概念图、语义图、因果图、思维导图、知识动画、视觉隐喻等，不同形态适配不同类型知识的呈现需求。
