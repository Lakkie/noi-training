# 1 Logical Structure 逻辑结构

## 1.1 Set Structure 集合结构‌

定义‌：数据元素之间除了“同属一个集合”的关联外，不存在其他任何特定逻辑关系，元素组织形式松散。

## 1.2 Linear Structure 线性结构‌

定义‌：数据元素之间存在‌一对一‌的线性相邻关系，有且仅有一个开始结点和终端结点，所有结点最多仅有一个直接前驱和一个直接后继。

### 1.2.1 General Linear List 一般线性表‌

定义‌：一般线性表是具有相同数据类型的n（n≥0）个数据元素的有限序列，除首元素无直接前驱、尾元素无直接后继外，其余每个元素有且仅有一个直接前驱和一个直接后继，且支持在表中‌任意位置‌执行插入、删除、查找、遍历等操作，没有操作位置的特殊约束。

#### 1.2.1.1 Sequential List 顺序表

定义:顺序表是采用‌顺序存储结构‌实现的线性表，它将线性表中所有元素依次存放在一组地址连续的内存单元中，借助物理上的邻接关系直接体现数据元素之间的一对一逻辑相邻关系。

#### 1.2.1.2 Singly Linked List 单向链表

定义：单向链表是用‌任意地址的内存单元‌存储线性表元素的链式结构，每个结点由数据域和唯一的指针域组成，数据域存放元素值，指针域指向序列中的下一个结点，所有结点通过指针串联成一条单向的链式序列。

#### 1.2.1.3 Doubly Linked List 双向链表

定义：双向链表的每个结点都包含‌数据域、前驱指针域、后继指针域‌三个部分，数据域存放元素值，前驱指针指向该结点的前一个结点，后继指针指向该结点的后一个结点，所有结点通过双向指针串联成可双向遍历的链式序列。

#### 1.2.1.4 Circular Linked List 循环链表

定义：循环链表是将普通链表的首尾结点通过指针相连形成的环形链式结构，最常见的实现是单向循环链表，它把最后一个结点的后继指针指向链表的头结点，让整个链表形成一个首尾相接的闭合环形。

### 1.2.2 Restricted Linear List 受限线性表‌

‌定义：受限线性表是对基础线性表的操作位置做了明确约束的线性结构，它保留了线性表“元素为一对一的有限序列”的基础逻辑特征，但仅允许在‌指定的一个或两个端点‌完成元素的插入、删除操作，不支持在表中任意位置执行增删操作。

#### 1.2.2.1 Stack 栈

定义：栈是限定仅能在表的同一端进行插入和删除操作的受限线性表，允许操作的一端被称为‌栈顶‌，另一端被称为‌栈底‌，元素的访问遵循‌后进先出（LIFO, Last In First Out）‌的核心规则。

#### 1.2.2.2 Queue 队列

定义：队列是限定仅能在表的不同两端分别进行插入和删除操作的受限线性表，允许插入元素的一端被称为‌队尾‌，允许删除元素的一端被称为‌队头‌，元素的访问遵循‌先进先出（FIFO, First In First Out）‌的核心规则。

#### 1.2.2.3 Deque (Double-Ended Queue) 双端队列

定义：双端队列（Deque，全称为Double-Ended Queue）是允许在队列的‌两端‌都执行插入和删除操作的受限线性表，它没有普通队列“仅队尾插入、仅队头删除”的严格约束，两端都可作为队头或队尾完成元素的增删操作。

#### 1.2.2.4 String 串

定义：串在计算机数据结构中通常指字符串，是由零个或多个字符组成的有限序列，是一种特殊的线性表，它的数据元素被限定为字符集中的字符。

### 1.2.3 Extended Linear List 扩展线性表‌（推广线性表）

定义：扩展线性表是对基础线性表的元素组成规则进行扩展的线性结构，它保留了线性表的部分基础逻辑特征，同时突破了“元素必须为原子类型”的约束，允许元素为子表或多维结构，以此实现更复杂的数据关系表达。

#### 1.2.3.1 Array 数组

定义：数组是用一组‌连续的内存空间‌，存储一组‌相同数据类型元素‌的线性结构，每个元素都有唯一的整数索引（通常从0开始计数），可通过索引直接定位元素，实现O(1)时间复杂度的随机访问。

#### 1.2.3.2 Generalized List 广义表

定义：广义表是n（n≥0）个元素的有限序列，其中每个元素既可以是不可拆分的原子数据元素，也可以是另一个广义表（子表），它突破了基础线性表元素必须为原子类型的约束，是线性表的深度扩展形式。

## 1.3 Tree Structure 树结构‌

定义‌：数据元素之间存在‌一对多‌的层次关系，有且仅有一个根结点，其余结点有且仅有一个直接前驱，可拥有多个直接后继。

## 1.4 Graph Structure 图结构‌

定义‌：数据元素之间存在‌多对多‌的任意关联关系，任意两个结点之间都可能存在连接路径。

# 2 Storage Structure / Physical Structure 存储结构 / 物理结构

## 2.1 Sequential Storage Structure 顺序存储结构‌

定义‌：将逻辑上相邻的元素，存放在物理地址也连续的存储单元中，元素间的逻辑关系通过存储单元的邻接关系来体现，典型实现为数组。

核心特点‌：存储密度大、支持随机存取，但插入和删除元素时需要移动大量元素，操作效率较低。

## 2.2 Linked Storage Structure 链式存储结构‌

‌定义‌：不要求数据元素在物理地址上连续，每个元素结点包含数据域和指针域，通过指针记录下一个结点的地址，以此串联起所有元素，典型实现为链表。

‌核心特点‌：插入和删除操作仅需修改指针，无需移动元素，但不支持随机存取，查找元素需要沿指针依次遍历。

## 2.3 Indexed Storage Structure 索引存储结构‌

定义‌：在存储数据元素的同时，额外建立一张索引表，索引表的每一项都包含关键字和对应元素的存储地址，通过索引可以快速定位目标元素。

‌核心特点‌：元素查找效率高，但需要额外的空间存储索引表，会增加一定的空间开销。

## 2.4 Hash Storage Structure 哈希存储结构（散列存储结构）

‌定义‌：不依赖元素的物理位置或指针关联，直接根据元素的关键字，通过预先设定的散列函数计算出该元素的存储地址，将元素存入对应位置。

‌核心特点‌：元素的查找、插入和删除操作速度极快，但如果散列函数设计不合理，容易出现地址冲突，需要额外机制处理冲突。
