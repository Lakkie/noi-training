```c
//B2025 输出字符菱形
#include<iostream>
using namespace std;
int main(){
	char space=' ';
	char star='*';
	int n=5;
	int mid=n/2;
    for(int i=0;i<=mid;++i){ //上半部分
        int spaces=mid-i;
        int stars=2*i+1;
        cout<<string(spaces,space)<<string(stars,star)<<"\n";
    }
    for(int i=mid-1;i>=0;--i){ //下半部分
        int spaces=mid-i;
        int stars=2*i+1;
        cout<<string(spaces,space)<<string(stars,star)<<"\n";
    }
	return 0;
}
```

```c
//P5705 【深基2.例7】数字反转
#include<iostream>
using namespace std;
int main(){
    double a;
    cin>>a;
    int b=(int)(a*10);       //将小数转换为整数
    int digit1=b%10;         //原小数位，现整数最高位
    int digit2=(b/10)%10;    //原个位
    int digit3=(b/100)%10;   //原十位
    int digit4=b/1000;       //原百位
    cout<<digit1<<"."<<digit2<<digit3<<digit4<<endl;
    return 0;
}
```

```c
//P5708 【深基2.习2】三角形面积
#include<iostream>
#include<iomanip>
#include<cmath>
using namespace std;
int main(){
    double a,b,c,p,s;
    cin>>a>>b>>c;
    p=(a+b+c)/2;
    s=sqrt(p*(p-a)*(p-b)*(p-c));
    cout<<fixed<<setprecision(1)<<s<<endl;
    return 0;
}
```
