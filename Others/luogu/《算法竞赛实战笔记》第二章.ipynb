#include<iostream>
#include<string>
using namespace std;
string add(string a,string b){
    string c;//计算结果(反序)
    int lenA=a.size();
    int lenB=b.size();
    int maxLen=0;//取较长的位数
    if(lenA>=lenB){
        maxLen=lenA;
    }else{
        maxLen=lenB;
    }
    int carry=0;//进位
    for(int i=0;i<maxLen;++i){//按较长的位数逐位计算
        int sum=carry;
        if(i<lenA) sum+=a[lenA-1-i]-'0';//获取a的当前位（从后往前取）
        if(i<lenB) sum+=b[lenB-1-i]-'0';//获取b的当前位（从后往前取）
        c+=sum%10+'0';//当前位
        carry=sum/10;//更新进位
    }
    if(carry>0){//最高一位进位
        c+=carry+'0';
    }
    string r;//计算结果(正序)
    for(int i=c.size()-1;i>=0;--i){
        r+=c[i];
    }
    return r;
}
int main(){
    //输入
    string a,b;
    cin>>a>>b;
    //计算
    string s=add(a,b);
    //输出
    cout<<s<<endl;
    return 0;
}

    int carry=0;//进位
    int lenA=A.len;
    int lenB=B.len;
    int maxLen=max(lenA,lenB);//遍历较长的位数
    for(int i=0;i<maxLen;++i){
        int sum=carry;
        if(i<lenA) sum+=A.d[i];//获取a的当前位（从后往前取）
        if(i<lenB) sum+=B.d[i];//获取b的当前位（从后往前取）

		C.d[C.len-1]=carry;
	}

string add(string a,string b){
    string result="";
    int i=a.size()-1;//指向a的末尾
    int j=b.size()-1;//指向b的末尾
    int carry=0;//进位
    //当还有数字未处理，或还有进位时继续循环
    while(i>=0||j>=0||carry>0){
        int sum=carry;
        //获取a的当前位，如果索引越界则视为0
        if(i>=0){
            sum+=a[i]-'0';
            i--;
        }
        //获取b的当前位，如果索引越界则视为0
        if(j>=0){
            sum+=b[j]-'0';
            j--;
        }
        //计算新的进位和当前位的结果
        carry=sum/10;
        int digit=sum%10;
        //将当前位字符追加到结果字符串
        result.push_back(digit+'0');
    }
    //因为是从低位到高位追加的，所以结果需要反转
    reverse(result.begin(),result.end());
    return result;
}

//比较两个字符串数字值的大小
bool compare(string a,string b){
    if(a.size()!=b.size()){
        return a.size()>b.size();
    }
    return a>=b;//如果长度相同，直接字典序比较
}
//计算较大数字减较小数字
string subtract_abs(string larger,string smaller){
    string c="";
    int borrow=0;
    for(int i=0;i<larger.size();++i){
        int digitA=(larger[larger.size()-1-i]-'0');
        int digitB=(i<smaller.size())?(smaller[smaller.size()-1-i]-'0'):0;
        int diff=digitA-digitB-borrow;
        if(diff<0){
            diff+=10;
            borrow=1;
        }else{
            borrow=0;
        }
        c+=diff+'0';
    }
	int f=0;//前导零的个数
	for(int i=c.size()-1;i>=0;i--){
		if(c[i]=='0'){
			f++;
		}else{
			break;//遇到第一个不是0的字符就结束循环。
		}
	}
	if(f<c.size()){
	    return c.substr(0,c.size()-f);	
	}else{
		return "0";//特殊情况：全零处理
	}
}
string subtract(string a,string b){
    bool is_negative=false;
    string larger,smaller,c,result;
    if(compare(a,b)){
        larger=a;
        smaller=b;
        is_negative=false;
    }else{
        larger=b;
        smaller=a;
        is_negative=true;
    } 
    c=subtract_abs(larger,smaller);
    if(is_negative){
        c+='-';
    }
    for(int i=c.size()-1;i>=0;--i){
        result+=c[i];
    }
    return result;
}

    //输入
    string a,b;
    cin>>a>>b;
    //计算
    string s=subtract(a,b);

    int d[MAXN];//存储每一位数字，d[0]是个位

    BigInt(){
        memset(d,0,sizeof(d));

    }
    void init(string s){//从字符串初始化
        memset(d,0,sizeof(d));
        len=s.size();

            d[i]=s[len-1-i]-'0';

            cout<<d[i];
        }

BigInt subtract(const BigInt& A,const BigInt& B){
    BigInt C;
    int borrow=0;
    for(int i=0;i<A.len;++i){
        int temp=A.d[i]-borrow;
        if(i<B.len) temp-=B.d[i];
        if(temp<0){
            temp+=10;
            borrow=1;
        }else{
            borrow=0;
        }
        C.d[C.len++]=temp;
    }
    //去除前导零
    while(C.len>1&&C.d[C.len-1]==0){
        C.len--;
    }    
    return C;
}
//辅助函数：比较两个高精度数的大小 (A >= B 返回 true)

    if(A.len!=B.len) return A.len>B.len;
    for(int i=A.len-1;i>=0;--i){
        if(A.d[i]!=B.d[i]) return A.d[i]>B.d[i];
    }
    return true;//相等也算A>=B
}

#include<iostream>
#include<string>
#include<vector>
using namespace std;
string multiply(string a,string b){
	if(a=="0"||b=="0"){//特殊情况
		return "0";
	}
    int na=a.size();
    int nb=b.size();
    vector<int> c(na+nb,0);//临时计算结果(逆序)
    for(int i=na-1;i>=0;--i){
        for(int j=nb-1;j>=0;--j){
            int x=a[i]-'0';
            int y=b[j]-'0';
            int ia=na-1-i;
            int jb=nb-1-j;
            c[ia+jb]+=x*y;
        }
    }
    for(int i=0;i<(na+nb-1);++i){//统一处理进位
        if(c[i]>=10){
            c[i+1]+=c[i]/10;
            c[i]%=10;
        }
    }
    int len=na+nb;
    while(len>1&&c[len-1]==0){//确定结果的实际长度并去除前导零
        len--;
    }
    string result;
    for(int i=len-1;i>=0;--i){//计算结果(正序)
        result.push_back(char(c[i]+'0'));
    }
    return result;
}
int main(){
    //输入
    string a,b;
    cin>>a>>b;
    //计算
    string s=multiply(a,b);

#include<iostream>
#include<vector>

#include<algorithm>
using namespace std;
//比较两个高精度数的大小(去掉前导零后)返回 1 if a>b, -1 if a<b, 0 if a==b
int compare_vector(const vector<int>& a,const vector<int>& b){
    if(a.size()!=b.size()){
        return a.size()>b.size()?1:-1;
    }
    for(int i=a.size()-1;i>=0;--i){
        if(a[i]!=b[i]){
            return a[i]>b[i]?1:-1;
        }
    }
    return 0;
}
//高精度减法:a=a-b (假设 a>=b)
void subtract_vector(vector<int>& a,const vector<int>& b){
    int borrow=0;
    for(size_t i=0;i<b.size();++i){
        a[i]-=b[i]+borrow;
        if(a[i]<0){
            a[i]+=10;
            borrow=1;
        }else{
            borrow=0;
        }
    }
    for(size_t i=b.size();borrow;++i){//处理剩余借位
        a[i]-=borrow;
        if(a[i]<0){
            a[i]+=10;
            borrow=1;
        }else{
            borrow=0;
        }
    }
    while(a.size()>1&&a.back()==0){//去除前导零
        a.pop_back();
    }
}
//高精度除法: a/b
vector<int> divide_vector(const vector<int>& a,const vector<int>& b){
    vector<int> quotient(a.size(),0);//商，初始化为0
    vector<int> remainder=a;//余数，初始化为被除数
    int shift=a.size()-b.size();//商的最高位位置
    if(shift<0){
        return {0};//a<b, 商为0
    }
    for(int i=shift;i>=0;--i){//从高位到低位确定商的每一位
        vector<int> current_divisor(b.begin(),b.end());//构造当前位的除数: b*10^i
        current_divisor.insert(current_divisor.begin(),i,0);//在b的低位补i个0
        int count=0;
        while(compare_vector(remainder,current_divisor)>=0){//试商：不断减去current_divisor，直到remainder<current_divisor
            subtract_vector(remainder,current_divisor);
            count++;
        }
        quotient[i]=count;
    }
    while(quotient.size()>1&&quotient.back()==0){//去除商的前导零
        quotient.pop_back();
    }
    return quotient;
}
string divide(const string& a,const string& b){
    vector<int> va,vb;//将字符串转换为向量，低位在前
    for(int i=a.size()-1;i>=0;--i){
        va.push_back(a[i]-'0');
    }
    for(int i=b.size()-1;i>=0;--i){
        vb.push_back(b[i]-'0');
    }
    vector<int> vr=divide_vector(va,vb);
    string result;
    for(int i=vr.size()-1;i>=0;--i){//输出结果(高位在后，所以逆序输出)
        result.push_back(vr[i]+'0');
    }
    return result;
}
int main(){
    //输入
    string a,b;
    cin>>a>>b;
    //计算
    string s=divide(a,b);
    //输出

}

### P5432 A/B Problem（高精度除法Ⅱ）

````python
import sys

# 增加递归深度
sys.setrecursionlimit(2000000)

def solve():
    # 读取输入
    input_data = sys.stdin.read().split()
    if not input_data:
        return
    
    a_str = input_data[0]
    b_str = input_data[1]
    
    # 1. 高精度除法
    a = int(a_str)
    b = int(b_str)
    r = a // b
    
    if r == 0:
        print(0)
        return
        
    # 2. 构造多项式 F(x)
    # 系数从低位到高位
    s_r = str(r)
    digits = [int(c) for c in s_r[::-1]]
    
    # 去除高位多余的零
    while len(digits) > 1 and digits[-1] == 0:
        digits.pop()
        
    n = len(digits)
    
    # 3. NTT 和多项式求逆
    MOD = 998244353
    G_ROOT = 3
    
    def ntt(a, invert):
        n_len = len(a)
        j = 0
        for i in range(1, n_len):
            bit = n_len >> 1
            while j & bit:
                j ^= bit
                bit >>= 1
            j ^= bit
            if i < j:
                a[i], a[j] = a[j], a[i]
        
        len_seg = 2
        while len_seg <= n_len:
            wlen = pow(G_ROOT, (MOD - 1) // len_seg, MOD)
            if invert:
                wlen = pow(wlen, MOD - 2, MOD)
            for i in range(0, n_len, len_seg):
                w = 1
                for j_idx in range(len_seg // 2):
                    u = a[i + j_idx]
                    v = a[i + j_idx + len_seg // 2] * w % MOD
                    a[i + j_idx] = (u + v) % MOD
                    a[i + j_idx + len_seg // 2] = (u - v + MOD) % MOD
                    w = w * wlen % MOD
            len_seg <<= 1
            
        if invert:
            n_inv = pow(n_len, MOD - 2, MOD)
            for i in range(n_len):
                a[i] = a[i] * n_inv % MOD

    def multiply_poly(a, b):
        n_len = 1
        while n_len < len(a) + len(b):
            n_len <<= 1
        fa = a[:] + [0] * (n_len - len(a))
        fb = b[:] + [0] * (n_len - len(b))
        
        ntt(fa, False)
        ntt(fb, False)
        for i in range(n_len):
            fa[i] = fa[i] * fb[i] % MOD
        ntt(fa, True)
        return fa

    def poly_inv(a, n_req):
        if n_req == 1:
            return [pow(a[0], MOD - 2, MOD)]
        
        m = (n_req + 1) // 2
        b_prev = poly_inv(a, m)
        
        # 牛顿迭代: B_new = B_old * (2 - A * B_old) mod x^n_req
        
        # 计算 A * B_old
        # 只需要前 n_req 项，但为了 NTT 效率，长度需补齐
        len_ntt = 1
        while len_ntt < n_req + len(b_prev):
            len_ntt <<= 1
            
        a_trunc = a[:n_req] + [0] * (len_ntt - min(n_req, len(a)))
        b_pad = b_prev + [0] * (len_ntt - len(b_prev))
        
        ntt(a_trunc, False)
        ntt(b_pad, False)
        
        for i in range(len_ntt):
            a_trunc[i] = a_trunc[i] * b_pad[i] % MOD
            
        ntt(a_trunc, True)
        
        # 计算 2 - A*B_old
        for i in range(n_req):
            a_trunc[i] = (2 - a_trunc[i]) % MOD
        for i in range(n_req, len_ntt):
            a_trunc[i] = 0
            
        # 计算 B_old * (2 - A*B_old)
        ntt(a_trunc, False)
        
        # 重新变换 B_old
        b_pad = b_prev + [0] * (len_ntt - len(b_prev))
        ntt(b_pad, False)
        
        for i in range(len_ntt):
            a_trunc[i] = a_trunc[i] * b_pad[i] % MOD
            
        ntt(a_trunc, True)
        
        return a_trunc[:n_req]

    G_coeffs = poly_inv(digits, n)
    
    print(" ".join(map(str, G_coeffs)))

if __name__ == '__main__':
    solve()
````
