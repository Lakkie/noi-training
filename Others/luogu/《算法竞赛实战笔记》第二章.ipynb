### P1601 高精度加法

```c
//非负整数字符串
#include<iostream>
#include<string>
using namespace std;
string add(string a,string b){
    string c;//计算结果(反序)
    int lenA=a.size();
    int lenB=b.size();
    int maxLen=0;//取较长的位数
    if(lenA>=lenB){
        maxLen=lenA;
    }else{
        maxLen=lenB;
    }
    int carry=0;//进位
    for(int i=0;i<maxLen;++i){//按较长的位数逐位计算
        int sum=carry;
        if(i<lenA) sum+=a[lenA-1-i]-'0';//获取a的当前位（从后往前取）
        if(i<lenB) sum+=b[lenB-1-i]-'0';//获取b的当前位（从后往前取）
        c+=sum%10+'0';//当前位
        carry=sum/10;//更新进位
    }
    if(carry>0){//最高一位进位
        c+=carry+'0';
    }
    string r;//计算结果(正序)
    for(int i=c.size()-1;i>=0;--i){
        r+=c[i];
    }
    return r;
}
int main(){
    //输入
    string a,b;
    cin>>a>>b;
    //计算
    string s=add(a,b);
    //输出
    cout<<s<<endl;
    return 0;
}
```

```c
//非负整数结构体
#include<iostream>
#include<string>
#include<cstring>
#include<algorithm>//调用max函数
using namespace std;
const int MAXN=505;//定义最大位数
struct BigInt{
    int d[MAXN];//存储每一位数字，d[0]是个位
    int len;//有效位数
    BigInt(){
        memset(d,0,sizeof(d));
        len=1;
    }
    void init(string s){//从字符串初始化
        memset(d,0,sizeof(d));
        len=s.size();
        for(int i=0;i<len;++i){//倒序存储：字符串最后一位(个位)存入d[0]
            d[i]=s[len-1-i]-'0';
        }
        while(len>1&&d[len-1]==0){//处理前导零，例如"00123"->len=3,d=[3,2,1]
            len--;
        }
    }
    void print() const{//输出
        for(int i=len-1;i>=0;--i){
            cout<<d[i];
        }
        cout<<endl;
    }
};
BigInt add(const BigInt& A,const BigInt& B){
    BigInt C;//计算结果
    int carry=0;//进位
    int lenA=A.len;
    int lenB=B.len;
    int maxLen=max(lenA,lenB);//遍历较长的位数
    for(int i=0;i<maxLen;++i){
        int sum=carry;
        if(i<lenA) sum+=A.d[i];//获取a的当前位（从后往前取）
        if(i<lenB) sum+=B.d[i];//获取b的当前位（从后往前取）
        C.d[i]=sum%10;//当前位
        carry=sum/10;//更新进位
    }
    C.len=maxLen;//循环结束后，i即为结果的总位数
    if(carry>0){
		C.len+=1;
		C.d[C.len-1]=carry;
	}
    if(C.len==0) C.len=1;//注意：如果结果是0，len至少为1
    return C;
}
int main(){
    //输入
    string a,b;
    cin>>a>>b;
    //计算
    BigInt A,B,C;
    A.init(a);
    B.init(b);
    C=add(A,B);
    //输出
    C.print();
    return 0;
}
```

```c
//非负整数字符串2
#include<iostream>
#include<string>
#include<algorithm>//调用reverse函数
using namespace std;
string add(string a,string b){
    string result="";
    int i=a.size()-1;//指向a的末尾
    int j=b.size()-1;//指向b的末尾
    int carry=0;//进位
    //当还有数字未处理，或还有进位时继续循环
    while(i>=0||j>=0||carry>0){
        int sum=carry;
        //获取a的当前位，如果索引越界则视为0
        if(i>=0){
            sum+=a[i]-'0';
            i--;
        }
        //获取b的当前位，如果索引越界则视为0
        if(j>=0){
            sum+=b[j]-'0';
            j--;
        }
        //计算新的进位和当前位的结果
        carry=sum/10;
        int digit=sum%10;
        //将当前位字符追加到结果字符串
        result.push_back(digit+'0');
    }
    //因为是从低位到高位追加的，所以结果需要反转
    reverse(result.begin(),result.end());
    return result;
}
int main(){
    //输入
    string a,b;
    cin>>a>>b;
    //计算
    string s=add(a,b);
    //输出
    cout<<s<<endl;
    return 0;
}
```

### P2142 高精度减法

```c
//非负整数字符串
#include<iostream>
#include<string>
using namespace std;
//比较两个字符串数字值的大小
bool compare(string a,string b){
    if(a.size()!=b.size()){
        return a.size()>b.size();
    }
    return a>=b;//如果长度相同，直接字典序比较
}
//计算较大数字减较小数字
string subtract_abs(string larger,string smaller){
    string c="";
    int borrow=0;
    for(int i=0;i<larger.size();++i){
        int digitA=(larger[larger.size()-1-i]-'0');
        int digitB=(i<smaller.size())?(smaller[smaller.size()-1-i]-'0'):0;
        int diff=digitA-digitB-borrow;
        if(diff<0){
            diff+=10;
            borrow=1;
        }else{
            borrow=0;
        }
        c+=diff+'0';
    }
	int f=0;//前导零的个数
	for(int i=c.size()-1;i>=0;i--){
		if(c[i]=='0'){
			f++;
		}else{
			break;//遇到第一个不是0的字符就结束循环。
		}
	}
	if(f<c.size()){
	    return c.substr(0,c.size()-f);	
	}else{
		return "0";//特殊情况：全零处理
	}
}
string subtract(string a,string b){
    bool is_negative=false;
    string larger,smaller,c,result;
    if(compare(a,b)){
        larger=a;
        smaller=b;
        is_negative=false;
    }else{
        larger=b;
        smaller=a;
        is_negative=true;
    } 
    c=subtract_abs(larger,smaller);
    if(is_negative){
        c+='-';
    }
    for(int i=c.size()-1;i>=0;--i){
        result+=c[i];
    }
    return result;
}
int main(){
    //输入
    string a,b;
    cin>>a>>b;
    //计算
    string s=subtract(a,b);
    //输出
    cout<<s<<endl;
    return 0;
}
```

```c
//非负整数结构体
#include<iostream>
#include<string>
#include<cstring>
using namespace std;
const int MAXN=10090;//定义最大位数
struct BigInt{
    int d[MAXN];//存储每一位数字，d[0]是个位
    int len;//有效位数
    BigInt(){
        memset(d,0,sizeof(d));
        len=0;
    }
    void init(string s){//从字符串初始化
        memset(d,0,sizeof(d));
        len=s.size();
        for(int i=0;i<len;++i){//倒序存储：字符串最后一位(个位)存入d[0]
            d[i]=s[len-1-i]-'0';
        }
        while(len>1&&d[len-1]==0){//处理前导零，例如"00123"->len=3,d=[3,2,1]
            len--;
        }
    }
    void print() const{//输出
        for(int i=len-1;i>=0;--i){
            cout<<d[i];
        }
        cout<<endl;
    }
};
BigInt subtract(const BigInt& A,const BigInt& B){
    BigInt C;
    int borrow=0;
    for(int i=0;i<A.len;++i){
        int temp=A.d[i]-borrow;
        if(i<B.len) temp-=B.d[i];
        if(temp<0){
            temp+=10;
            borrow=1;
        }else{
            borrow=0;
        }
        C.d[C.len++]=temp;
    }
    //去除前导零
    while(C.len>1&&C.d[C.len-1]==0){
        C.len--;
    }    
    return C;
}
//辅助函数：比较两个高精度数的大小 (A >= B 返回 true)
bool compare(const BigInt& A,const BigInt& B){
    if(A.len!=B.len) return A.len>B.len;
    for(int i=A.len-1;i>=0;--i){
        if(A.d[i]!=B.d[i]) return A.d[i]>B.d[i];
    }
    return true;//相等也算A>=B
}
int main(){
    //输入
    string a,b;
    cin>>a>>b;
    //计算并输出
    BigInt A,B,C;
    A.init(a);
    B.init(b);
    if(compare(A,B)){
        C=subtract(A,B);
    }else{
        C=subtract(B,A);
        cout<<'-';
    }
    C.print();
    return 0;
}
```
