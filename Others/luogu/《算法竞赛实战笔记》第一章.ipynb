### T96987 表达式括号匹配（3种类型括号）

```c
#include<iostream>
#include<stack>
#include<string>
#include<map>
using namespace std;
int main(){
    //输入
    string s;
    cin>>s;
    //计算
    stack<char> st;
    char openParenthesis='(';
    char openBracket='[';
    char openBrace='{';
    char closeParenthesis=')';
    char closeBracket=']';
    char closeBrace='}';
    map<char,char> matchMap;
    matchMap[closeParenthesis]=openParenthesis;
    matchMap[closeBracket]=openBracket;
    matchMap[closeBrace]=openBrace;
    bool is_valid=true;
    for(char c:s){
        if(c==openParenthesis||c==openBracket||c==openBrace){
            st.push(c);
        }else if(c==closeParenthesis||c==closeBracket||c==closeBrace){
            if(st.empty()){
                is_valid=false;
                break;
            }
            if(st.top()!=matchMap[c]){
                is_valid=false;
                break;
            }
            st.pop();
        }
    }
    //输出
    if(is_valid&&st.empty()){
        cout<<"YES"<<endl;
    } else {
        cout<<"NO"<<endl;
    }
    return 0;
}
```

### U117014 后缀表达式求值简单版

```c
#include<iostream>
#include<stack>
#include<string>
#include<cctype>
using namespace std;
int main(){
    //输入
    string s;
    cin>>s;
    //计算
    stack<int> st;
    for(int i=0;i<s.length();++i){
        char c=s[i];
        if(isdigit(c)){
            st.push(c-'0');
        }else if(c=='+'||c=='-'||c=='*'||c=='/'){
            if(st.size()<2){
                return 0;
            }
            int right=st.top();
            st.pop();
            int left=st.top();
            st.pop();
            int result=0;
            if(c=='+'){
                result=left+right;
            }else if(c=='-'){
                result=left-right;
            }else if(c=='*'){
                result=left*right;
            }else if(c=='/'){
                result=left/right;
            }
            st.push(result);
        }
    }
    //输出
    if(!st.empty()){
        cout<<st.top()<<endl;
    }
    return 0;
}
```

### T110281 中缀表达式求值简单版

```c
#include<iostream>
#include<stack>
using namespace std;
int c(int a,int b,char op){//基础运算
	if(op=='+'){
		return a+b;
	}else if(op=='*'){
		return a*b;
	}
	return 0;
} 
int main(){
	//输入
	string s;
	cin>>s;
	//计算
	int r,l;
	char op;
	stack<int> nums;//存储数字
	stack<char> ops;//存储符号
	for(int i=0;i<s.size();i++){
		if(s[i]>='0'&&s[i]<='9'){
			nums.push(s[i]-'0');
		}else if(s[i]=='('){ 
			ops.push(s[i]);
		}else if(s[i]==')'){//匹配
			while(!ops.empty()&&ops.top()!='('){//括号内先计算
				r=nums.top(); nums.pop();
				l=nums.top(); nums.pop();
				op=ops.top(); ops.pop();
				nums.push(c(l,r,op));
			}
			if(!ops.empty()&&ops.top()=='('){
				ops.pop();//去掉对应的括号
			}
		}else if(s[i]=='+'){
			if(!ops.empty()&&ops.top()!='('){//当前符号是+，上一个是*or+时，都能先计算
				r=nums.top(); nums.pop();
				l=nums.top(); nums.pop();
				op=ops.top(); ops.pop();
				nums.push(c(l,r,op));
			}
			ops.push(s[i]);
		}else if(s[i]=='*'){
			if(!ops.empty()&&ops.top()=='*'){//当前符号是*，上一个也是*时，才能先计算
				r=nums.top(); nums.pop();
				l=nums.top(); nums.pop();
				op=ops.top(); ops.pop();
				nums.push(c(l,r,op));
			}
			ops.push(s[i]);
		} 
	}
	while(!ops.empty()){//剩余的统一运算
		r=nums.top(); nums.pop();
		l=nums.top(); nums.pop();
		op=ops.top(); ops.pop();
		nums.push(c(l,r,op));
	}
	cout<<nums.top()<<endl;
	return 0;
}
```

```c
#include<iostream>
#include<stack>
#include<string>
#include<cctype>
using namespace std;
stack<int> nums;
stack<char> ops;
int getPriority(char op){
    if(op == '+') return 1;
    if(op == '*') return 2;
    return 0;
}
void calculate(){
    if(nums.size()<2||ops.empty()) return;
    int b=nums.top(); nums.pop();
    int a=nums.top(); nums.pop();
    char op=ops.top(); ops.pop();
    int res=0;
    switch(op){
        case '+': res=a+b; break;
        case '*': res=a*b; break;
    }
    nums.push(res);
}
int main(){
    //输入
    string s;
    cin>>s;
    //计算
    for(int i=0;i<s.length();++i){
        char c=s[i];
        if(isdigit(c)){
            nums.push(c-'0');
        }else if(c=='('){
            ops.push(c);
        }else if(c==')'){
            while(!ops.empty()&&ops.top()!='('){
                calculate();
            }
            if(!ops.empty()&&ops.top()=='('){
                ops.pop();
            }
        }else if(c=='+'||c=='*'){
            while(!ops.empty()&&ops.top()!='('&&getPriority(ops.top())>=getPriority(c)){
                calculate();
            }
            ops.push(c);
        }
    }
    while(!ops.empty()){
        calculate();
    }
    //输出
    if(!nums.empty()){
        cout<<nums.top()<<endl;
    }
    return 0;
}
```

### P2058 [NOIP 2016 普及组] 海港

```c
#include<iostream>
#include<vector>
#include<queue>
using namespace std;
struct Passenger{
    int time;
    int nation;
};
int main(){
	//输入
	int n;
	cin>>n;
	queue<Passenger> q;
	vector<int> country_count(100005,0);
	int distinct_countries=0;
	for(int i=1;i<=n;++i){
		int t,k,x;
		cin>>t>>k;
		for(int j=1;j<=k;++j){//统计全部信息
			cin>>x;
			q.push({t,x});
            if (country_count[x]==0){
                distinct_countries++;
            }
            country_count[x]++;
		}
        while(!q.empty()&&t-q.front().time>=86400){//剔除不符合条件的信息
            Passenger front_p=q.front(); q.pop();
            country_count[front_p.nation]--;
            if(country_count[front_p.nation]==0){
                distinct_countries--;
            }
        }
        //输出
        cout<<distinct_countries<<endl;
	}
    return 0;
}
```

```c
//动态规划 / Kadane算法
#include<iostream>
#include<climits>//用于LLONG_MIN
using namespace std;
int main(){
    //输入
    int n;
    cin>>n;
    long long current_sum=0;
    long long max_sum=LLONG_MIN;//初始化为最小值，处理全负数的情况
    for(int i=0;i<n;++i){
        int x;
        cin>>x;
        //计算
        if(current_sum<0){//状态转移：如果之前的累加和小于0，则抛弃之前的段，从当前x开始
            current_sum=x;
        }else{//否则，将x加入当前的段
            current_sum+=x;
        }
        if(current_sum>max_sum){//更新全局最大值
            max_sum=current_sum;
        }
    }
    //输出
    cout<<max_sum<<endl;
    return 0;
}
```

### T111583 max值PK

```c
#include<iostream>
#include<vector>
#include<algorithm>
#include<cmath>
using namespace std;
int main(){
    //输入
    int n;
    cin>>n;
    vector<int> a(n+1);
    for(int i=1;i<=n;++i){
        cin>>a[i];
    }
    //计算
    vector<int> prefix_max(n+1);//前缀最大值
    prefix_max[1]=a[1];
    for(int i=2;i<=n;++i){
        prefix_max[i]=max(prefix_max[i-1],a[i]);
    }
    vector<int> suffix_max(n+1);//后缀最大值
    suffix_max[n]=a[n];
    for(int i=n-1;i>=1;--i){
        suffix_max[i]=max(suffix_max[i+1],a[i]);
    }
    long long max_diff=0;
    for(int m=1;m<n;++m){
        long long left_max=prefix_max[m];
        long long right_max=suffix_max[m+1];
        long long current_diff=abs(left_max-right_max);
        if(current_diff>max_diff){
            max_diff=current_diff;
        }
    }
    //输出
    cout<<max_diff<<endl;
    return 0;
}
```

```c
#include<iostream>
#include<algorithm>
#include<cmath>
using namespace std;
int getDepth(int k){//计算节点k的深度（根节点深度为 1）
    int depth=0;
    while(k>0){
        depth++;
        k/=2;
    }
    return depth;
}
int findLCA(int x,int y){//二叉树最近公共祖先（Lowest Common Ancestor, LCA）
    int depthX=getDepth(x);//获取两个节点的深度
    int depthY=getDepth(y);
    while(depthX>depthY){//将较深的节点向上移动，直到两个节点深度相同
        x/=2;
        depthX--;
    }
    while(depthY>depthX){
        y/=2;
        depthY--;
    }
    while(x!=y){//两个节点同时向上移动，直到相遇
        x/=2;
        y/=2;
    }
    return x;
}
int main(){
    int x,y;
    cin>>x>>y;
	cout<<findLCA(x,y)<<endl;
    return 0;
}
```

### T112246 二叉树基本操作

```c
#include<iostream>
#include<string>
#include<algorithm>
#include<vector>
using namespace std;
struct TreeNode{//定义二叉树节点结构
    char val;
    TreeNode *left;
    TreeNode *right;
    TreeNode(char x):val(x),left(NULL),right(NULL){}
};
int idx=0;//全局变量用于跟踪当前读取到的字符串位置
string preOrderStr;
TreeNode* buildTree(){//根据先序序列建立二叉树
    if(idx>=preOrderStr.length()){
        return NULL;
    }
    char ch=preOrderStr[idx];
    idx++;//移动到下一个字符
    if(ch=='#'){
        return NULL;
    }
    TreeNode* root=new TreeNode(ch);
    root->left=buildTree();//递归建立左子树
    root->right=buildTree();//递归建立右子树
    return root;
}
void inOrderTraversal(TreeNode* root){//中序遍历: 左 -> 根 -> 右
    if(root==NULL) return;
    inOrderTraversal(root->left);
    cout<<root->val;
    inOrderTraversal(root->right);
}
void postOrderTraversal(TreeNode* root){//后序遍历: 左 -> 右 -> 根
    if(root==NULL) return;
    postOrderTraversal(root->left);
    postOrderTraversal(root->right);
    cout<<root->val;
}
int countLeaves(TreeNode* root){//计算叶子节点数
    if(root==NULL) return 0;
    if(root->left==NULL&&root->right==NULL){
        return 1;
    }
    return countLeaves(root->left)+countLeaves(root->right);
}
int getHeight(TreeNode* root){//计算二叉树的高度
    if(root==NULL) return 0;
    int leftHeight=getHeight(root->left);
    int rightHeight=getHeight(root->right);
    return max(leftHeight,rightHeight)+1;
}
void freeTree(TreeNode* root){// 释放内存，防止泄漏
    if(root==NULL) return;
    freeTree(root->left);
    freeTree(root->right);
    delete root;
}
int main() {
    //输入
    cin>>preOrderStr;
    //计算并输出
    idx=0;//初始化索引
    TreeNode* root=buildTree();
    inOrderTraversal(root);
    cout<<endl;
    postOrderTraversal(root);
    cout<<endl;
    cout<<countLeaves(root)<<endl;
    cout<<getHeight(root)<<endl;
    freeTree(root);
    return 0;
}
```

### P1030 [NOIP 2001 普及组] 求先序排列

```c
#include<iostream>
#include<string>
#include<vector>
using namespace std;
string inorder,postorder;
/**
 * 递归构建并输出先序遍历
 * @param inL 中序序列的起始索引
 * @param inR 中序序列的结束索引
 * @param postL 后序序列的起始索引
 * @param postR 后序序列的结束索引
 */
void buildAndPrintPreorder(int inL,int inR,int postL,int postR){
    if(inL>inR||postL>postR){//递归终止条件：区间无效
        return;
    }
    char root=postorder[postR];//后序的最后一个元素是当前子树的根
    cout<<root;
    //在中序序列中找到根节点的位置
    int k=inL;
    while(k<=inR&&inorder[k]!=root){
        k++;
    }
    //计算左子树的节点个数
    int leftSize=k-inL;
    //递归处理左子树
    //左子树中序区间: [inL,k-1]
    //左子树后序区间: [postL,postL+leftSize-1]
    buildAndPrintPreorder(inL,k-1,postL,postL+leftSize-1);
    //递归处理右子树
    //右子树中序区间: [k+1,inR]
    //右子树后序区间: [postL+leftSize,postR-1]
    buildAndPrintPreorder(k+1,inR,postL+leftSize,postR-1);
}
int main(){
    //输入
    cin>>inorder>>postorder;
    //计算并输出
    int n=inorder.length();
    if(n>0){
        buildAndPrintPreorder(0,n-1,0,n-1);
    }
    cout<<endl;
    return 0;
}
```

### P1449 后缀表达式

```c
#include<iostream>
#include<stack>
#include<string>
#include<cctype>//isdigit
using namespace std;
int main(){
    //输入
    string s;
    getline(cin, s);//读取整行表达式
    //计算
    stack<long long> stk;
    long long num=0;
    for(char c:s){
        if(isdigit(c)){//如果是数字字符，累加构建当前数值
            num=num*10+(c-'0');
        }else if(c=='.'){//遇到'.'表示一个数字结束，将构建好的数字入栈
            stk.push(num);
            num=0;
        }else if(c=='+'||c=='-'||c=='*'||c=='/'){//遇到运算符，弹出两个操作数进行计算
            if(stk.size()<2){//理论上题目保证合法，但健壮性检查是个好习惯
                return 0; 
            }
            long long b=stk.top(); stk.pop();//右操作数（后入栈）
            long long a=stk.top(); stk.pop();//左操作数（先入栈）
            long long res=0;
            switch(c){
                case '+': res=a+b; break;
                case '-': res=a-b; break;//注意顺序：左 - 右
                case '*': res=a*b; break;
                case '/': res=a/b; break;//注意顺序：左 / 右，C++整数除法向零取整
            }
            stk.push(res);//将结果压回栈
        }else if(c=='@'){//遇到 '@' 表示表达式结束
            break;
        }
    }
    //输出
    if(!stk.empty()){//最终结果在栈顶
        cout<<stk.top()<<endl;
    }
    return 0;
}
```

```c
#include<iostream>
using namespace std;
//字符转数字：'0'~'9' -> 0~9，'A'~'F' -> 10~15
int charToInt(char c){
    if(c>='0'&&c<='9') return c-'0';
    return c-'A'+10;
}
//数字转字符：0~9 -> '0'~'9'，10~15 -> 'A'~'F'
char intToChar(int x){
    if(x<10) return x+'0';
    return x-10+'A';
}
int main(){
    int n,m;
    string s;
    cin>>n>>s>>m;
    //n进制转十进制
    long long dec=0;
    for(int i=0;i<s.size();i++){
        dec=dec*n+charToInt(s[i]);
    }
    //十进制转m进制
    if(dec==0){//剪枝
        cout<<0<<endl;
        return 0;
    }
    string ans="";
    while(dec>0){//短除法，余数倒序
        ans=intToChar(dec%m)+ans;
        dec/=m;
    }
    cout<<ans<<endl;
    return 0;
}
```

```c
#include<iostream>
#include<stdexcept>//引入标准异常类
#include<climits>//引入LLONG_MAX常量
using namespace std;
int n,m;
//字符转数字：'0'~'9'->0~9，'A'~'Z'->10~35
int charToInt(char c){
    if(c>='0'&&c<='9'){
		return c-'0';
	}else if(c>='A'&&c<='Z'){
		return c-'A'+10;
	}else{
		throw invalid_argument(string("Invalid character in input: ")+c);
	}
}
//数字转字符：0~9->'0'~'9'，10~35->'A'~'Z'
char intToChar(int x){
    if(x<10){
		return x+'0';
	}else if(x>=10&&x<=35){
	    return x-10+'A';	
	}else{
		throw domain_error("Digit out of range for base conversion (0-35): "+to_string(x));
	}
}
//n进制转十进制
long long baseNToDecimal(const string& s){
    long long dec=0;
    for(int i=0;i<s.size();i++){
    	int digit=charToInt(s[i]);
    	//检查数位是否小于进制n
        if(digit>=n){
            throw invalid_argument(string("Digit '")+s[i]+"' is not valid for base "+to_string(n));
        }
        //检查数据是否溢出
        if(dec>(LLONG_MAX-digit)/n){
		    throw overflow_error("Conversion result exceeds long long limit");
		}
        dec=dec*n+digit;
    }
    return dec;
}
//十进制转m进制
string decimalToBaseM(long long dec){
    if(dec==0){
        return "0";
    }
    string ans="";
    while(dec>0){//短除法，余数倒序
        ans=intToChar(dec%m)+ans;
        dec/=m;
    }
    return ans;
}
int main(){
    string s;
    try{
	    cin>>n>>s>>m;
	    long long dec=baseNToDecimal(s);
	    string ans=decimalToBaseM(dec);
	    cout<<ans<<endl;
    }catch(const invalid_argument& e){
        cerr<<"Invalid Argument Error: "<<e.what()<<endl;
        return 1;
    }catch(const domain_error& e){
        cerr<<"Domain Error: "<<e.what()<<endl;
        return 1;
    }catch(const overflow_error& e){
        cerr<<"Overflow Error: "<<e.what()<<endl;
        return 1;
    }catch(const exception& e){
        cerr<<"Unknown Exception: "<<e.what()<<endl;
        return 1;
    }
    return 0;
}
```

### P1996 约瑟夫问题

```c
#include<iostream>
#include<queue>
using namespace std;
int main(){
    int n,m;
    cin>>n>>m;
    queue<int> q;
    for(int i=1;i<=n;i++) q.push(i);
    int cnt=0;
    while(!q.empty()){
        int t=q.front();
        q.pop();
        cnt++;
        if(cnt==m){
            cout<<t<<" ";
            cnt=0;
        }else{
            q.push(t);
        }
    }
    return 0;
}
```

### P1165 日志分析

```c
//双栈
#include<iostream>
#include<stack>
#include<algorithm>
using namespace std;
int main(){
    int n;
    cin>>n;
    stack<int> data_stk;//数据栈
    stack<int> max_stk;//最大值辅助栈
    while(n--){
        int op;
        cin>>op;
        if(op==0){//维护最大值栈：存入当前状态下的最大值
            int x;
            cin>>x;
            data_stk.push(x);
            if(max_stk.empty()){
                max_stk.push(x);
            }else{
                max_stk.push(max(x,max_stk.top()));
            }
        }else if(op==1){//出栈操作，需保持两栈同步
            if(!data_stk.empty()){
                data_stk.pop();
                max_stk.pop();
            }
        }else if(op==2){//查询最大值
            if(max_stk.empty()){
                cout<<0<<endl;
            }else{
                cout<<max_stk.top()<<endl;
            }
        }
    }
    return 0;
}
```

### P1305 新二叉树

```c
#include<iostream>
#include<string>
using namespace std;
int l[30],r[30];//l[i]存储节点i的左孩子下标，r[i]存储右孩子下标，下标范围0-25对应'a'-'z'
char root_char;//记录根节点字符
//前序遍历递归函数，u: 当前节点的下标(0-25)
void preOrder(int u){
    if(u==-1) return;//终止条件：如果下标为-1，表示空节点，直接返回
    cout<<char(u+'a');//访问根节点：将下标还原为字符并输出
    preOrder(l[u]);//递归遍历左子树
    preOrder(r[u]);//递归遍历右子树
}
int main(){
    int n;
    cin>>n;
    for(int i=0;i<n;++i){
        char u,left,right;
        cin>>u>>left>>right;
        if(i==0){//记录根节点
            root_char=u;
        }
        int idx=u-'a';//将字符转换为下标进行处理
        if(left=='*'){//处理左孩子：如果是'*'则标记为-1，否则转换为下标
            l[idx]=-1;
        }else{
            l[idx]=left-'a';
        }
        if(right=='*'){//处理右孩子：如果是'*'则标记为-1，否则转换为下标
            r[idx]=-1;
        }else{
            r[idx]=right-'a';
        }
    }
    preOrder(root_char-'a');//从根节点开始进行前序遍历
    cout<<endl;
    return 0;
}
```

### P1087 [NOIP 2004 普及组] FBI 树

```c
//P1087 [NOIP 2004 普及组] FBI 树
#include<iostream>
#include<string>
#include<cmath>
using namespace std;
string s;
int n;
//递归函数：处理字符串s的下标从l到r的部分
void solve(int l,int r){
    //递归分裂：如果区间长度大于1，则先处理左右子树
    if(l<r){
        int mid=(l+r)/2;
        solve(l,mid);//后序遍历：先左
        solve(mid+1,r);//后序遍历：再右
    }
    //统计当前区间[l,r]中'0'和'1'的个数
    int count0=0;
    int count1=0;
    for(int i=l;i<=r;++i){
        if(s[i]=='0'){
            count0++;
        }else{
            count1++;
        }
    }
    //根据统计结果输出当前节点的类型（后序遍历：最后根）
    if(count0>0&&count1>0){
        cout<<'F';//既有0又有1
    }else if(count0>0){
        cout<<'B';//全0
    }else{
        cout<<'I';//全1
    }
}
int main(){
    //输入
    cin>>n>>s;
    //计算
    solve(0,s.size()-1);//调用递归函数，初始区间为整个字符串，字符串长度为2^n，下标从0到2^n-1
    //输出
	cout<<endl;
    return 0;
}
```
