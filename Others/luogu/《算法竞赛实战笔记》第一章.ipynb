```c
//T110143 清空栈
#include<iostream>
using namespace std;
int main(){
    //输入
    int a[105];
    int i=1;
    int ans=0;
    int t=0;
    while(true){
        cin>>a[i];
        if(a[i]==0){
            t=i-1;
            break;
        }
        ans+=a[i]*i;
        i++;
    }
    //输出
    cout<<ans<<endl;
    for(int j=t;j>=1;j--){
        cout<<a[j];
        if(j>1){
            cout<<" ";
        }
    }
    cout<<endl;
    return 0;
}
```

```c
//T96987 表达式括号匹配（3种类型括号）
#include<iostream>
#include<stack>
#include<string>
#include<map>
using namespace std;
int main(){
    //输入
    string s;
    cin>>s;
    //计算
    stack<char> st;
    char openParenthesis='(';
    char openBracket='[';
    char openBrace='{';
    char closeParenthesis=')';
    char closeBracket=']';
    char closeBrace='}';
    map<char,char> matchMap;
    matchMap[closeParenthesis]=openParenthesis;
    matchMap[closeBracket]=openBracket;
    matchMap[closeBrace]=openBrace;
    bool is_valid=true;
    for(char c:s){
        if(c==openParenthesis||c==openBracket||c==openBrace){
            st.push(c);
        }else if(c==closeParenthesis||c==closeBracket||c==closeBrace){
            if(st.empty()){
                is_valid=false;
                break;
            }
            if(st.top()!=matchMap[c]){
                is_valid=false;
                break;
            }
            st.pop();
        }
    }
    //输出
    if(is_valid&&st.empty()){
        cout<<"YES"<<endl;
    } else {
        cout<<"NO"<<endl;
    }
    return 0;
}
```

```c
//U117014 后缀表达式求值简单版
#include<iostream>
#include<stack>
#include<string>
#include<cctype>
using namespace std;
int main(){
    //输入
    string s;
    cin>>s;
    //计算
    stack<int> st;
    for(int i=0;i<s.length();++i){
        char c=s[i];
        if(isdigit(c)){
            st.push(c-'0');
        }else if(c=='+'||c=='-'||c=='*'||c=='/'){
            if(st.size()<2){
                return 0;
            }
            int right=st.top();
            st.pop();
            int left=st.top();
            st.pop();
            int result=0;
            if(c=='+'){
                result=left+right;
            }else if(c=='-'){
                result=left-right;
            }else if(c=='*'){
                result=left*right;
            }else if(c=='/'){
                result=left/right;
            }
            st.push(result);
        }
    }
    //输出
    if(!st.empty()){
        cout<<st.top()<<endl;
    }
    return 0;
}
```

```c
//T110281 中缀表达式求值简单版
#include<iostream>
#include<stack>
using namespace std;
int c(int a,int b,char op){//基础运算
	if(op=='+'){
		return a+b;
	}else if(op=='*'){
		return a*b;
	}
	return 0;
} 
int main(){
	//输入
	string s;
	cin>>s;
	//计算
	int r,l;
	char op;
	stack<int> nums;//存储数字
	stack<char> ops;//存储符号
	for(int i=0;i<s.size();i++){
		if(s[i]>='0'&&s[i]<='9'){
			nums.push(s[i]-'0');
		}else if(s[i]=='('){ 
			ops.push(s[i]);
		}else if(s[i]==')'){//匹配
			while(!ops.empty()&&ops.top()!='('){//括号内先计算
				r=nums.top(); nums.pop();
				l=nums.top(); nums.pop();
				op=ops.top(); ops.pop();
				nums.push(c(l,r,op));
			}
			if(!ops.empty()&&ops.top()=='('){
				ops.pop();//去掉对应的括号
			}
		}else if(s[i]=='+'){
			if(!ops.empty()&&ops.top()!='('){//当前符号是+，上一个是*or+时，都能先计算
				r=nums.top(); nums.pop();
				l=nums.top(); nums.pop();
				op=ops.top(); ops.pop();
				nums.push(c(l,r,op));
			}
			ops.push(s[i]);
		}else if(s[i]=='*'){
			if(!ops.empty()&&ops.top()=='*'){//当前符号是*，上一个也是*时，才能先计算
				r=nums.top(); nums.pop();
				l=nums.top(); nums.pop();
				op=ops.top(); ops.pop();
				nums.push(c(l,r,op));
			}
			ops.push(s[i]);
		} 
	}
	while(!ops.empty()){//剩余的统一运算
		r=nums.top(); nums.pop();
		l=nums.top(); nums.pop();
		op=ops.top(); ops.pop();
		nums.push(c(l,r,op));
	}
	cout<<nums.top()<<endl;
	return 0;
}
```

```c
//T110281 中缀表达式求值简单版
#include<iostream>
#include<stack>
#include<string>
#include<cctype>
using namespace std;
stack<int> nums;
stack<char> ops;
int getPriority(char op){
    if(op == '+') return 1;
    if(op == '*') return 2;
    return 0;
}
void calculate(){
    if(nums.size()<2||ops.empty()) return;
    int b=nums.top(); nums.pop();
    int a=nums.top(); nums.pop();
    char op=ops.top(); ops.pop();
    int res=0;
    switch(op){
        case '+': res=a+b; break;
        case '*': res=a*b; break;
    }
    nums.push(res);
}
int main(){
    //输入
    string s;
    cin>>s;
    //计算
    for(int i=0;i<s.length();++i){
        char c=s[i];
        if(isdigit(c)){
            nums.push(c-'0');
        }else if(c=='('){
            ops.push(c);
        }else if(c==')'){
            while(!ops.empty()&&ops.top()!='('){
                calculate();
            }
            if(!ops.empty()&&ops.top()=='('){
                ops.pop();
            }
        }else if(c=='+'||c=='*'){
            while(!ops.empty()&&ops.top()!='('&&getPriority(ops.top())>=getPriority(c)){
                calculate();
            }
            ops.push(c);
        }
    }
    while(!ops.empty()){
        calculate();
    }
    //输出
    if(!nums.empty()){
        cout<<nums.top()<<endl;
    }
    return 0;
}
```

```c
//P2058 [NOIP 2016 普及组] 海港 *
#include<iostream>
#include<vector>
#include<queue>
using namespace std;
struct Passenger{
    int time;
    int nation;
};
int main(){
	//输入
	int n;
	cin>>n;
	queue<Passenger> q;
	vector<int> country_count(100005,0);
	int distinct_countries=0;
	for(int i=1;i<=n;++i){
		int t,k,x;
		cin>>t>>k;
		for(int j=1;j<=k;++j){//统计全部信息
			cin>>x;
			q.push({t,x});
            if (country_count[x]==0){
                distinct_countries++;
            }
            country_count[x]++;
		}
        while(!q.empty()&&t-q.front().time>=86400){//剔除不符合条件的信息
            Passenger front_p=q.front(); q.pop();
            country_count[front_p.nation]--;
            if(country_count[front_p.nation]==0){
                distinct_countries--;
            }
        }
        //输出
        cout<<distinct_countries<<endl;
	}
    return 0;
}
```

## 1.3 前缀和

```c
//T111575 子段和
#include<iostream>
#include<vector>
using namespace std;
int main(){
    //输入&计算
    int n;
    cin>>n;
    vector<long long> prefix(n+1,0);
    for(int i=1;i<=n;i++){
        int val;
        cin>>val;
        prefix[i]=prefix[i-1]+val;//构建前缀和：当前前缀和=上一个前缀和+当前元素值
    }
    int t;
    cin>>t;
    while(t--){
        int l,r;
        cin>>l>>r;
        //输出
        cout<<prefix[r]-prefix[l-1]<<endl;
    }
    return 0;
}
```

```c
//P1115 最大子段和 动态规划 / Kadane 算法
#include<iostream>
#include<climits>//用于LLONG_MIN
using namespace std;
int main(){
    //输入
    int n;
    cin>>n;
    long long current_sum=0;
    long long max_sum=LLONG_MIN;//初始化为最小值，处理全负数的情况
    for(int i=0;i<n;++i){
        int x;
        cin>>x;
        //计算
        if(current_sum<0){//状态转移：如果之前的累加和小于0，则抛弃之前的段，从当前x开始
            current_sum=x;
        }else{//否则，将x加入当前的段
            current_sum+=x;
        }
        if(current_sum>max_sum){//更新全局最大值
            max_sum=current_sum;
        }
    }
    //输出
    cout<<max_sum<<endl;
    return 0;
}
```

```c
//T111583 max 值 PK
#include<iostream>
#include<vector>
#include<algorithm>
#include<cmath>
using namespace std;
int main(){
    //输入
    int n;
    cin>>n;
    vector<int> a(n+1);
    for(int i=1;i<=n;++i){
        cin>>a[i];
    }
    //计算
    vector<int> prefix_max(n+1);//前缀最大值
    prefix_max[1]=a[1];
    for(int i=2;i<=n;++i){
        prefix_max[i]=max(prefix_max[i-1],a[i]);
    }
    vector<int> suffix_max(n+1);//后缀最大值
    suffix_max[n]=a[n];
    for(int i=n-1;i>=1;--i){
        suffix_max[i]=max(suffix_max[i+1],a[i]);
    }
    long long max_diff=0;
    for(int m=1;m<n;++m){
        long long left_max=prefix_max[m];
        long long right_max=suffix_max[m+1];
        long long current_diff=abs(left_max-right_max);
        if(current_diff>max_diff){
            max_diff=current_diff;
        }
    }
    //输出
    cout<<max_diff<<endl;
    return 0;
}
```

```c
//T112067 二叉树
#include<iostream>
#include<algorithm>
#include<cmath>
using namespace std;
int getDepth(int k){//计算节点k的深度（根节点深度为 1）
    int depth=0;
    while(k>0){
        depth++;
        k/=2;
    }
    return depth;
}
int findLCA(int x,int y){//二叉树最近公共祖先（Lowest Common Ancestor, LCA）
    int depthX=getDepth(x);//获取两个节点的深度
    int depthY=getDepth(y);
    while(depthX>depthY){//将较深的节点向上移动，直到两个节点深度相同
        x/=2;
        depthX--;
    }
    while(depthY>depthX){
        y/=2;
        depthY--;
    }
    while(x!=y){//两个节点同时向上移动，直到相遇
        x/=2;
        y/=2;
    }
    return x;
}
int main(){
    int x,y;
    cin>>x>>y;
	cout<<findLCA(x,y)<<endl;
    return 0;
}
```

```c
//T112246 二叉树基本操作
#include<iostream>
#include<string>
#include<algorithm>
#include<vector>
using namespace std;
struct TreeNode{//定义二叉树节点结构
    char val;
    TreeNode *left;
    TreeNode *right;
    TreeNode(char x):val(x),left(NULL),right(NULL){}
};
int idx=0;//全局变量用于跟踪当前读取到的字符串位置
string preOrderStr;
TreeNode* buildTree(){//根据先序序列建立二叉树
    if(idx>=preOrderStr.length()){
        return NULL;
    }
    char ch=preOrderStr[idx];
    idx++;//移动到下一个字符
    if(ch=='#'){
        return NULL;
    }
    TreeNode* root=new TreeNode(ch);
    root->left=buildTree();//递归建立左子树
    root->right=buildTree();//递归建立右子树
    return root;
}
void inOrderTraversal(TreeNode* root){//中序遍历: 左 -> 根 -> 右
    if(root==NULL) return;
    inOrderTraversal(root->left);
    cout<<root->val;
    inOrderTraversal(root->right);
}
void postOrderTraversal(TreeNode* root){//后序遍历: 左 -> 右 -> 根
    if(root==NULL) return;
    postOrderTraversal(root->left);
    postOrderTraversal(root->right);
    cout<<root->val;
}
int countLeaves(TreeNode* root){//计算叶子节点数
    if(root==NULL) return 0;
    if(root->left==NULL&&root->right==NULL){
        return 1;
    }
    return countLeaves(root->left)+countLeaves(root->right);
}
int getHeight(TreeNode* root){//计算二叉树的高度
    if(root==NULL) return 0;
    int leftHeight=getHeight(root->left);
    int rightHeight=getHeight(root->right);
    return max(leftHeight,rightHeight)+1;
}
void freeTree(TreeNode* root){// 释放内存，防止泄漏
    if(root==NULL) return;
    freeTree(root->left);
    freeTree(root->right);
    delete root;
}
int main() {
    //输入
    cin>>preOrderStr;
    //计算并输出
    idx=0;//初始化索引
    TreeNode* root=buildTree();
    inOrderTraversal(root);
    cout<<endl;
    postOrderTraversal(root);
    cout<<endl;
    cout<<countLeaves(root)<<endl;
    cout<<getHeight(root)<<endl;
    freeTree(root);
    return 0;
}
```

```c
//B3614 【模板】栈 数组模拟
#include<iostream>
#include<string>
using namespace std;
const int MAXN=1000005;//根据题目数据范围设定足够大的容量
unsigned long long stk[MAXN];
int top_idx=0;//栈顶指针，0表示空栈
int main(){
    //输入
    int t;
    cin>>t;
    while(t--){
        top_idx=0;//每组数据开始前重置栈
        int n;
        cin>>n;
        while(n--){
            string op;
            cin>>op;
            //计算&输出
            if(op=="push"){
                unsigned long long x;
                cin>>x;
                stk[++top_idx]=x;//指针上移并存入元素
            }else if(op=="pop"){
                if(top_idx==0){
                    cout<<"Empty"<<endl;
                }else{
                    top_idx--;//指针下移，逻辑上移除元素
                }
            }else if(op=="query"){
                if(top_idx==0){
                    cout<<"Anguei!"<<endl;
                }else{
                    cout<<stk[top_idx]<<endl;//访问栈顶元素
                }
            }else if(op=="size"){
                cout<<top_idx<<endl;//当前指针位置即为元素个数
            }
        }
    }
    return 0;
}
```

```c
//P1449 后缀表达式
#include<iostream>
#include<stack>
#include<string>
#include<cctype>//isdigit
using namespace std;
int main(){
    //输入
    string s;
    getline(cin, s);//读取整行表达式
    //计算
    stack<long long> stk;
    long long num=0;
    for(char c:s){
        if(isdigit(c)){//如果是数字字符，累加构建当前数值
            num=num*10+(c-'0');
        }else if(c=='.'){//遇到'.'表示一个数字结束，将构建好的数字入栈
            stk.push(num);
            num=0;
        }else if(c=='+'||c=='-'||c=='*'||c=='/'){//遇到运算符，弹出两个操作数进行计算
            if(stk.size()<2){//理论上题目保证合法，但健壮性检查是个好习惯
                return 0; 
            }
            long long b=stk.top(); stk.pop();//右操作数（后入栈）
            long long a=stk.top(); stk.pop();//左操作数（先入栈）
            long long res=0;
            switch(c){
                case '+': res=a+b; break;
                case '-': res=a-b; break;//注意顺序：左 - 右
                case '*': res=a*b; break;
                case '/': res=a/b; break;//注意顺序：左 / 右，C++整数除法向零取整
            }
            stk.push(res);//将结果压回栈
        }else if(c=='@'){//遇到 '@' 表示表达式结束
            break;
        }
    }
    //输出
    if(!stk.empty()){//最终结果在栈顶
        cout<<stk.top()<<endl;
    }
    return 0;
}
```
