# 1 教育基础知识和基本原理

## 1.1 教育及其产生与发展

### 1.1.1 教育的由来与定义

在我国一般认为“教育”一词最早见于《孟子·尽心上》中“得天下英才而教育之，三乐也”。这里的“教育”取“教”和“育”共同具有的词义，即培养。

广义的教育泛指一切有目的地增进人的知识和技能、发展人的智力和体力、影响人的思想观念的活动，包括家庭教育、社会教育和学校教育。它可能是无组织的、零碎的，也可能是有组织的、系统的。

狭义的教育主要指学校教育，是指教育者根据一定的社会要求，有目的、有计划、有组织、有系统地对受教育者的身心施加影响，促使他们朝着期望的方向变化发展的活动。

### 1.1.2 教育的本质与社会属性

教育是一种有目的地培养人的社会活动，这是教育区别于其他事务现象的根本特征，是教育的本质。

“社会活动”体现了教育的社会性。一方面，教育是人类特有的活动，动物界是没有教育的。另一方面，人的先天本能不是教育。

“培养人”强调使人朝着好的方向发展。机械的灌输、错误观念的影响不是教育。

“有目的”区别于无目的、无意识的活动，教育以影响和发展人的身心为直接目标。但影响人的身心发展的活动不都是教育，那些无目的的偶然影响不是教育。
